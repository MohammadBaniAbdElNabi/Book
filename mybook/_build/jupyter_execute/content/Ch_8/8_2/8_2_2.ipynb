# 8.2.2 Generating Random Events

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/drive/1efb2rLy85WEEJkxMzbXI5CqDXs4VC8g0?usp=sharing)

In a discrete event simulation, once an event of any type has been simulated, the most important piece of information we need to know, in order to advance the simulation is: how long until the next event? Once a customer has arrived, we need to know when the next arrival will occur so that we can schedule that event within the simulation. Similarly, upon completion of a service or upon arrival of a customer to an idle server, we need to know the length of time this next service will take so that we can schedule this customer's departure from the system.

If we are assuming that interarrival times and service times come from some particular probability distributions, then we must have a mechanism within the simulation program to generate the lengths of these intervals of time and therefore to generate the next events in the simulated system. The general procedure will be first to generate a random number from the uniform distribution, to apply a mathematical transformation
to the uniform deviate to obtain a random number from the desired distribution, and then to use this random number in the simulation (perhaps as the interval of time until the next event).

A great deal of effort has been put into the study and development of computer programs to generate random numbers. Truly random numbers are typically obtained from some physical process, but sequences of numbers generated in this way are unfortunately not reproducible. Pseudorandom numbers are numbers that satisfy certain statistical tests for randomness but are generated by a systematic algorithmic procedure that can be repeated if desired. The purpose of generating pseudorandom numbers is to simulate sampling from a continuous uniform distribution over the interval $[0,1]$.

The most frequently implemented algorithms belong to the class of congruential generator methods. These generators are fully described in books by Knuth (1981), Graybeal and Pooch (1980), Banks et al. (1984), Marsaglia (2003) and most introductory texts on simulation; and they are almost always available in any computer installation through simple subroutine calls. Because of the easy accessibility of these pseudorandom number generators, it is doubtful that a simulation analyst would need to develop software from scratch for this purpose. Yet, from a practical standpoint, analysts are encouraged to heed the following warning. Because almost every computer system offers at least one means of generating uniform random variates, most computer users employ these capabilities with faith, assume their correctness, and feel happy with the results. Nevertheless, blatantly bad random number generators are prevalent and may fail some of the standard theoretical or empirical statistical tests for randomness, or may generate strings of numbers exhibiting detectable regular patterns (Marsaglia 1985, Park and Miller 1988, Ripley 1988, L'Ecuyer 1990).

Although many simulation models appear to work well despite these defects in the stream of random numbers, there have been simulation studies that yield totally misleading results because they are more sensitive to the quality of the generators. And although such failures are rare, they can be disastrous; therefore, researchers are still actively investigating better ways to generate random numbers.

In any case, it is quite unlikely that a simulation analyst would need to develop his own software for this purpose. Instead we will discuss how to use a uniform deviate from the interval $[0,1]$ to produce a random number from an exponential distribution, thus simulating a sampling from an exponential distribution.

A commonly used method for doing this, called the inverse transform method, can be applied whenever the inverse of the cumulative distribution function of the desired distribution can be computed analytically.

Recall that the probability density function for the exponential distribution is given by:

$$
f(x)= \begin{cases}\lambda e^{-\lambda x} & \text { for } x \geq 0 \\ 0 & \text { for } x<0\end{cases}
$$

The corresponding cumulative distribution function is given by:

$$
F(x)=\int_{-\infty}^{x} f(t) d t= \begin{cases}1-e^{-\lambda x} & \text { for } x \geq 0 \\ 0 & \text { for } x<0\end{cases}
$$

Inverse transform method.

Figure 8.1 illustrates that the range of $\mathrm{F}(\mathrm{x})$ is the interval ( 0,1 ), and suggests that uniform random numbers from ( 0,1 ) can be transformed into exponentially distributed numbers as follows. Let $R$ denote the uniform random number from ( 0,1 ), and set $F(x)=R$. Then, $x=F^{-1}(R)$ and $x$ can be solved in terms of $R$ by evaluating:

$$
\begin{aligned}
\mathrm{F}(\mathrm{x}) & =1-\mathrm{e}^{-\lambda \mathrm{x}}=\mathrm{R} \\
\mathrm{e}^{-\lambda \mathrm{x}} & =1-\mathrm{R} \\
-\lambda \mathrm{x} & =\ln (1-\mathrm{R}) \\
\mathrm{x} & =-\frac{1}{\lambda} \ln (1-\mathrm{R})
\end{aligned}
$$

This formula is called a random variate generator for the exponential distribution. It is often simplified by replacing ( $1-\mathrm{R}$ ) by R , since both R and ( $1-\mathrm{R}$ ) are uniformly distributed on ( 0,1 ), to obtain the generating formula

$$
x=-\frac{1}{\lambda} \ln R
$$

Therefore, whenever a simulation program requires a sample from an exponential distribution, R is obtained from a standard pseudorandom number generator, and x is computed by this formula and used in the simulation.

The inverse transform method is not necessarily the most efficient method, but it is straightforward and can be used to generate deviates from a variety of statistical distributions other than the exponential. Unfortunately, for most distributions (including the normal), the cumulative probability function does not have a closed form inverse. In particular, the distribution may be derived from an empirical study of the actual system. In practice, the distributions may not fit any of the theoretical functions. For example, consider a server who can perform several different types of service depending on the customer's need (doctor, bank teller, mechanic). Each type of service has a non-zero minimum required time plus a random variable time. However, when all types of service are aggregated, the resulting distribution is likely to be multi-modal, and very non-standard.

In these situations, it is common to approximate the cumulative distribution by a piecewise linear function, and then to apply the inverse transform method using linear

Internally, this distribution can be stored in table form, as in Table 8.4. Then, when we want to generate a random variate from this distribution, we select a uniform random number r , then search for the entry $\mathrm{r}_{\mathrm{i}}$ in the table such that $\mathrm{r}_{\mathrm{i}} \leq \mathrm{r}<\mathrm{r}_{\mathrm{i}+1}$. The corresponding service time is obtained using standard linear interpolation:

**TABLE 8.4** <br>
Piecewise Linear Approximation of Cumulative Service Time Distribution

| Inflection Point | x Service Time | F(x) Cumulative Probability |
| :--- | :--- | :--- |
| 1 | $\mathrm{x}_{1}$ | $\mathrm{r}_{1}=0$ |
| 2 | $\mathrm{x}_{2}$ | $\mathrm{r}_{2}$ |
| 3 | $\mathrm{x}_{3}$ | $\mathrm{r}_{3}$ |
| 4 | $\mathrm{x}_{4}$ | $\mathrm{r}_{4}$ |
| 5 | $\mathrm{x}_{5}$ | $\mathrm{r}_{5}$ |
| 6 | $\mathrm{x}_{6}$ | $\mathrm{r}_{6}$ |
| 7 | $\mathrm{x}_{7}$ | $\mathrm{r}_{7}=1.0$ |


$$
\mathrm{x}=\left(\mathrm{r}-\mathrm{r}_{\mathrm{i}}\right) / \text { slope }_{\mathrm{i}}+\mathrm{x}_{\mathrm{i}}
$$

where

$$
\text { slope }_{\mathrm{i}}=\frac{\left(\mathrm{r}_{\mathrm{i}+1}-\mathrm{r}_{\mathrm{i}}\right)}{\left(\mathrm{x}_{\mathrm{i}+1}-\mathrm{x}_{\mathrm{i}}\right)}
$$

Clearly, by using a large number of piecewise linear segments, it is possible to achieve any desired level of accuracy. However, there is a trade-off between accuracy and the time required to search repeatedly through large tables.

For further and more general discussions of this method, see the books by Ravindran et al. (1987), Schmeiser (1980), and Law (2007). These references also contain descriptions of other methods, such as the rejection method, the composition method, a derivation technique, and approximation techniques.

```python
# ── Section 8.2.2 — Interactive Generating Random Events ────────────────

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# Interactive controls
lam_slider = widgets.FloatSlider(
    value=3,
    min=0.1,
    max=10,
    step=0.1,
    description='λ:',
    continuous_update=False
)

N_slider = widgets.IntSlider(
    value=10000,
    min=100,
    max=50000,
    step=100,
    description='N:',
    continuous_update=False
)

R_slider = widgets.FloatSlider(
    value=0.70,
    min=0.01,
    max=0.99,
    step=0.01,
    description='R example:',
    continuous_update=False
)

seed_slider = widgets.IntSlider(
    value=42,
    min=1,
    max=999,
    step=1,
    description='Seed:',
    continuous_update=False
)

out = widgets.Output()


def run_inverse_transform(lam, N, R_ex, seed):
    with out:
        clear_output(wait=True)

        # ── Inverse transform formula ────────────────────────────
        np.random.seed(seed)

        # Step 1: generate R ~ U(0,1)
        R = np.random.uniform(0, 1, N)

        # Step 2: apply inverse transform
        # We use 1-R because F(x)=1-e^(-λx), so x = -ln(1-R)/λ
        x_inv = -(1 / lam) * np.log(1 - R)

        print('=== Section 8.2.2 — Interactive Inverse Transform Method ===')
        print()
        print('Formula:')
        print('  F(x) = 1 - e^(-λx)')
        print('  R = F(x)')
        print('  x = -(1/λ) ln(1 - R)')
        print()
        print(f'λ = {lam}')
        print(f'N = {N}')
        print(f'Random seed = {seed}')
        print()
        print(f'Expected mean = 1/λ = {1/lam:.4f}')
        print(f'Simulated mean = {x_inv.mean():.4f}')

        if abs(x_inv.mean() - 1/lam) < 0.01:
            print('Mean check: close to expected ✓')
        else:
            print('Mean check: not very close; increase N to improve it')

        print()
        print(f'Expected std = 1/λ = {1/lam:.4f}')
        print(f'Simulated std = {x_inv.std():.4f}')

        # ── Figure 8.1 — illustrate the inverse transform ─────────
        fig, axes = plt.subplots(1, 2, figsize=(12, 4))

        x_max = max(3, np.percentile(x_inv, 99.5))
        x_range = np.linspace(0, x_max, 300)

        F_x = 1 - np.exp(-lam * x_range)

        # Example point
        x_ex = -(1 / lam) * np.log(1 - R_ex)

        # Left: CDF with inverse transform illustrated
        axes[0].plot(
            x_range,
            F_x,
            'b-',
            lw=2,
            label=f'F(x) = 1 - e^(-λx), λ={lam}'
        )

        axes[0].axhline(R_ex, color='red', ls='--', lw=1.2, alpha=0.8)
        axes[0].axvline(x_ex, color='red', ls='--', lw=1.2, alpha=0.8)
        axes[0].scatter([x_ex], [R_ex], color='red', s=80, zorder=5)

        axes[0].annotate(
            f'R={R_ex:.2f} → x={x_ex:.3f}',
            (x_ex, R_ex),
            textcoords='offset points',
            xytext=(10, -15),
            fontsize=9
        )

        axes[0].set_xlabel('x')
        axes[0].set_ylabel('F(x)')
        axes[0].set_title('Figure 8.1 — Inverse Transform\nSet F(x)=R, solve for x')
        axes[0].legend(fontsize=9)
        axes[0].grid(True, alpha=0.3)

        # Right: histogram of generated values vs theoretical PDF
        axes[1].hist(
            x_inv,
            bins=50,
            density=True,
            color='steelblue',
            alpha=0.6,
            label='Inverse transform samples'
        )

        axes[1].plot(
            x_range,
            lam * np.exp(-lam * x_range),
            'r-',
            lw=2,
            label=f'Exponential PDF λ={lam}'
        )

        axes[1].set_xlabel('x')
        axes[1].set_ylabel('Density')
        axes[1].set_title(f'Generated samples vs Exponential PDF\nN={N}, λ={lam}')
        axes[1].legend(fontsize=9)
        axes[1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # Show a few generated values
        print()
        print('First 10 generated random events:')
        print('-' * 45)
        for i in range(10):
            print(f'{i+1:2d}. R = {R[i]:.4f}  →  x = {x_inv[i]:.4f}')

        # ── Piecewise linear explanation ──────────────────────────
        print()
        print('Piecewise linear approximation (Figure 8.3):')
        print('  For non-standard distributions, store (x_i, F(x_i)) in a table.')
        print('  Then for random R: find i such that r_i <= R < r_{i+1}.')
        print('  x = (R - r_i) / slope_i + x_i')
        print('  where slope_i = (r_{i+1}-r_i)/(x_{i+1}-x_i)')


def update(change=None):
    run_inverse_transform(
        lam_slider.value,
        N_slider.value,
        R_slider.value,
        seed_slider.value
    )


for slider in [lam_slider, N_slider, R_slider, seed_slider]:
    slider.observe(update, names='value')


display(widgets.VBox([
    widgets.HTML('<h3>Section 8.2.2 — Generating Random Events</h3>'),
    lam_slider,
    N_slider,
    R_slider,
    seed_slider
]))

display(out)

update()
```

```python
# ── Section 8.2.2 — Generating Random Events with NumPy + SimPy ─────────────
# NumPy generates the random exponential times.
# SimPy uses those generated times to schedule arrivals in simulation time.

import sys
import subprocess
import importlib.util

# Install SimPy if needed
if importlib.util.find_spec("simpy") is None:
    print("Installing SimPy...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "simpy"])

import simpy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

lam_slider = widgets.FloatSlider(
    value=3,
    min=0.1,
    max=10,
    step=0.1,
    description='λ:',
    continuous_update=False,
    style={'description_width': '120px'},
    layout=widgets.Layout(width='420px')
)

N_slider = widgets.IntSlider(
    value=10000,
    min=100,
    max=10000,
    step=100,
    description='Events N:',
    continuous_update=False,
    style={'description_width': '120px'},
    layout=widgets.Layout(width='420px')
)

R_slider = widgets.FloatSlider(
    value=0.70,
    min=0.01,
    max=0.99,
    step=0.01,
    description='Example R:',
    continuous_update=False,
    style={'description_width': '120px'},
    layout=widgets.Layout(width='420px')
)

seed_slider = widgets.IntSlider(
    value=42,
    min=1,
    max=999,
    step=1,
    description='Seed:',
    continuous_update=False,
    style={'description_width': '120px'},
    layout=widgets.Layout(width='420px')
)

show_log_checkbox = widgets.Checkbox(
    value=True,
    description='Show SimPy event log'
)

out = widgets.Output()


# -------------------------------------------------------------------
# SimPy process
# -------------------------------------------------------------------

def arrival_process(env, interarrival_times, event_log):
    """
    SimPy process that creates arrivals using generated interarrival times.
    """
    for i, interarrival in enumerate(interarrival_times, start=1):
        yield env.timeout(interarrival)

        event_log.append({
            'Event Number': i,
            'Clock Time': env.now,
            'Interarrival Time': interarrival,
            'Event': f'Arrival {i}'
        })


# -------------------------------------------------------------------
# Main interactive function
# -------------------------------------------------------------------

def run_822_simpy(lam, N, R_ex, seed, show_log):
    with out:
        clear_output(wait=True)

        np.random.seed(seed)

        # ------------------------------------------------------------
        # Step 1: Generate R ~ Uniform(0,1)
        # ------------------------------------------------------------
        R = np.random.uniform(0, 1, N)

        # ------------------------------------------------------------
        # Step 2: Inverse transform for exponential interarrival times
        # F(x) = 1 - e^(-λx)
        # R = F(x)
        # x = -ln(1 - R) / λ
        # ------------------------------------------------------------
        interarrival_times = -np.log(1 - R) / lam

        # Example inverse transform point
        x_ex = -np.log(1 - R_ex) / lam

        # ------------------------------------------------------------
        # Step 3: Use SimPy to schedule the events
        # ------------------------------------------------------------
        env = simpy.Environment()
        event_log = []

        env.process(arrival_process(env, interarrival_times, event_log))
        env.run()

        event_df = pd.DataFrame(event_log)

        # ------------------------------------------------------------
        # Verification
        # ------------------------------------------------------------
        expected_mean = 1 / lam
        simulated_mean = interarrival_times.mean()

        expected_std = 1 / lam
        simulated_std = interarrival_times.std()

        print("=== Section 8.2.2 — Generating Random Events with NumPy + SimPy ===")
        print()
        print("Important idea:")
        print("  NumPy generates the random interarrival times.")
        print("  SimPy uses those times to move the simulation clock and create events.")
        print()
        print("Formula:")
        print("  F(x) = 1 - e^(-λx)")
        print("  R = F(x)")
        print("  x = -ln(1 - R) / λ")
        print()
        print(f"λ = {lam}")
        print(f"N = {N}")
        print(f"Seed = {seed}")
        print()
        print(f"Expected mean = 1/λ = {expected_mean:.4f}")
        print(f"Simulated mean = {simulated_mean:.4f}")

        if abs(simulated_mean - expected_mean) < 0.05:
            print("Mean check: close to expected ✓")
        else:
            print("Mean check: not very close; increase N to improve it")

        print()
        print(f"Expected std = 1/λ = {expected_std:.4f}")
        print(f"Simulated std = {simulated_std:.4f}")

        print()
        print("Example inverse transform:")
        print(f"R = {R_ex:.2f}")
        print(f"x = -ln(1 - {R_ex:.2f}) / {lam:.2f} = {x_ex:.4f}")

        # ------------------------------------------------------------
        # Plot 1: CDF inverse transform
        # ------------------------------------------------------------
        x_max = max(3 / lam, np.percentile(interarrival_times, 99.5))
        x_range = np.linspace(0, x_max, 300)

        F_x = 1 - np.exp(-lam * x_range)
        pdf_x = lam * np.exp(-lam * x_range)

        fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))

        axes[0].plot(
            x_range,
            F_x,
            linewidth=2,
            label=f'F(x) = 1 - e^(-λx), λ={lam}'
        )

        axes[0].axhline(R_ex, color='red', linestyle='--', linewidth=1.2)
        axes[0].axvline(x_ex, color='red', linestyle='--', linewidth=1.2)
        axes[0].scatter([x_ex], [R_ex], color='red', s=80, zorder=5)

        axes[0].annotate(
            f'R={R_ex:.2f} → x={x_ex:.3f}',
            (x_ex, R_ex),
            textcoords='offset points',
            xytext=(10, -15),
            fontsize=9
        )

        axes[0].set_xlabel('x')
        axes[0].set_ylabel('F(x)')
        axes[0].set_title('Inverse Transform Method\nSet F(x)=R, solve for x')
        axes[0].legend(fontsize=9)
        axes[0].grid(True, alpha=0.3)

        # ------------------------------------------------------------
        # Plot 2: Histogram vs theoretical PDF
        # ------------------------------------------------------------
        axes[1].hist(
            interarrival_times,
            bins=30,
            density=True,
            color='steelblue',
            alpha=0.6,
            edgecolor='black',
            label='Generated interarrival times'
        )

        axes[1].plot(
            x_range,
            pdf_x,
            color='red',
            linewidth=2,
            label=f'Exponential PDF λ={lam}'
        )

        axes[1].set_xlabel('Interarrival time')
        axes[1].set_ylabel('Density')
        axes[1].set_title(f'Generated Samples vs Exponential PDF\nN={N}, λ={lam}')
        axes[1].legend(fontsize=9)
        axes[1].grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        # ------------------------------------------------------------
        # Display generated values
        # ------------------------------------------------------------
        print()
        print("First 10 generated events:")
        print("-" * 55)

        for i in range(min(10, N)):
            print(
                f"{i+1:2d}. R = {R[i]:.4f}  "
                f"Interarrival = {interarrival_times[i]:.4f}  "
                f"Clock Time = {event_df.loc[i, 'Clock Time']:.4f}"
            )

        if show_log:
            print()
            print("=== SimPy Event Log ===")
            display(event_df)

        print()
        print("How to explain this:")
        print("  In 8.2.2, we generate random event times.")
        print("  SimPy is not replacing the inverse transform formula.")
        print("  SimPy uses the generated times to schedule events in the simulation clock.")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

def update(change=None):
    run_822_simpy(
        lam_slider.value,
        N_slider.value,
        R_slider.value,
        seed_slider.value,
        show_log_checkbox.value
    )


for widget in [
    lam_slider,
    N_slider,
    R_slider,
    seed_slider,
    show_log_checkbox
]:
    widget.observe(update, names='value')


display(widgets.VBox([
    widgets.HTML('<h3>Section 8.2.2 — Generating Random Events Using NumPy + SimPy</h3>'),
    widgets.HTML('Change λ, N, R, or the seed. NumPy generates the random event times, and SimPy schedules them as arrivals.'),
    lam_slider,
    N_slider,
    R_slider,
    seed_slider,
    show_log_checkbox
]))

display(out)

update()
```
