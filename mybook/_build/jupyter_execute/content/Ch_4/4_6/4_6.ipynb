[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/drive/1OT6GbX4ifAxvixEaTCu0cNZMpGlafjl2?usp=sharing)

One of the most successful approaches to $0-1$ problems has been the introduction of cover inequalities (Crowder et al. 1983). A cover inequality is a specialized type of cutting plane. It defines a constraint that is added to the original problem in the hope that the extreme point solutions will occur at $0-1$ points. After generating as many cover inequality constraints as possible, the reduced problem is solved using a standard branch-and-bound algorithm. This technique was able to dramatically decrease computation time on large,
sparse $0-1$ programming problems, and practical problems with over $10,0000-1$ variables were solved to optimality. Prior to the introduction of this method, problems with $5000-1$ variables were considered very difficult.
As before, the problem is formulated as a standard linear program with the additional restriction that all variables must be either 0 or 1 . The constraints are partitioned into two types. Type I constraints are called Special Ordered Set (SOS) constraints. Type II constraints are simply all of the non-SOS inequalities. The simplest form of SOS constraint is as follows:

$$
\sum_{j \in L} x_{j} \leq 1 \quad \text { for some subset } L \text { of the variables }
$$

In practical problems, we will often find that the vast majority of constraints are SOS. For example, if the variables $\mathrm{x}_{\mathrm{ij}}$ are equal to 1 if resource i is assigned to location j , then we will have a number of SOS constraints which state that each resource can be assigned to at most one location. We may also get SOS equality constraints if resource i must be assigned to exactly one location.
SOS constraints have a very useful property with respect to $0-1$ integer programming. Observe that, when we consider only one constraint (plus the non-negativity constraints on the variables), every extreme point solution occurs at a $0-1$ point. For example, consider a simple system: $\mathrm{x}_{1}+\mathrm{x}_{2}+\mathrm{x}_{3}=1 ; \mathrm{x}_{1}, \mathrm{x}_{2}, \mathrm{x}_{3} \geq 0$. The extreme points occur at (1,0,0), (0,1,0), $(0,0,1)$, and ( $0,0,0$ ). Unfortunately, when several SOS constraints intersect, fractional LP solutions are introduced, but the property of having many $0-1$ extreme points is still very attractive.
In a sense, SOS constraints produce easy problems, while the remaining inequalities are difficult. In general, the vast majority of extreme points using non-SOS constraints will lead to fractional solutions. Cover inequalities can be considered a simple technique for converting an individual non-SOS constraint into a set of equivalent SOS inequalities.
Before we present a precise definition, consider the following simple constraint as an example:

$$
3 x_{1}+4 x_{2}+5 x_{3} \leq 6
$$

Observe that if we consider only $0-1$ solutions, no two of these $x_{\mathrm{j}}$ 's are allowed to have a value equal to 1 . In particular, we can express this as:

$$
\begin{aligned}
& x_{1}+x_{2} \leq 1 \\
& x_{1}+x_{3} \leq 1 \\
& x_{2}+x_{3} \leq 1
\end{aligned}
$$

All of these constraints are cover inequalities; if any two variables are equal to 1 , then the left-hand side will be greater than (or cover) the right-hand side. As an example, if $\mathrm{x}_{1}$ and $x_{2}=1$, then $3 x_{1}+4 x_{2}=7>6$. In fact, we can represent all three of these constraints in one by observing that only one of these $\mathrm{x}_{\mathrm{j}}$ 's can equal 1 in any feasible $0-1$ solution:

$$
x_{1}+x_{2}+x_{3} \leq 1
$$

Here, we can replace the original non-SOS inequality with its cover. As far as any $0-1$ solutions are concerned, the two constraints are equivalent. With respect to the LP solution, however, the cover inequality is much more restrictive. For example, the point $(1,0.75,0)$ is feasible for the LP but infeasible under the cover.
As a more general illustration, consider the inequality:

$$
3 x_{1}+4 x_{2}+5 x_{3}+6 x_{4}+7 x_{5}+9 x_{6} \leq 12
$$

Any subset of $\mathrm{x}_{\mathrm{j}}$ 's that results in a sum greater than 12 can be eliminated by a cover inequality such as $x_{2}+x_{3}+x_{4} \leq 2$ because we cannot have all three of these variables equal to one. (The sum would be at least 15.)
A cover for a single inequality is a subset of variables, the sum of whose (positive) coefficients is greater than (or covers) the right-hand side value, b. A cover assumes that the inequality is in less than or equal ( $\leq$ ) form, and that all of the coefficients are positive (or zero). We can convert a greater than or equal to constraint into less than or equal to form by multiplying through by -1 . We can also represent an equality constraint by two inequalities (one $\geq$ and $a \leq$ ) and then multiply the $\geq$ by -1 . Each of these would be considered separately. If the constraint has a negative coefficient for variable $\mathrm{x}_{\mathrm{j}}$, we can perform a temporary variable substitution of $\mathrm{x}_{\mathrm{j}}=1-\mathrm{x}_{\mathrm{j}}{ }^{\prime}$ to make all coefficients positive.
Suppose, for example, that a problem contains the constraint:

$$
4 x_{1}-5 x_{2}+3 x_{3}-4 x_{4}-7 x_{5}+5 x_{6}=1
$$

We can replace this constraint with two inequalities:

$$
\begin{aligned}
& 4 x_{1}-5 x_{2}+3 x_{3}-4 x_{4}-7 x_{5}+5 x_{6} \leq 1 \\
& 4 x_{1}-5 x_{2}+3 x_{3}-4 x_{4}-7 x_{5}+5 x_{6} \geq 1
\end{aligned}
$$

(We do not really replace the constraint. We simply transform it for the purpose of finding cover inequalities.) The second ( $\geq$ ) constraint can be written as:

$$
-4 x_{1}+5 x_{2}-3 x_{3}+4 x_{4}+7 x_{5}-5 x_{6} \leq-1
$$

Substitute $\mathrm{x}_{1}=1-\mathrm{x}_{1}^{\prime}, \mathrm{x}_{3}=1-\mathrm{x}_{3}^{\prime}$, and $\mathrm{x}_{6}=1-\mathrm{x}_{6}^{\prime}$ to get

$$
4 x_{1}^{\prime}+5 x_{2}+3 x_{3}^{\prime}+4 x_{4}+7 x_{5}+5 x_{6}^{\prime} \leq 11
$$

Similarly, for the first inequality, we get:

$$
4 \mathrm{x}_{1}+5 \mathrm{x}_{2}^{\prime}+3 \mathrm{x}_{3}+4 \mathrm{x}_{4}^{\prime}+7 \mathrm{x}_{5}^{\prime}+5 \mathrm{x}_{6} \leq 17
$$

We can then use each of these independently to construct cover inequalities. The preceding constraint implies (among others) that:

$$
x_{1}+x_{3}+x_{4}^{\prime}+x_{5}^{\prime} \leq 3 \text { (that is, the variables cannot all have the value } 1 \text { ) }
$$

Converting back to original variables, we get:

$$
x_{1}+x_{3}-x_{4}-x_{5} \leq 1
$$

and we could add this new SOS constraint to the original LP, and resolve it.
In general, any non-SOS constraint can be written in the form:

$$
\sum_{j \in K} a_{j} x_{j} \leq b
$$

where K refers to the subset of non-zero coefficients and we can assume that $\mathrm{a}_{\mathrm{j}}>0$. We have deleted the subscript $i$ for the row to simplify the notation.

Let S be any subset of K such that:

$$
\sum_{\mathrm{j} \in \mathrm{~S}} \mathrm{a}_{\mathrm{j}}>\mathrm{b}
$$

The set $S$ defines a cover. $S$ is called a minimal cover if:

$$
\sum_{\mathrm{j} \in \mathrm{~S}} \mathrm{a}_{\mathrm{j}}-\mathrm{a}_{\mathrm{k}}<\mathrm{b} \text { for all } \mathrm{k} \in \mathrm{~S}
$$

that is, every element of S must cover b . In our example, for

$$
4 x_{1}+5 x_{2}^{\prime}+3 x_{3}+4 x_{4}^{\prime}+7 x_{5}^{\prime}+5 x_{6} \leq 17
$$

we could say that the set $S=\{1,2,3,4,5,6\}$ is a cover. The sum of the coefficients is greater than 17. However, there are a number of smaller covers. If we remove $\mathrm{x}_{2}{ }^{\prime}$, the set is still a cover. If we also remove $x_{3}$, the result, $S=\{1,4,5,6\}$ is still a cover. However, if we remove any other element, S is no longer a cover; the sum will not be greater than 17 . This set is called a minimal cover, and the cover inequality is:

$$
x_{1}+x_{4}^{\prime}+x_{5}^{\prime}+x_{6} \leq 3
$$

or, equivalently,

$$
x_{1}-x_{4}-x_{5}+x_{6} \leq 1
$$

If the set S is a cover, then every $0-1$ solution must satisfy the cover inequality:

$$
\sum_{\mathrm{j} \in \mathrm{~S}} \mathrm{x}_{\mathrm{j}} \leq|\mathrm{S}|-1
$$

There is a simple procedure for finding a minimal cover. Begin with $\mathrm{S}=\mathrm{K}$. Pick any index to delete from S such that the remaining indices still form a cover. Repeat until no index can be deleted without making the coefficient sum less than or equal to $b$. By repeating this process several times in a systematic way, we could generate all possible minimal cover inequalities. However, for large practical problems, the number of cover inequalities can be exponential. Therefore, we need a method for efficiently finding a good cover.

Unfortunately, the approach described earlier is not very practical for large problems. Suppose that one of the non-SOS constraints contains 50 variables, and each cover inequality has approximately 25 variables; then the constraint allows only half of the variables to be used in any $0-1$ solution. The number of potential minimal cover
inequalities is $\binom{50}{25} \approx 1.26 \times 10^{14}$. Generating all possible covers is not a very practical strategy, for even if we could generate all covers, we would discover that most of them were unnecessary in the following sense. The original purpose behind constructing these constraints was to force the LP into a $0-1$ extreme point. Most of the covers, although perfectly valid, will have no effect on the current optimal solution to the LP. The preferred approach would be to solve the LP, and then, if the solution contains fractional values, to look for a single cover inequality that makes the current LP solution infeasible.

To illustrate this process, consider the following simple problem:

$$
\begin{array}{ll}
\operatorname{maximize} & \mathrm{z}=12 \mathrm{x}_{1}+13 \mathrm{x}_{2}+11 \mathrm{x}_{3}+10 \mathrm{x}_{4} \\
\text { subject to: } & 12 \mathrm{x}_{1}+13 \mathrm{x}_{2}+12 \mathrm{x}_{3}+11 \mathrm{x}_{4} \leq 29 \\
& \mathrm{x}_{\mathrm{j}}=0 \text { or } 1
\end{array}
$$

Solving this problem as an LP (with constraints $0 \leq x_{j} \leq 1$ ), we find that $x^{*}=(1,1,0.333,0)$, with $z^{*}=28.667$. We want to find a set $S$ such that:

1. The set $S$ forms a cover of the constraint:

$$
\sum_{\mathrm{j} \in S} \mathrm{a}_{\mathrm{j}}>\mathrm{b}
$$

therefore,

$$
\sum_{j \in S} x_{j} \leq|S|-1
$$

2. The current LP solution violates the cover inequality:

$$
\sum_{j \in S} x_{j}^{*}>|S|-1
$$

It is fairly easy to show that, if $x_{j}^{*}=0$, then $j$ will never occur in the set S . Because every $\mathrm{x}_{\mathrm{j}}^{*} \leq 1$, if any of them are zero, the constraint will never violate the cover inequality. It is also easy to prove that, if $\mathrm{x}_{\mathrm{j}}^{*}=1$, then we can always include it in the set S . If the corresponding $j$ is not in a set $S$ that satisfies the aforementioned criteria, then adding $j$ to $S$ will still be a cover. Therefore, in our example, we will include $\mathrm{x}_{1}$ and $\mathrm{x}_{2}$ and ignore $\mathrm{x}_{4}$. The only question is whether to include $\mathrm{x}_{3}$. Observe that when we do not include it, we do not get a cover; but, when we do add it to $S$, we get a cover and the current solution violates the cover inequality, as required:

$$
\mathrm{x}_{1}+\mathrm{x}_{2}+\mathrm{x}_{3} \leq 2
$$

We now add this constraint to the original problem and solve the LP again. If the new solution is fractional, we look for another cover inequality.

We now present a simple algorithm for finding effective cover inequalities. Let $x^{*}$ be the optimal solution to the linear programming problem with $0 \leq \mathrm{x}_{\mathrm{j}} \leq 1$, and suppose that we
want to find a valid cover inequality for one of the non-SOS constraints that will cut off the current LP solution. Consider any non-SOS constraint of the form:

$$
\sum_{\mathrm{j} \in \mathrm{~K}} \mathrm{a}_{\mathrm{j}} \mathrm{x}_{\mathrm{j}} \geq \mathrm{b}+1
$$

(We will repeat this procedure for each of the non-SOS constraints separately.)
Define the elements of S using the $0-1$ variables $\mathrm{s}_{\mathrm{j}}$, where:

$$
s_{\mathrm{j}} \begin{cases}1 & \text { if } \mathrm{j} \in \mathrm{~s} \\ 0 & \text { Otherwise }\end{cases}
$$

We claim that this problem is equivalent to solving the following $0-1$ knapsack problem:

$$
\begin{array}{ll}
\text { minimize } & z=\sum_{j \in K}\left(1-x_{j}^{*}\right) s_{j} \\
\text { subject to } & \sum_{j \in S} a_{j} x_{j} \leq b
\end{array}
$$

The constraint ensures that the solution will be a cover. If the optimal value of z in this problem is less than 1 , then the corresponding cover inequality will make the current LP solution infeasible. For a proof of this claim, refer to the work of Crowder et al. (1983).

In this subproblem, we do not actually require the optimal value of z . It is only necessary to find a z value less than 1 , so we can use a variation of the biggest bang for your buck heuristic, which will be described in the following, to find an approximate solution efficiently. This method may miss a valid cover; but if it does find one, it will be acceptable.
We present a method for finding an approximate solution to the following $0-1$ knapsack problem:

$$
\begin{array}{ll}
\text { maximize } & \mathrm{z}=\sum_{\mathrm{j} \in \mathrm{~S}} \mathrm{t}_{\mathrm{j}} \mathrm{x}_{\mathrm{j}} \\
\text { subject to } & \sum_{\mathrm{j} \in \mathrm{~S}} \mathrm{a}_{\mathrm{j}} \mathrm{x}_{\mathrm{j}} \leq \mathrm{b}
\end{array}
$$

The LP version of the knapsack problem is very easy to solve optimally. The algorithm sorts all of the variables in decreasing order of bang for buck. The cost coefficient $\mathrm{t}_{\mathrm{j}}$ represents the value (bang) that we get from each $\mathrm{x}_{\mathrm{j}}$, while $\mathrm{a}_{\mathrm{j}}$ represents the cost (buck) or weight associated with the limited resource $b$. Process the variables in decreasing order of $\left\{t_{j} / a_{j}\right\}$, and set $\mathrm{x}_{\mathrm{j}}=1$ as long as the constraint is still satisfied. Let k be the index of the first variable that will not fit in the knapsack. Define the amount of space left in the knapsack (the residual) as:

$$
\mathrm{r}=\mathrm{b}-\sum_{\mathrm{j}<\mathrm{k}} \mathrm{a}_{\mathrm{j}}
$$

and set $\mathrm{x}_{\mathrm{k}}$ equal to the fraction just large enough to use all remaining capacity:

$$
x_{k}=\frac{r}{a_{k}}
$$

The rest of the $\mathrm{x}_{\mathrm{j}}$ 's for $\mathrm{j}>\mathrm{k}$ are set to 0 .
This simple one-pass assignment gives the optimal objective function for the LP and has only one possible fractional variable. Let $\mathrm{z}^{*}$ be the objective function value. The optimal value z for the $0-1$ knapsack problem will be less than or equal to $\mathrm{z}^{*}$. If $\mathrm{z}^{*}$ is not integer valued, we can round it down, and use it to approximate the $0-1$ knapsack solution. Thus, we do not actually solve the $0-1$ knapsack problem.

The bang for buck heuristic also gives us a lower bound on the $0-1$ knapsack problem. If we ignore the fractional variable $\mathrm{x}_{\mathrm{k}}$, we have a feasible $0-1$ solution and, therefore, the optimal $0-1$ solution is bounded below by $\mathrm{z}^{*}-\mathrm{t}_{\mathrm{k}} \mathrm{x}_{\mathrm{k}}$ and above by $\mathrm{z}^{*}$. In particular, if the LP has no fractional variable, the solution $\mathrm{z}^{\mathrm{I}}$ must be optimal.

Our situation presents a type of reverse knapsack problem: minimize a cost function and have at least $(\mathrm{b}+1)$ selected for inclusion in the knapsack. We can apply the same bang for buck heuristic; only we select the variable with the smallest ratio first, and keep selecting until the solution is feasible.

Consider the previous example: $12 \mathrm{x}_{1}+13 \mathrm{x}_{2}+12 \mathrm{x}_{3}+11 \mathrm{x}_{4} \leq 29$ and $\mathrm{x}^{*}=(1,1,0.333,0)$. The knapsack problem becomes:

$$
\begin{array}{ll}
\text { minimize } & \mathrm{z}=0 \mathrm{~s}_{1}+0 \mathrm{~s}_{2}+0.667 \mathrm{~s}_{3}+1 \mathrm{~s}_{4} \\
\text { subject to } & 12 \mathrm{~s}_{1}+13 \mathrm{~s}_{2}+12 \mathrm{~s}_{3}+11 \mathrm{~s}_{4} \geq 30
\end{array}
$$

The heuristic solution is: $\mathrm{s}_{1}=\mathrm{s}_{2}=\mathrm{s}_{3}=1$ or $\mathrm{S}=\{1,2,3\}$ with the value of $\mathrm{z}=0.667$, which is less than 1 . Therefore, the corresponding cover inequality, $x_{1}+x_{2}+x_{3} \leq 2$ cuts off the current LP solution, as required.

Cover inequalities are included, as an option, in most of the higher quality commercial packages. These implementations usually develop as many cover inequalities as possible in a preprocessor, and then solve the reduced problem using branch-and-bound or other techniques. Some implementations may use the technique repeatedly, after each iteration of branch-and-bound.

In large practical test problems, Crowder et al. (1983) have discovered that the main advantage of cover inequalities does not rely on getting $0-1$ extreme points. However, the objective function value for the resulting LP is much closer to the final integer optimum. In other words, the cover inequalities appear to be defining very strong cuts into the feasible region. This has a dramatic effect on the branch-and-bound routine because tree nodes will now be fathomed much earlier, and the bounds will tend to be considerably stronger. As mentioned at the outset, it is possible to solve pure $0-1$ problems with up to $10,0000-1$ variables to optimality in a reasonable amount of computer time.

Since then, many other inequalities have been developed and incorporated into commercial software. We are now solving problems with millions of $0-1$ variables routinely. See Johnson et al. (2000) for several additional examples.

**Code Implementation**

```python
# ── Chapter 4.6 — Interactive Cover Inequalities ─────────────────────
# Same notebook idea:
# 1) Solve LP relaxation
# 2) Check cover S
# 3) Add cover inequality if valid
# 4) Solve LP again
# 5) Compare with original 0-1 integer solution

import sys
import subprocess
import importlib.util

# Install Pyomo if needed
if importlib.util.find_spec("pyomo") is None:
    print("Installing pyomo...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pyomo", "--quiet"])

# Install CBC in Colab/Linux if needed
import shutil
if shutil.which("cbc") is None:
    try:
        print("Installing CBC solver...")
        subprocess.run(["apt-get", "install", "-y", "-qq", "coinor-cbc"], check=False)
    except Exception:
        print("CBC could not be installed automatically. If needed, run: !apt-get install -y -qq coinor-cbc")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display as ipython_display, clear_output
from pyomo.environ import *

# -------------------------------------------------------------------
# Default notebook data
# -------------------------------------------------------------------

default_profits = [12, 13, 11, 10]
default_weights = [12, 13, 12, 11]
default_capacity = 29
default_cover = (1, 2, 3)

n = 4

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

style = {'description_width': '160px'}
layout = widgets.Layout(width='460px')

capacity_slider = widgets.IntSlider(
    value=default_capacity,
    min=10,
    max=60,
    step=1,
    description='Capacity:',
    continuous_update=False,
    style=style,
    layout=layout
)

cover_selector = widgets.SelectMultiple(
    options=[1, 2, 3, 4],
    value=default_cover,
    description='Cover S:',
    style={'description_width': '80px'},
    layout=widgets.Layout(width='230px', height='110px')
)

add_cover_checkbox = widgets.Checkbox(
    value=True,
    description='Add cover cut if valid'
)

show_chart_checkbox = widgets.Checkbox(
    value=True,
    description='Show comparison chart'
)

show_editor_checkbox = widgets.Checkbox(
    value=False,
    description='Edit profits/weights'
)

profit_inputs = []
weight_inputs = []

for i in range(n):
    profit_inputs.append(
        widgets.IntText(
            value=default_profits[i],
            description=f'p{i+1}:',
            style={'description_width': '35px'},
            layout=widgets.Layout(width='120px')
        )
    )

    weight_inputs.append(
        widgets.IntText(
            value=default_weights[i],
            description=f'w{i+1}:',
            style={'description_width': '35px'},
            layout=widgets.Layout(width='120px')
        )
    )

update_button = widgets.Button(
    description='Update solution',
    button_style='primary',
    icon='refresh'
)

reset_button = widgets.Button(
    description='Reset textbook data',
    icon='undo'
)

out = widgets.Output()

# -------------------------------------------------------------------
# Helper functions
# -------------------------------------------------------------------

def get_data():
    profits = [float(w.value) for w in profit_inputs]
    weights = [float(w.value) for w in weight_inputs]
    capacity = float(capacity_slider.value)
    cover = list(cover_selector.value)

    return profits, weights, capacity, cover


def solve_model(profits, weights, capacity, binary=False, cover=None, add_cover=False):
    """
    Solves either:
    - LP relaxation: variables continuous from 0 to 1
    - Binary IP: variables are 0 or 1

    If add_cover=True, adds:
        sum(x_i for i in cover) <= |cover| - 1
    """

    n = len(profits)

    model = ConcreteModel()
    model.I = RangeSet(1, n)

    if binary:
        model.x = Var(model.I, within=Binary)
    else:
        model.x = Var(model.I, bounds=(0, 1))

    model.obj = Objective(
        expr=sum(profits[i-1] * model.x[i] for i in model.I),
        sense=maximize
    )

    model.capacity = Constraint(
        expr=sum(weights[i-1] * model.x[i] for i in model.I) <= capacity
    )

    if add_cover and cover is not None and len(cover) > 0:
        model.cover_cut = Constraint(
            expr=sum(model.x[i] for i in cover) <= len(cover) - 1
        )

    solver = SolverFactory("cbc")
    result = solver.solve(model, tee=False)

    solution = [value(model.x[i]) for i in model.I]
    objective = value(model.obj)
    total_weight = sum(weights[i] * solution[i] for i in range(n))

    return solution, objective, total_weight


def cover_info(weights, capacity, cover, x_lp):
    """
    Check whether selected S is a valid cover and whether the LP violates the cover inequality.
    """

    if len(cover) == 0:
        return {
            "cover_weight": 0,
            "rhs": None,
            "lhs_lp": None,
            "is_valid_cover": False,
            "is_violated": False
        }

    cover_weight = sum(weights[i-1] for i in cover)
    rhs = len(cover) - 1
    lhs_lp = sum(x_lp[i-1] for i in cover)

    is_valid_cover = cover_weight > capacity
    is_violated = lhs_lp > rhs + 1e-8

    return {
        "cover_weight": cover_weight,
        "rhs": rhs,
        "lhs_lp": lhs_lp,
        "is_valid_cover": is_valid_cover,
        "is_violated": is_violated
    }


def solution_table(profits, weights, x_lp, x_cut, x_ip):
    rows = []

    for i in range(len(profits)):
        rows.append({
            "Variable": f"x{i+1}",
            "Profit": profits[i],
            "Weight": weights[i],
            "LP relaxation": round(x_lp[i], 4),
            "After cover cut": round(x_cut[i], 4) if x_cut is not None else "-",
            "Binary IP": int(round(x_ip[i]))
        })

    return pd.DataFrame(rows)


def draw_comparison_chart(x_lp, x_cut, x_ip):
    labels = [f"x{i+1}" for i in range(len(x_lp))]
    x = np.arange(len(labels))
    width = 0.25

    plt.figure(figsize=(9, 4.8))

    plt.bar(x - width, x_lp, width, label="LP relaxation")

    if x_cut is not None:
        plt.bar(x, x_cut, width, label="After cover cut")

    plt.bar(x + width, x_ip, width, label="Binary IP")

    plt.title("Variable Values Before and After Cover Inequality", fontsize=13, pad=16)
    plt.xlabel("Decision variable")
    plt.ylabel("Value")
    plt.xticks(x, labels)
    plt.ylim(0, 1.15)
    plt.grid(True, axis='y', alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()


def reset_data(change=None):
    for i in range(n):
        profit_inputs[i].value = default_profits[i]
        weight_inputs[i].value = default_weights[i]

    capacity_slider.value = default_capacity
    cover_selector.value = default_cover
    update()


# -------------------------------------------------------------------
# Main update function
# -------------------------------------------------------------------

def update(change=None):
    with out:
        clear_output(wait=True)

        profits, weights, capacity, cover = get_data()

        print("=== Chapter 4.6 — Interactive Cover Inequalities ===")
        print()
        print("Problem:")
        print("Maximize:")
        print("  z = " + " + ".join([f"{int(profits[i])}x{i+1}" for i in range(n)]))
        print()
        print("Subject to:")
        print("  " + " + ".join([f"{int(weights[i])}x{i+1}" for i in range(n)]) + f" ≤ {int(capacity)}")
        print("  0 ≤ xi ≤ 1 for LP relaxation")
        print("  xi ∈ {0,1} for integer model")
        print()

        data_df = pd.DataFrame({
            "Variable": [f"x{i+1}" for i in range(n)],
            "Profit": profits,
            "Weight": weights
        })

        print("Data:")
        ipython_display(data_df)

        # Step 1: LP relaxation
        x_lp, z_lp, w_lp = solve_model(
            profits,
            weights,
            capacity,
            binary=False,
            cover=None,
            add_cover=False
        )

        print()
        print("=" * 55)
        print("Original LP Relaxation")
        print("=" * 55)

        for i, val in enumerate(x_lp, start=1):
            print(f"x{i} = {val:.4f}")

        print(f"z = {z_lp:.4f}")
        print(f"Total weight used = {w_lp:.4f} / {capacity:.0f}")

        # Step 2: Cover check
        info = cover_info(weights, capacity, cover, x_lp)

        print()
        print("=" * 55)
        print("Cover Inequality Check")
        print("=" * 55)

        print(f"Selected cover S = {set(cover)}")
        print(f"Total weight of cover = {info['cover_weight']:.0f}")
        print(f"Capacity = {capacity:.0f}")

        if info["is_valid_cover"]:
            print("This is a valid cover because its total weight exceeds the capacity.")
        else:
            print("⚠️ This is NOT a valid cover because its total weight does not exceed the capacity.")
            print("So the cover inequality should not be added.")

        if len(cover) > 0:
            cover_expr = " + ".join([f"x{i}" for i in cover])
            print()
            print("Cover inequality:")
            print(f"  {cover_expr} ≤ {info['rhs']}")

            print()
            print("At the LP relaxation solution:")
            print(f"  {cover_expr} = {info['lhs_lp']:.4f}")
            print(f"  Right-hand side = {info['rhs']}")

            if info["is_violated"]:
                print("The current LP solution violates the cover inequality.")
            else:
                print("The current LP solution does not violate the cover inequality.")

        # Step 3: Add cover cut only if valid
        actually_add_cut = (
            add_cover_checkbox.value
            and info["is_valid_cover"]
            and len(cover) > 0
        )

        x_cut, z_cut, w_cut = None, None, None

        if actually_add_cut:
            x_cut, z_cut, w_cut = solve_model(
                profits,
                weights,
                capacity,
                binary=False,
                cover=cover,
                add_cover=True
            )

            print()
            print("=" * 55)
            print("LP Relaxation After Adding the Cover Cut")
            print("=" * 55)

            for i, val in enumerate(x_cut, start=1):
                print(f"x{i} = {val:.4f}")

            print(f"z = {z_cut:.4f}")
            print(f"Total weight used = {w_cut:.4f} / {capacity:.0f}")

        else:
            print()
            print("=" * 55)
            print("LP After Cover Cut")
            print("=" * 55)
            print("Cover cut was not added.")
            print("Reason: either the checkbox is off, the cover is empty, or the selected cover is not valid.")

        # Step 4: Binary IP
        x_ip, z_ip, w_ip = solve_model(
            profits,
            weights,
            capacity,
            binary=True,
            cover=None,
            add_cover=False
        )

        print()
        print("=" * 55)
        print("Integer Programming Solution")
        print("=" * 55)

        for i, val in enumerate(x_ip, start=1):
            print(f"x{i} = {int(round(val))}")

        print(f"Total weight = {w_ip:.0f}")
        print(f"Objective value = {int(round(z_ip))}")

        # Result table
        print()
        print("=== Comparison Table ===")
        ipython_display(solution_table(profits, weights, x_lp, x_cut, x_ip))

        # Chart
        if show_chart_checkbox.value:
            print()
            print("=== Comparison Chart ===")
            draw_comparison_chart(x_lp, x_cut, x_ip)

        print()
        print("Interpretation:")
        print("A cover is a set of items whose total weight exceeds the capacity.")
        print("Therefore, not all items in that cover can be selected together.")
        print("The cover inequality cuts off fractional LP solutions but keeps valid 0-1 integer solutions.")

        # Default verification
        if (
            profits == default_profits and
            weights == default_weights and
            capacity == default_capacity and
            tuple(cover) == default_cover
        ):
            print()
            print("=== Textbook Default Verification ===")
            print("Cover S = {1, 2, 3}")
            print("Total cover weight = 12 + 13 + 12 = 37 > 29")
            print("Cover inequality:")
            print("  x1 + x2 + x3 ≤ 2")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

update_button.on_click(update)
reset_button.on_click(reset_data)

for w in [
    capacity_slider,
    cover_selector,
    add_cover_checkbox,
    show_chart_checkbox
]:
    w.observe(update, names='value')


# -------------------------------------------------------------------
# Display
# -------------------------------------------------------------------

editor_rows = []

for i in range(n):
    editor_rows.append(
        widgets.HBox([
            profit_inputs[i],
            weight_inputs[i]
        ])
    )

editor_box = widgets.VBox([
    widgets.HTML("<h4>Edit profits and weights</h4>"),
    *editor_rows,
    widgets.HBox([update_button, reset_button])
])

def toggle_editor(change=None):
    if show_editor_checkbox.value:
        editor_box.layout.display = ''
    else:
        editor_box.layout.display = 'none'

show_editor_checkbox.observe(toggle_editor, names='value')
editor_box.layout.display = 'none'

ipython_display(widgets.VBox([
    widgets.HTML("<h3>Chapter 4.6 — Interactive Cover Inequalities</h3>"),
    widgets.HTML("This follows the notebook: solve LP relaxation, check cover S, add the cover cut, then compare with the 0-1 integer solution."),

    capacity_slider,
    cover_selector,
    add_cover_checkbox,
    show_chart_checkbox,
    show_editor_checkbox,
    editor_box
]))

ipython_display(out)

update()
```
