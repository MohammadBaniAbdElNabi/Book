If we accept the validity of these axioms, then it is possible to define a preference function or a utility function, $\mathrm{u}(\mathrm{A})$, with the properties that:

1. For any two outcomes, A and $\mathrm{B}, \mathrm{u}(\mathrm{A})>\mathrm{u}(\mathrm{B})$ if and only if outcome A is preferred to outcome B.
2. If an outcome C is indifferent to a lottery L with probability p of outcome A and probability ( $1-\mathrm{p}$ ) of outcome B , then

$$
u(C)=p \times u(A)+(1-p) \times u(B)
$$

That is, we can define a utility function such that the utility of the lottery is equal to the mathematical expectation of the utilities of the prizes.

A utility function that satisfies these properties is invariant under linear scaling. If we add a constant to all utility values, or if me multiply all utilities by a constant, the new function will still satisfy both of the aforementioned properties. Therefore, we can assume any convenient scale for our function. In particular, we will assume that the best possible outcome has a utility, $\mathrm{u}($ Best $)=1$, and the worst possible outcome has a utility, $\mathrm{u}($ Worst $)=0$. Note that we could use any convenient scale (e.g., from 1 to 100, or from 0 to 10).
Consider the decision problem from the previous section which is displayed in Figure 9.8. We wish to repeat the analysis, but this time, we will incorporate the decisionmaker's attitude toward risk using utility theory. Note that the gate associated with paying for the survey has been removed. Instead, the $\$ 20,000$ cost of the survey has been subtracted from the final outcomes for all corresponding tree branches. This does not affect the EMV of the decision, but, in order to evaluate utilities, all outcomes must be expressed as a net effect of that complete branch.

In this decision problem, the best possible outcome is $\$ 1,000,000$ and the worst possible outcome is $-\$ 520,000$. Therefore, we can arbitrarily assign utilities:

$$
u(\$ 1,000,000)=1
$$

and

$$
u(-\$ 520,000)=0
$$

There are two commonly used methods for determining the utilities of the intermediate values. As seen in Figure 9.9, each will give us one new point for the utility function. In Figure 9.9a, the decision-maker is asked, "For what value of $X$ are you indifferent to the alternatives at the decision fork?" Observe that the expected utility of the lottery is:

$$
[(0.5) \times u(\$ 1,000,000)+(0.5) \times u(-\$ 520,000)]=0.5
$$

By the definition of a utility function, the utility of $X$ must be $u(X)=0.5$. Thus, the decisionmaker is essentially saying that the utility of the lottery is equal to the utility of X .
In the approach illustrated in Figure 9.9b, the decision-maker is asked, "For what value of $p$ are you indifferent to the options at the decision fork?" The expected utility of the lottery, in this case, is given by

$$
(p) \times u(\$ 1,000,000)+(1-p) \times u(-\$ 520,000)=p
$$

We conclude that $\mathrm{u}(\$ 200,000)=\mathrm{p}$, again relying on the definition. There are a variety of other assessment techniques, but the two approaches described here are the simplest, and the most common.
Suppose that we decide to use the first method, and our decision-maker selects a value of $\mathrm{X}=-\$ 100,000$. For this person, $\mathrm{u}(-\$ 100,000)=0.5$. This decision-maker is very risk averse. Given a $50-50$ chance of earning $\$ 1,000,000$ or losing $\$ 520,000$, he would prefer not to play. The chance of a loss of \$520,000 is simply too great. In fact, he would prefer a certain loss of up to $\$ 100,000$ to the lottery. Presumably, this decision maker feels that the smaller loss could be absorbed, while the potential large loss would be nearly intolerable. This rather dramatic behavior is not uncommon among actual decision-makers, and we will consider other examples later.

Let us now repeat the utility assessment procedure using the two lotteries shown in Figure 9.10. Observe in Figure 9.10a that, when the decision-maker specifies a value for $X$ at which he is indifferent, the utility of X is equal to the expected utility of the lottery:

$$
\begin{aligned}
\mathrm{u}(\mathrm{X}) & =(0.5) \times \mathrm{u}(-\$ 100,000)+(0.5) \times \mathrm{u}(-\$ 520,000) \\
& =(0.5) \times(0.5)+(0.5) \times(0)=0.25
\end{aligned}
$$

Similarly, in Figure 9.10b, when a value of Y is selected, we find that the $\mathrm{u}(\mathrm{Y})=0.75$.
Suppose that values of $\mathrm{X}=-\$ 350,000$ and $\mathrm{Y}=\$ 250,000$ are selected. We therefore have five sample points for the utility function. By plotting these points, the remaining values can be estimated by drawing a smooth curve through them, as shown in Figure 9.11.
Using this admittedly approximate utility function, we can now answer several lottery questions. For example, suppose he were faced with a lottery having a $50-50$ chance of $\$ 500,000$ or $-\$ 200,000$. From the curve, $\mathrm{u}(\$ 500,000) \approx 0.86$ and $\mathrm{u}(-\$ 200,000) \approx 0.41$. The expected utility of this lottery is $(0.5) \times(0.86)+(0.5) \times(0.41)=0.635$. Again using the utility curve, we find that the $\mathrm{u}(\$ 80,000) \approx 0.635$. Therefore, the lottery is approximately equivalent to a certain outcome of $\$ 80,000$. Hence, our decision-maker should be indifferent to a certain outcome of $\$ 80,000$ or a $50-50$ lottery of either $\$ 500,000$ or $-\$ 200,000$.

This difference is very small, especially when we consider that the numbers are approximate. The decision-maker should probably consider these two options as being of equal value. Perhaps other criteria could be used to resolve the tie.
When we based our decision on expected monetary value in the previous section, the survey strategy produced an expected profit of $\$ 128,000$. If we decided to market the product without a survey, the expected profit was $\$ 80,000$. Both alternatives were clearly preferred to Do Not Market. When we consider the decision-maker's attitude toward risk, the survey strategy is only marginally preferred, while the alternative of marketing without a survey is definitely dominated by all other options.

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from IPython.display import display

style = {'description_width': '180px'}
layout = widgets.Layout(width='500px')

# ── Default probabilities ────────────────────────────────────────────────
s_pHi = widgets.FloatSlider(value=0.20, min=0, max=0.90, step=0.01,
    description='P(High demand)', style=style, layout=layout, readout_format='.0%')

s_pMed = widgets.FloatSlider(value=0.40, min=0, max=0.90, step=0.01,
    description='P(Med demand)', style=style, layout=layout, readout_format='.0%')

# Survey likelihoods
s_lHi = widgets.FloatSlider(value=0.70, min=0.01, max=0.99, step=0.01,
    description='P(Fav | High)', style=style, layout=layout, readout_format='.0%')

s_lMed = widgets.FloatSlider(value=0.60, min=0.01, max=0.99, step=0.01,
    description='P(Fav | Med)', style=style, layout=layout, readout_format='.0%')

s_lLow = widgets.FloatSlider(value=0.30, min=0.01, max=0.99, step=0.01,
    description='P(Fav | Low)', style=style, layout=layout, readout_format='.0%')

# No survey payoffs
s_payHi_NS = widgets.FloatSlider(value=1000000, min=-1000000, max=1500000, step=10000,
    description='No Survey: High', style=style, layout=layout)

s_payMed_NS = widgets.FloatSlider(value=200000, min=-1000000, max=1500000, step=10000,
    description='No Survey: Med', style=style, layout=layout)

s_payLow_NS = widgets.FloatSlider(value=-500000, min=-1000000, max=1500000, step=10000,
    description='No Survey: Low', style=style, layout=layout)

s_dont_NS = widgets.FloatSlider(value=0, min=-1000000, max=1500000, step=10000,
    description="No Survey: Don’t Market", style=style, layout=layout)

# Survey payoffs
s_payHi_S = widgets.FloatSlider(value=980000, min=-1000000, max=1500000, step=10000,
    description='Survey: High', style=style, layout=layout)

s_payMed_S = widgets.FloatSlider(value=180000, min=-1000000, max=1500000, step=10000,
    description='Survey: Med', style=style, layout=layout)

s_payLow_S = widgets.FloatSlider(value=-520000, min=-1000000, max=1500000, step=10000,
    description='Survey: Low', style=style, layout=layout)

s_dont_S = widgets.FloatSlider(value=-20000, min=-1000000, max=1500000, step=10000,
    description="Survey: Don’t Market", style=style, layout=layout)

col1 = widgets.VBox([
    widgets.HTML('<b>Prior Probabilities</b>'),
    s_pHi, s_pMed,
    widgets.HTML('<b>No Survey Payoffs</b>'),
    s_payHi_NS, s_payMed_NS, s_payLow_NS, s_dont_NS
])

col2 = widgets.VBox([
    widgets.HTML('<b>Survey Payoffs</b>'),
    s_payHi_S, s_payMed_S, s_payLow_S, s_dont_S,
    widgets.HTML('<b>Survey Accuracy</b>'),
    s_lHi, s_lMed, s_lLow
])

controls = widgets.HBox([col1, widgets.HTML('&nbsp;&nbsp;&nbsp;'), col2])
out = widgets.Output()


def draw_money(pHi, pMed,
               payHi_NS, payMed_NS, payLow_NS, dont_NS,
               payHi_S, payMed_S, payLow_S, dont_S,
               lHi, lMed, lLow):

    pLow = max(0, 1 - pHi - pMed)

    # No survey EMV
    emv_market_NS = pHi*payHi_NS + pMed*payMed_NS + pLow*payLow_NS
    best_noSurvey = max(emv_market_NS, dont_NS)
    act_noSurvey = 'Market' if emv_market_NS >= dont_NS else "Don’t Market"

    # Survey probabilities
    pF = pHi*lHi + pMed*lMed + pLow*lLow
    pUNF = 1 - pF

    # Posterior probabilities
    pF_Hi = pHi*lHi/pF
    pF_Med = pMed*lMed/pF
    pF_Low = pLow*lLow/pF

    pU_Hi = pHi*(1-lHi)/pUNF
    pU_Med = pMed*(1-lMed)/pUNF
    pU_Low = pLow*(1-lLow)/pUNF

    # Survey EMV
    emv_F = pF_Hi*payHi_S + pF_Med*payMed_S + pF_Low*payLow_S
    emv_UNF = pU_Hi*payHi_S + pU_Med*payMed_S + pU_Low*payLow_S

    best_F = max(emv_F, dont_S)
    best_UNF = max(emv_UNF, dont_S)

    act_F = 'Market' if emv_F >= dont_S else "Don’t Market"
    act_UNF = 'Market' if emv_UNF >= dont_S else "Don’t Market"

    emv_survey = pF*best_F + pUNF*best_UNF
    do_survey = emv_survey >= best_noSurvey

    def money(v):
        sign = '-' if v < 0 else ''
        return f"{sign}${abs(v):,.0f}"

    def fmtp(v):
        return f"{v:.2f}"

    with out:
        out.clear_output(wait=True)

        fig, ax = plt.subplots(figsize=(15, 10))
        ax.set_xlim(0, 15)
        ax.set_ylim(0, 10)
        ax.axis('off')

        OPT = '#0F6E56'
        DARK = '#2C2C2A'
        MUTED = '#888780'

        def edge(x1, y1, x2, y2, label='', opt=False):
            ax.plot([x1, x2], [y1, y2], color=OPT if opt else MUTED, lw=2.5 if opt else 1.1)
            if label:
                ax.text((x1+x2)/2, (y1+y2)/2 + 0.15, label,
                        ha='center', fontsize=9,
                        bbox=dict(facecolor='white', edgecolor='none', alpha=0.85))

        def dec(x, y, opt=False):
            ax.add_patch(plt.Rectangle((x-0.16, y-0.16), 0.32, 0.32,
                                       facecolor=OPT if opt else 'white',
                                       edgecolor=DARK, lw=1.1, zorder=5))

        def chance(x, y, opt=False):
            ax.add_patch(plt.Circle((x, y), 0.18,
                                    facecolor='white',
                                    edgecolor=OPT if opt else DARK,
                                    lw=1.3, zorder=5))

        def label(x, y, text):
            ax.text(x, y, text, fontsize=9, va='center')

        def node_value(x, y, value):
            ax.text(x-0.15, y+0.35, money(value), fontsize=8, ha='center')

        # Layout
        x0, x1, x2, x3, x4 = 1.2, 3.3, 5.7, 7.7, 10.0
        y0 = 4.7

        ySurvey = 7.0
        yNoSurvey = 2.4

        yF = 8.35
        yUNF = 5.75

        yMktF = 8.9
        yDontF = 7.55

        yMktU = 6.25
        yDontU = 4.9

        yHiF, yMedF, yLowF = 9.55, 8.9, 8.25
        yHiU, yMedU, yLowU = 6.9, 6.25, 5.6

        yMktNS = 3.15
        yDontNS = 1.7

        yHiNS, yMedNS, yLowNS = 3.85, 3.15, 2.45

        # Root
        dec(x0, y0, True)
        node_value(x0, y0, emv_survey if do_survey else best_noSurvey)

        # Survey
        edge(x0, y0, x1, ySurvey, 'Survey', do_survey)
        chance(x1, ySurvey, do_survey)
        node_value(x1, ySurvey, emv_survey)

        edge(x1, ySurvey, x2, yF, f'F ({fmtp(pF)})', do_survey)
        dec(x2, yF)
        node_value(x2, yF, best_F)

        edge(x2, yF, x3, yMktF, 'Market', do_survey and act_F == 'Market')
        chance(x3, yMktF)
        node_value(x3, yMktF, emv_F)

        edge(x3, yMktF, x4, yHiF, f'Hi ({fmtp(pF_Hi)})')
        label(x4, yHiF, money(payHi_S))

        edge(x3, yMktF, x4, yMedF, f'Med ({fmtp(pF_Med)})')
        label(x4, yMedF, money(payMed_S))

        edge(x3, yMktF, x4, yLowF, f'Low ({fmtp(pF_Low)})')
        label(x4, yLowF, money(payLow_S))

        edge(x2, yF, x3, yDontF, "Don’t market", do_survey and act_F == "Don’t Market")
        label(x3, yDontF, money(dont_S))

        # Unfavorable
        edge(x1, ySurvey, x2, yUNF, f'UNF ({fmtp(pUNF)})')
        dec(x2, yUNF)
        node_value(x2, yUNF, best_UNF)

        edge(x2, yUNF, x3, yMktU, 'Market', do_survey and act_UNF == 'Market')
        chance(x3, yMktU)
        node_value(x3, yMktU, emv_UNF)

        edge(x3, yMktU, x4, yHiU, f'Hi ({fmtp(pU_Hi)})')
        label(x4, yHiU, money(payHi_S))

        edge(x3, yMktU, x4, yMedU, f'Med ({fmtp(pU_Med)})')
        label(x4, yMedU, money(payMed_S))

        edge(x3, yMktU, x4, yLowU, f'Low ({fmtp(pU_Low)})')
        label(x4, yLowU, money(payLow_S))

        edge(x2, yUNF, x3, yDontU, "Don’t market", do_survey and act_UNF == "Don’t Market")
        label(x3, yDontU, money(dont_S))

        # No Survey
        edge(x0, y0, x1, yNoSurvey, "Don’t survey", not do_survey)
        dec(x1, yNoSurvey, not do_survey)
        node_value(x1, yNoSurvey, best_noSurvey)

        edge(x1, yNoSurvey, x2, yMktNS, 'Market', not do_survey and act_noSurvey == 'Market')
        chance(x2, yMktNS)
        node_value(x2, yMktNS, emv_market_NS)

        edge(x2, yMktNS, x3, yHiNS, f'Hi ({fmtp(pHi)})')
        label(x3, yHiNS, money(payHi_NS))

        edge(x2, yMktNS, x3, yMedNS, f'Med ({fmtp(pMed)})')
        label(x3, yMedNS, money(payMed_NS))

        edge(x2, yMktNS, x3, yLowNS, f'Low ({fmtp(pLow)})')
        label(x3, yLowNS, money(payLow_NS))

        edge(x1, yNoSurvey, x2, yDontNS, "Don’t market", not do_survey and act_noSurvey == "Don’t Market")
        label(x2, yDontNS, money(dont_NS))

        title = (
            f"Figure 9.8 — Marketing Decision Tree with Survey Information\n"
            f"Optimal: {'Survey' if do_survey else 'Don’t Survey'} | "
            f"Expected Value = {money(emv_survey if do_survey else best_noSurvey)}"
        )
        fig.suptitle(title, fontsize=11)

        plt.tight_layout()
        plt.show()

        print("Key results:")
        print(f"No survey market EMV = {money(emv_market_NS)}")
        print(f"Favorable market EMV = {money(emv_F)}")
        print(f"Unfavorable market EMV = {money(emv_UNF)}")
        print(f"Survey EMV = {money(emv_survey)}")


def update_money(change):
    draw_money(
        s_pHi.value, s_pMed.value,
        s_payHi_NS.value, s_payMed_NS.value, s_payLow_NS.value, s_dont_NS.value,
        s_payHi_S.value, s_payMed_S.value, s_payLow_S.value, s_dont_S.value,
        s_lHi.value, s_lMed.value, s_lLow.value
    )


for s in [
    s_pHi, s_pMed,
    s_payHi_NS, s_payMed_NS, s_payLow_NS, s_dont_NS,
    s_payHi_S, s_payMed_S, s_payLow_S, s_dont_S,
    s_lHi, s_lMed, s_lLow
]:
    s.observe(update_money, names='value')


display(controls, out)

draw_money(
    s_pHi.value, s_pMed.value,
    s_payHi_NS.value, s_payMed_NS.value, s_payLow_NS.value, s_dont_NS.value,
    s_payHi_S.value, s_payMed_S.value, s_payLow_S.value, s_dont_S.value,
    s_lHi.value, s_lMed.value, s_lLow.value
)
```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
from IPython.display import display
from scipy.interpolate import PchipInterpolator

style = {'description_width': '220px'}
layout = widgets.Layout(width='520px')

# ── Fixed best and worst outcomes from the book ───────────────────────────
worst_payoff = -520000
best_payoff = 1000000

# ── Utility assessment points ─────────────────────────────────────────────
# These default values are chosen to match the book example approximately.
# You can change them interactively to show different risk attitudes.

s_u_neg500 = widgets.FloatSlider(
    value=0.045, min=0, max=1, step=0.001,
    description='U(-$500,000)', style=style, layout=layout, readout_format='.3f'
)

s_u_neg100 = widgets.FloatSlider(
    value=0.500, min=0, max=1, step=0.001,
    description='U(-$100,000)', style=style, layout=layout, readout_format='.3f'
)

s_u_neg20 = widgets.FloatSlider(
    value=0.575, min=0, max=1, step=0.001,
    description='U(-$20,000)', style=style, layout=layout, readout_format='.3f'
)

s_u_180 = widgets.FloatSlider(
    value=0.705, min=0, max=1, step=0.001,
    description='U($180,000)', style=style, layout=layout, readout_format='.3f'
)

s_u_200 = widgets.FloatSlider(
    value=0.720, min=0, max=1, step=0.001,
    description='U($200,000)', style=style, layout=layout, readout_format='.3f'
)

s_u_980 = widgets.FloatSlider(
    value=0.997, min=0, max=1, step=0.001,
    description='U($980,000)', style=style, layout=layout, readout_format='.3f'
)

out = widgets.Output()


def money(v):
    sign = '-' if v < 0 else ''
    return f"{sign}${abs(v):,.0f}"


def draw_utility_curve(u_neg500, u_neg100, u_neg20, u_180, u_200, u_980):

    # Payoff points from the book example
    x_points = np.array([
        -520000,
        -500000,
        -100000,
        -20000,
        180000,
        200000,
        980000,
        1000000
    ])

    y_points = np.array([
        0.000,
        u_neg500,
        u_neg100,
        u_neg20,
        u_180,
        u_200,
        u_980,
        1.000
    ])

    # Sort points just in case
    order = np.argsort(x_points)
    x_points = x_points[order]
    y_points = y_points[order]

    # Smooth monotonic interpolation
    utility_function = PchipInterpolator(x_points, y_points)

    x_smooth = np.linspace(worst_payoff, best_payoff, 400)
    y_smooth = utility_function(x_smooth)

    # Values needed for the two trees
    payoff_values = [
        -520000,
        -500000,
        -20000,
        0,
        180000,
        200000,
        980000,
        1000000
    ]

    with out:
        out.clear_output(wait=True)

        fig, ax = plt.subplots(figsize=(9, 6))

        ax.plot(x_smooth, y_smooth, linewidth=2)
        ax.scatter(x_points, y_points, s=60)

        for x, y in zip(x_points, y_points):
            ax.text(x, y + 0.025, f"{money(x)}\nU={y:.3f}",
                    ha='center', fontsize=8)

        ax.set_title("Interactive Utility Function\nConverting Payoffs into Utilities")
        ax.set_xlabel("Payoff ($)")
        ax.set_ylabel("Utility")
        ax.set_ylim(-0.05, 1.08)
        ax.grid(True, alpha=0.3)

        plt.tight_layout()
        plt.show()

        print("Payoff-to-Utility Conversion Table")
        print("-" * 42)

        for p in payoff_values:
            u = float(utility_function(p))
            print(f"{money(p):>14}  →  Utility = {u:.3f}")

        print("\nThese are the numbers used in Figure 9.12:")
        print(f"{money(1000000)}  →  1.000")
        print(f"{money(980000)}  →  {float(utility_function(980000)):.3f}")
        print(f"{money(200000)}  →  {float(utility_function(200000)):.3f}")
        print(f"{money(180000)}  →  {float(utility_function(180000)):.3f}")
        print(f"{money(-500000)}  →  {float(utility_function(-500000)):.3f}")
        print(f"{money(-520000)}  →  0.000")


def update(change):
    draw_utility_curve(
        s_u_neg500.value,
        s_u_neg100.value,
        s_u_neg20.value,
        s_u_180.value,
        s_u_200.value,
        s_u_980.value
    )


for s in [s_u_neg500, s_u_neg100, s_u_neg20, s_u_180, s_u_200, s_u_980]:
    s.observe(update, names='value')


controls = widgets.VBox([
    widgets.HTML("<h3>Step 2: Utility Assessment</h3>"),
    widgets.HTML("""
    <p>
    The first tree uses <b>dollar payoffs</b>.
    The second tree uses <b>utilities</b>.
    This part explains how each dollar amount is converted into a utility value between 0 and 1.
    </p>
    """),
    s_u_neg500,
    s_u_neg100,
    s_u_neg20,
    s_u_180,
    s_u_200,
    s_u_980
])

display(controls, out)

draw_utility_curve(
    s_u_neg500.value,
    s_u_neg100.value,
    s_u_neg20.value,
    s_u_180.value,
    s_u_200.value,
    s_u_980.value
)
```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
from IPython.display import display

style = {'description_width': '180px'}
layout = widgets.Layout(width='500px')

# ── Default probabilities ────────────────────────────────────────────────
s_pHi = widgets.FloatSlider(value=0.20, min=0, max=0.90, step=0.01,
    description='P(High demand)', style=style, layout=layout, readout_format='.0%')

s_pMed = widgets.FloatSlider(value=0.40, min=0, max=0.90, step=0.01,
    description='P(Med demand)', style=style, layout=layout, readout_format='.0%')

# Survey likelihoods
s_lHi = widgets.FloatSlider(value=0.70, min=0.01, max=0.99, step=0.01,
    description='P(Fav | High)', style=style, layout=layout, readout_format='.0%')

s_lMed = widgets.FloatSlider(value=0.60, min=0.01, max=0.99, step=0.01,
    description='P(Fav | Med)', style=style, layout=layout, readout_format='.0%')

s_lLow = widgets.FloatSlider(value=0.30, min=0.01, max=0.99, step=0.01,
    description='P(Fav | Low)', style=style, layout=layout, readout_format='.0%')

# No survey utilities
s_uHi_NS = widgets.FloatSlider(value=1.000, min=0, max=1, step=0.001,
    description='No Survey: U(High)', style=style, layout=layout, readout_format='.3f')

s_uMed_NS = widgets.FloatSlider(value=0.720, min=0, max=1, step=0.001,
    description='No Survey: U(Med)', style=style, layout=layout, readout_format='.3f')

s_uLow_NS = widgets.FloatSlider(value=0.045, min=0, max=1, step=0.001,
    description='No Survey: U(Low)', style=style, layout=layout, readout_format='.3f')

s_dont_NS = widgets.FloatSlider(value=0.585, min=0, max=1, step=0.001,
    description="No Survey: Don’t Market", style=style, layout=layout, readout_format='.3f')

# Survey utilities
s_uHi_S = widgets.FloatSlider(value=0.997, min=0, max=1, step=0.001,
    description='Survey: U(High)', style=style, layout=layout, readout_format='.3f')

s_uMed_S = widgets.FloatSlider(value=0.705, min=0, max=1, step=0.001,
    description='Survey: U(Med)', style=style, layout=layout, readout_format='.3f')

s_uLow_S = widgets.FloatSlider(value=0.000, min=0, max=1, step=0.001,
    description='Survey: U(Low)', style=style, layout=layout, readout_format='.3f')

s_dont_S = widgets.FloatSlider(value=0.575, min=0, max=1, step=0.001,
    description="Survey: Don’t Market", style=style, layout=layout, readout_format='.3f')

col1 = widgets.VBox([
    widgets.HTML('<b>Prior Probabilities</b>'),
    s_pHi, s_pMed,
    widgets.HTML('<b>No Survey Utilities</b>'),
    s_uHi_NS, s_uMed_NS, s_uLow_NS, s_dont_NS
])

col2 = widgets.VBox([
    widgets.HTML('<b>Survey Utilities</b>'),
    s_uHi_S, s_uMed_S, s_uLow_S, s_dont_S,
    widgets.HTML('<b>Survey Accuracy</b>'),
    s_lHi, s_lMed, s_lLow
])

controls = widgets.HBox([col1, widgets.HTML('&nbsp;&nbsp;&nbsp;'), col2])
out = widgets.Output()


def draw_utility(pHi, pMed,
                 uHi_NS, uMed_NS, uLow_NS, dont_NS,
                 uHi_S, uMed_S, uLow_S, dont_S,
                 lHi, lMed, lLow):

    pLow = max(0, 1 - pHi - pMed)

    # No survey expected utility
    eu_market_NS = pHi*uHi_NS + pMed*uMed_NS + pLow*uLow_NS
    best_noSurvey = max(eu_market_NS, dont_NS)
    act_noSurvey = 'Market' if eu_market_NS >= dont_NS else "Don’t Market"

    # Survey probabilities
    pF = pHi*lHi + pMed*lMed + pLow*lLow
    pUNF = 1 - pF

    # Posterior probabilities
    pF_Hi = pHi*lHi/pF
    pF_Med = pMed*lMed/pF
    pF_Low = pLow*lLow/pF

    pU_Hi = pHi*(1-lHi)/pUNF
    pU_Med = pMed*(1-lMed)/pUNF
    pU_Low = pLow*(1-lLow)/pUNF

    # Survey expected utilities
    eu_F = pF_Hi*uHi_S + pF_Med*uMed_S + pF_Low*uLow_S
    eu_UNF = pU_Hi*uHi_S + pU_Med*uMed_S + pU_Low*uLow_S

    best_F = max(eu_F, dont_S)
    best_UNF = max(eu_UNF, dont_S)

    act_F = 'Market' if eu_F >= dont_S else "Don’t Market"
    act_UNF = 'Market' if eu_UNF >= dont_S else "Don’t Market"

    eu_survey = pF*best_F + pUNF*best_UNF
    do_survey = eu_survey >= best_noSurvey

    def fmt(v):
        return f"{v:.3f}"

    def fmtp(v):
        return f"{v:.2f}"

    with out:
        out.clear_output(wait=True)

        fig, ax = plt.subplots(figsize=(15, 10))
        ax.set_xlim(0, 15)
        ax.set_ylim(0, 10)
        ax.axis('off')

        OPT = '#0F6E56'
        DARK = '#2C2C2A'
        MUTED = '#888780'

        def edge(x1, y1, x2, y2, label='', opt=False):
            ax.plot([x1, x2], [y1, y2], color=OPT if opt else MUTED, lw=2.5 if opt else 1.1)
            if label:
                ax.text((x1+x2)/2, (y1+y2)/2 + 0.15, label,
                        ha='center', fontsize=9,
                        bbox=dict(facecolor='white', edgecolor='none', alpha=0.85))

        def dec(x, y, opt=False):
            ax.add_patch(plt.Rectangle((x-0.16, y-0.16), 0.32, 0.32,
                                       facecolor=OPT if opt else 'white',
                                       edgecolor=DARK, lw=1.1, zorder=5))

        def chance(x, y, opt=False):
            ax.add_patch(plt.Circle((x, y), 0.18,
                                    facecolor='white',
                                    edgecolor=OPT if opt else DARK,
                                    lw=1.3, zorder=5))

        def label(x, y, text):
            ax.text(x, y, text, fontsize=9, va='center')

        def node_value(x, y, value):
            ax.text(x-0.10, y+0.35, fmt(value), fontsize=8, ha='center')

        # Layout
        x0, x1, x2, x3, x4 = 1.2, 3.3, 5.7, 7.7, 10.0
        y0 = 4.7

        ySurvey = 7.0
        yNoSurvey = 2.4

        yF = 8.35
        yUNF = 5.75

        yMktF = 8.9
        yDontF = 7.55

        yMktU = 6.25
        yDontU = 4.9

        yHiF, yMedF, yLowF = 9.55, 8.9, 8.25
        yHiU, yMedU, yLowU = 6.9, 6.25, 5.6

        yMktNS = 3.15
        yDontNS = 1.7

        yHiNS, yMedNS, yLowNS = 3.85, 3.15, 2.45

        # Root
        dec(x0, y0, True)
        node_value(x0, y0, eu_survey if do_survey else best_noSurvey)

        # Survey
        edge(x0, y0, x1, ySurvey, 'Survey', do_survey)
        chance(x1, ySurvey, do_survey)
        node_value(x1, ySurvey, eu_survey)

        edge(x1, ySurvey, x2, yF, f'F ({fmtp(pF)})', do_survey)
        dec(x2, yF)
        node_value(x2, yF, best_F)

        edge(x2, yF, x3, yMktF, 'Market', do_survey and act_F == 'Market')
        chance(x3, yMktF)
        node_value(x3, yMktF, eu_F)

        edge(x3, yMktF, x4, yHiF, f'Hi ({fmtp(pF_Hi)})')
        label(x4, yHiF, fmt(uHi_S))

        edge(x3, yMktF, x4, yMedF, f'Med ({fmtp(pF_Med)})')
        label(x4, yMedF, fmt(uMed_S))

        edge(x3, yMktF, x4, yLowF, f'Low ({fmtp(pF_Low)})')
        label(x4, yLowF, fmt(uLow_S))

        edge(x2, yF, x3, yDontF, "Don’t market", do_survey and act_F == "Don’t Market")
        label(x3, yDontF, fmt(dont_S))

        # Unfavorable
        edge(x1, ySurvey, x2, yUNF, f'UNF ({fmtp(pUNF)})')
        dec(x2, yUNF)
        node_value(x2, yUNF, best_UNF)

        edge(x2, yUNF, x3, yMktU, 'Market', do_survey and act_UNF == 'Market')
        chance(x3, yMktU)
        node_value(x3, yMktU, eu_UNF)

        edge(x3, yMktU, x4, yHiU, f'Hi ({fmtp(pU_Hi)})')
        label(x4, yHiU, fmt(uHi_S))

        edge(x3, yMktU, x4, yMedU, f'Med ({fmtp(pU_Med)})')
        label(x4, yMedU, fmt(uMed_S))

        edge(x3, yMktU, x4, yLowU, f'Low ({fmtp(pU_Low)})')
        label(x4, yLowU, fmt(uLow_S))

        edge(x2, yUNF, x3, yDontU, "Don’t market", do_survey and act_UNF == "Don’t Market")
        label(x3, yDontU, fmt(dont_S))

        # No Survey
        edge(x0, y0, x1, yNoSurvey, "Don’t survey", not do_survey)
        dec(x1, yNoSurvey, not do_survey)
        node_value(x1, yNoSurvey, best_noSurvey)

        edge(x1, yNoSurvey, x2, yMktNS, 'Market', not do_survey and act_noSurvey == 'Market')
        chance(x2, yMktNS)
        node_value(x2, yMktNS, eu_market_NS)

        edge(x2, yMktNS, x3, yHiNS, f'Hi ({fmtp(pHi)})')
        label(x3, yHiNS, fmt(uHi_NS))

        edge(x2, yMktNS, x3, yMedNS, f'Med ({fmtp(pMed)})')
        label(x3, yMedNS, fmt(uMed_NS))

        edge(x2, yMktNS, x3, yLowNS, f'Low ({fmtp(pLow)})')
        label(x3, yLowNS, fmt(uLow_NS))

        edge(x1, yNoSurvey, x2, yDontNS, "Don’t market", not do_survey and act_noSurvey == "Don’t Market")
        label(x2, yDontNS, fmt(dont_NS))

        title = (
            f"Figure 9.12 — Marketing Decision Tree with Utilities\n"
            f"Optimal: {'Survey' if do_survey else 'Don’t Survey'} | "
            f"Expected Utility = {fmt(eu_survey if do_survey else best_noSurvey)}"
        )
        fig.suptitle(title, fontsize=11)

        plt.tight_layout()
        plt.show()

        print("Key results:")
        print(f"No survey market utility = {fmt(eu_market_NS)}")
        print(f"Favorable market utility = {fmt(eu_F)}")
        print(f"Unfavorable market utility = {fmt(eu_UNF)}")
        print(f"Survey expected utility = {fmt(eu_survey)}")


def update_utility(change):
    draw_utility(
        s_pHi.value, s_pMed.value,
        s_uHi_NS.value, s_uMed_NS.value, s_uLow_NS.value, s_dont_NS.value,
        s_uHi_S.value, s_uMed_S.value, s_uLow_S.value, s_dont_S.value,
        s_lHi.value, s_lMed.value, s_lLow.value
    )


for s in [
    s_pHi, s_pMed,
    s_uHi_NS, s_uMed_NS, s_uLow_NS, s_dont_NS,
    s_uHi_S, s_uMed_S, s_uLow_S, s_dont_S,
    s_lHi, s_lMed, s_lLow
]:
    s.observe(update_utility, names='value')


display(controls, out)

draw_utility(
    s_pHi.value, s_pMed.value,
    s_uHi_NS.value, s_uMed_NS.value, s_uLow_NS.value, s_dont_NS.value,
    s_uHi_S.value, s_uMed_S.value, s_uLow_S.value, s_dont_S.value,
    s_lHi.value, s_lMed.value, s_lLow.value
)
```
