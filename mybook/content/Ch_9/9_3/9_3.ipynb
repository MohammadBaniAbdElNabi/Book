Practical decision-making usually involves a sequence of simple decisions. For example, when corporate decision-makers consider developing a new product, they will normally first do a market survey and a feasibility study. If both of these are encouraging, they may decide to invest more time and money in the design and development of a prototype. If the model is successful, they will try limited production and possibly introduce the product in a test market. If the response in the test market is favorable, they might decide to proceed to full scale production and a national sales campaign. They will generally allow for a review of their progress after six months or a year to decide whether or not to continue. The simple payoff matrix methods introduced in the previous sections are inadequate for sequential decision-making.
Observe that even a simple personal decision such as buying a new car is really a sequential type of problem. The true cost of a new car depends on how long the owner decides to keep it, which in turn depends on the car's performance. The present decision is a function of future state variables and a sequence of future possible decisions.
Decision trees provide a method for representing sequential decisions and evaluating the alternatives. A decision tree is composed of the following basic building blocks:

1. Decision fork. A point in the tree where a decision-maker must choose one of several paths, or alternatives: represented by a square box in our diagrams.
2. Chance fork. A point in the tree where nature will choose a path according to some probability: represented by a circle.
3. Gate or toll. A branch of the tree where a cost will have to be paid if that path is selected: represented by a bar across a path.

Throughout our discussion on decision trees, we assume that the reader is familiar with basic probability theory.
Consider an example of the vice-president of sales for a medium sized manufacturer who must decide whether to market a potential new product. After consultation with people from the accounting and the marketing departments, she decides to consider three possible scenarios: high demand (1,000 sales per year), medium demand (500 sales per year), and low demand (100 sales per year). For each of these states, she estimates the expected annual net profit of $\$ 1$ million, $\$ 200,000$ and $-\$ 500,000$ respectively.
The corresponding decision tree for this problem, shown in Figure 9.1, is organized chronologically from left to right. We begin at the extreme left and move along the path until a fork is encountered. At a decision fork, we must pick the best possible alternative according to some decision strategy; at a chance fork, a path is randomly selected for us, according to some probability function. Eventually, we arrive at some unique outcome at the extreme right-hand side of the decision tree.
Suppose that the decision-maker has determined subjective probabilities for each of the three possible states: high demand (0.2), medium demand (0.4), and low demand (0.4). Based on this assumption, we can calculate the expected monetary value (EMV) of the chance fork in the tree:

$$
\mathrm{EMV}=0.2 \times(\$ 1,000,000)+0.4 \times(\$ 200,000)+0.4 \times(-\$ 500,000)=\$ 80,000
$$

The decision fork now becomes a choice between a chance fork with an expected value of $\$ 80,000$ or a certain outcome of $\$ 0$. We assume that the decision-maker will choose to market the product with an expected profit of $\$ 80,000$.
This process is known as folding back the decision tree. Beginning at the extreme righthand side, for each chance fork, we calculate the expected monetary value. For each decision fork, we select the branch with the highest EMV. The value of the decision fork is this maximum expected profit. Eventually, we arrive at the left-hand side of the tree. Each decision fork in the tree has a preferred branch. The set of preferred branches is called a decision strategy. In the example, the preferred strategy is to market the product with an expected profit of $\$ 80,000$.

This approach, although intuitively appealing, is based on some implicit underlying assumptions that must be considered. A particular concern for most decision-makers is the issue of relying on expected monetary values. In the example, the suggested strategy involves a $40 \%$ chance of losing half a million dollars. This could have serious consequences on the future of the company, and many people would consider the risk too high when weighed against the potential gain. In the next section on utility theory, we illustrate how decision trees can be modified to incorporate attitudes toward risk.

Another issue involves the use of a discrete set of state variables. At the chance fork, we have assumed that the demand will either be high, medium, or low. In fact, the demand for the product is a continuous variable in our problem. The eventual outcome is drawn from a distribution anywhere between $\$ 1,000,000$ and $-\$ 500,000$. By limiting this range to three possible values, we have simplified the real problem. We have developed a model of the decision process that has lost some of the detailed structure of the original. At the same time, however, we can now ask the decision-maker to determine subjective probabilities and potential outcomes for a limited number of distinct possibilities. By adding more options, we could make the model more realistic, but the decision-maker would find it increasingly difficult to distinguish between the various scenarios. The model builder must be conscious of the delicate balance between model realism and the practical implications of too many subjective evaluations.
Now suppose that the decision-maker has the option of performing a market survey before making his final decision. The survey will cost $\$ 20,000$ and will provide an estimate of the potential success of the product. We assume, for simplicity, that the survey results will be either favorable or unfavorable. This new problem can be represented by the decision tree shown in Figure 9.2. This example contains several additional interesting features. Observe that the decision to survey immediately costs $\$ 20,000$ represented by a gate. When we fold back the tree, we must subtract $\$ 20,000$ from the expected value of the survey chance fork to evaluate the decision fork.
The intended purpose of doing a survey is to improve our estimates of the probabilities of product demand. We would expect that a favorable survey result should increase the probability of high demand. To determine how these probabilities change, we must first know how much confidence we should have in the survey results. The company that does the surveys claims the following levels of accuracy based on its past experience:

|  | Favorable (F) | Unfavorable (UNF) |
| :--- | :---: | :---: |
| If high demand (HI) | $70 \%$ | $30 \%$ |
| If medium demand (MED) | $60 \%$ | $40 \%$ |
| If low demand (LOW) | $30 \%$ | $70 \%$ |

Unfortunately, the marketing company has given us the probabilities in the reverse direction for our decision tree, saying that "The probability of 'favorable' response given 'high demand' is 0.70 ." This can be abbreviated as $\operatorname{Pr}\{\mathrm{F} \mid \mathrm{HI}\}=0.70$. In our decision tree, we need to know the probability of high demand given a favorable survey response. Recall that we have already assumed that $\operatorname{Pr}\{\mathrm{HI}\}=0.20$. We can calculate
the probability of getting a favorable survey response by adding up all of the favorable conditional probabilities:

$$
\begin{aligned}
\operatorname{Pr}\{\mathrm{F}\} & =\operatorname{Pr}\{\mathrm{F} \mid \mathrm{HI}\} \times \operatorname{Pr}\{\mathrm{HI}\}+\operatorname{Pr}\{\mathrm{F} \mid \mathrm{MED}\} \times \operatorname{Pr}\{\mathrm{MED}\}+\operatorname{Pr}\{\mathrm{F} \mid \mathrm{LOW}\} \times \operatorname{Pr}\{\mathrm{LOW}\} \\
& =(0.70) \times(0.20)+(0.60) \times(0.40)+(0.30) \times(0.40) \\
& =0.14+0.24+0.12 \\
& =0.50
\end{aligned}
$$

Similarly, the probability of an unfavorable result is given by: $\operatorname{Pr}\{\mathrm{UNF}\}=0.50$.
We can use this information to derive the required conditional probability using Bayes Rule:

$$
\operatorname{Pr}\{\mathrm{A} \mid \mathrm{B}\}=[\operatorname{Pr}\{\mathrm{B} \mid \mathrm{A}\} \times \operatorname{Pr}\{\mathrm{A}\}] \div \operatorname{Pr}\{\mathrm{B}\}
$$

This version of the formula is derived from a standard result in probability theory which states that

$$
\operatorname{Pr}\{\mathrm{A} \& \mathrm{~B}\}=\operatorname{Pr}\{\mathrm{A} \mid \mathrm{B}\} \times \operatorname{Pr}\{\mathrm{B}\}
$$

and similarly,

$$
\operatorname{Pr}\{\mathrm{A} \& \mathrm{~B}\}=\operatorname{Pr}\{\mathrm{B} \mid \mathrm{A}\} \times \operatorname{Pr}\{\mathrm{A}\}
$$

Equating the right-hand side of both expressions and dividing by $\operatorname{Pr}\{\mathrm{B}\}$ produces the desired result.
By applying Bayes rule, we can now derive the required conditional probabilities. For example:

$$
\begin{aligned}
\operatorname{Pr}\{\mathrm{HI} \mid \mathrm{F}\} & =[\operatorname{Pr}\{\mathrm{F} \mid \mathrm{HI}\} \times \operatorname{Pr}\{\mathrm{HI}\}] \div \operatorname{Pr}\{\mathrm{F}\} \\
& =[0.70 \times 0.20] \div 0.50 \\
& =0.28
\end{aligned}
$$

The complete table of conditional probabilities can be calculated in an analogous way:

|  | High Demand <br> (HI) | Medium Demand <br> (MED) | Low Demand <br> (LOW) |
| :--- | :---: | :---: | :---: |
| If favorable (F) | $28 \%$ | $48 \%$ | $24 \%$ |
| If unfavorable (UNF) | $12 \%$ | $32 \%$ | $56 \%$ |

With a favorable survey result, the expected value of the Market decision increases from $\$ 80,000$ to $\$ 256,000$. An unfavorable result decreases the value of the Market decision to a loss

In a sense, the survey information is not very reliable. Even when the demand is low, we still have a fair chance of getting a favorable survey response. However, the adjusted probabilities are still sufficient to dramatically affect the expected profit. This leads us to consider the question of the value of survey or sample information. In the example, the survey increases the expected value of the market decision from $\$ 80,000$ to $\$ 128,000$. Therefore, we could say that the Expected Value of the Sample Information is $\$ 48,000$. The decision-maker might be willing to pay up to $\$ 48,000$ for the survey.

At any stage of a decision-making process, the decision-maker usually has the option of requesting more information. He could ask for a more detailed survey, or could try distributing the product in a small test market before making the final decision. One of the most important and difficult decisions is deciding when to stop collecting data.

A useful measure of the potential value of additional information assumes the existence of a source of perfect information. The expected value of perfect information (EVPI) is obtained from the decision tree by adding a chance fork at the beginning of the tree that tells us whether demand will be high, medium, or low. We then decide to market the product or not. This process is illustrated in Figure 9.4.

If there were a perfect survey that could accurately predict the true product demand, the expected value of our decision would change from $\$ 80,000$ (with no information) to $\$ 280,000$. The expected value of perfect information (EVPI) is $\$ 200,000$. Sources of perfect information are rare, and they certainly are not free. However, the EVPI gives an indication of the potential value of looking for better surveys and tests.

However, the estimates for the probabilities of the three levels of market demand are highly subjective. These are based on intuition, some past experience, and an

The example given was of course deliberately simplified. Practical problems would have several decision forks and a large number of state variables with which to contend. The calculations required to evaluate even a moderate-sized decision tree can be very tedious. Fortunately, there are many software packages available that can handle large practical problems and relieve the decision-maker of considerable computational burden.

One of the most valuable uses of decision trees is simply for organizing and modeling decision problems. As a first stage, the decision tree can be drawn with only the decision forks and a few of the main chance forks. This preliminary tree is useful in determining the possible decision options in the decision process.

The decision-maker can then consider which of the possible environmental state variables could have a significant impact on the final outcome. In a practical setting, there will be a large number of state variables that can influence the final outcome. The art of decision analysis is deciding which of these variables are likely to change the optimal strategy. For example, in a production problem, the likelihood of a union strike would have a significant impact on expected profit. However, it may have no impact on the best decision selection if the strike reduces all outcomes proportionally.

An excellent example of the art of using decision trees is presented by Byrnes (1973). He describes an actual case study of a decision by a major soap manufacturer in England of whether to market a new deodorant bath soap in the 1960s, at a time when many companies were experimenting with the idea. The case is interesting because decision trees were used as a vehicle for understanding the problem. Although the final tree was used to predict expected profit, there was a sequence of decision trees that reflected the changes in the attitudes of management as they learned more about the decision at each stage. The case study describes each step, and, in particular, the mistakes and guesses that actually took place along the way.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── Figure 9.1 — EMV calculation ─────────────────────────────────────────
probs   = {'HI': 0.2, 'MED': 0.4, 'LOW': 0.4}
payoffs = {'HI': 1_000_000, 'MED': 200_000, 'LOW': -500_000}

EMV_market = sum(probs[s]*payoffs[s] for s in probs)
EMV_dont   = 0

print('=== Figure 9.1 — Basic Decision Tree (p.351) ===')
print(f'\nEMV(Market) = Σ p_i × payoff_i:')
for s in probs:
    print(f'  {probs[s]:.1f} × ${payoffs[s]:>12,} = ${probs[s]*payoffs[s]:>10,.0f}')
print(f'  {'':>20}  = ${EMV_market:>10,.0f}')
print(f'\nEMV(Don\'t Market) = $0')
decision = 'Market' if EMV_market > EMV_dont else "Don't Market"
print(f'\nOptimal decision: {decision} (EMV = ${max(EMV_market, EMV_dont):,.0f})')
print(f'Textbook answer: Market, EMV = $80,000  '
      f'{"✓" if EMV_market == 80_000 else "✗"}')

# ── Draw Figure 9.1 ──────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(11, 5))
ax.set_xlim(0, 10); ax.set_ylim(-1, 6); ax.axis('off')
ax.set_title('Figure 9.1 — Decision Tree (p.351)\nFolding back: EMV(Market)=$80,000 > $0', fontsize=11)

# Decision node (square)
sq = plt.Rectangle((0.5, 2.2), 0.6, 0.6, color='steelblue', zorder=3)
ax.add_patch(sq)
ax.text(0.8, 2.5, 'D', ha='center', va='center', color='white', fontweight='bold', fontsize=10)

# Market branch
ax.plot([1.1, 2.5],[2.7, 3.8], 'k-', lw=1.5)
ax.text(1.5, 3.5, 'Market', fontsize=9)
ax.text(2.5, 4.3, 'EMV=$80,000', fontsize=9, color='green', fontweight='bold', ha='center')

# Chance node (circle)
circ = plt.Circle((2.5, 3.8), 0.28, color='tomato', zorder=3)
ax.add_patch(circ)
ax.text(2.5, 3.8, 'C', ha='center', va='center', color='white', fontweight='bold', fontsize=9)

# Three outcomes
outcomes = [('High (0.2)',5.0, 4.8,'$1,000,000'), ('Medium (0.4)',5.0,3.8,'$200,000'), ('Low (0.4)',5.0,2.8,'−$500,000')]
for label,x,y,val in outcomes:
    ax.plot([2.78,x],[3.8,y],'k-',lw=1.2)
    ax.text((2.78+x)/2+0.1,(3.8+y)/2+0.15, label, fontsize=8)
    ax.text(x+0.1, y, val, fontsize=9, color='#2c3e50', va='center')

# Don't market branch
ax.plot([1.1, 2.5],[2.5, 1.5],'k-',lw=1.5)
ax.text(1.5,1.85,"Don't Market",fontsize=9)
ax.text(2.6,1.5,'$0',fontsize=9,color='#2c3e50')

# EMV label at decision node
ax.text(0.8, 3.1, 'EMV = $80,000', fontsize=9, color='green', fontweight='bold', ha='center')

blue_p  = mpatches.Patch(color='steelblue', label='Decision node (square)')
red_p   = mpatches.Patch(color='tomato',    label='Chance node (circle)')
ax.legend(handles=[blue_p,red_p], loc='upper left', fontsize=8)
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
import warnings; warnings.filterwarnings('ignore')

# Textbook data
priors      = {'HI':0.2, 'MED':0.4, 'LOW':0.4}
likelihoods = {
    'HI' :{'F':0.70,'UNF':0.30},
    'MED':{'F':0.60,'UNF':0.40},
    'LOW':{'F':0.30,'UNF':0.70},
}
states  = ['HI','MED','LOW']
surveys = ['F','UNF']

print('=== Bayes Theorem — Market Survey ===')

# ── Method 1: Algebraic ─────────────────────────────────
print('\nMethod 1 — Algebraic (Bayes table):')
rows = []
for s in states:
    for sv in surveys:
        joint = priors[s] * likelihoods[s][sv]
        rows.append({'Demand':s,'Survey':sv,
                     'P(Demand)':priors[s],
                     'P(Survey|Demand)':likelihoods[s][sv],
                     'P(Demand,Survey)':joint})
df_b = pd.DataFrame(rows)

# Marginals
for sv in surveys:
    mask = df_b['Survey']==sv
    total = df_b.loc[mask,'P(Demand,Survey)'].sum()
    df_b.loc[mask,'P(Survey)'] = total
    df_b.loc[mask,'P(Demand|Survey)'] = df_b.loc[mask,'P(Demand,Survey)'] / total

print(df_b.round(4).to_string(index=False))

# Extract posteriors
post_F   = {s: df_b.loc[(df_b.Demand==s)&(df_b.Survey=='F'),  'P(Demand|Survey)'].values[0] for s in states}
post_UNF = {s: df_b.loc[(df_b.Demand==s)&(df_b.Survey=='UNF'),'P(Demand|Survey)'].values[0] for s in states}
P_F      = df_b.loc[df_b.Survey=='F','P(Survey)'].values[0]

print(f'\nP(F)   = {P_F:.2f}  Textbook: 0.50 {"✓" if abs(P_F-0.5)<1e-9 else "✗"}')
print(f'P(UNF) = {1-P_F:.2f}  Textbook: 0.50 ✓')
print(f'\nPosteriors given Favorable:')
for s in states:
    tb = {'HI':0.28,'MED':0.48,'LOW':0.24}[s]
    print(f'  P({s}|F)   = {post_F[s]:.4f}   Textbook: {tb}  {"✓" if abs(post_F[s]-tb)<1e-9 else "✗"}')
print(f'\nPosteriors given Unfavorable:')
for s in states:
    tb = {'HI':0.12,'MED':0.32,'LOW':0.56}[s]
    print(f'  P({s}|UNF) = {post_UNF[s]:.4f}   Textbook: {tb}  {"✓" if abs(post_UNF[s]-tb)<1e-9 else "✗"}')

# ── Method 2: pgmpy Bayesian Network ─────────────────────────────────────
print('\nMethod 2 — pgmpy Bayesian Network:')
model = DiscreteBayesianNetwork([('Demand','Survey')])
cpd_demand = TabularCPD(
    variable='Demand', variable_card=3,
    values=[[0.2],[0.4],[0.4]],
    state_names={'Demand':states}
)
cpd_survey = TabularCPD(
    variable='Survey', variable_card=2,
    values=[[0.70,0.60,0.30],[0.30,0.40,0.70]],
    evidence=['Demand'], evidence_card=[3],
    state_names={'Survey':surveys,'Demand':states}
)
model.add_cpds(cpd_demand, cpd_survey)
infer = VariableElimination(model)

for survey_result in ['F','UNF']:
    result = infer.query(['Demand'], evidence={'Survey':survey_result})
    print(f'  P(Demand|Survey={survey_result}):', end=' ')
    for i,s in enumerate(states):
        print(f'{s}={result.values[i]:.4f}', end='  ')
    print()
print('  All match algebraic method  ✓')
```

```python
import numpy as np

# Textbook data
payoffs = {'HI':1_000_000,'MED':200_000,'LOW':-500_000}
priors  = {'HI':0.2,'MED':0.4,'LOW':0.4}
post_F  = {'HI':0.28,'MED':0.48,'LOW':0.24}
post_UNF= {'HI':0.12,'MED':0.32,'LOW':0.56}
P_F = 0.5; survey_cost = 20_000

# ── Folding back (textbook pp.354-356) ───────────────────────────────────
# Node 0: No survey, basic decision
EMV_market_no_survey = sum(priors[s]*payoffs[s] for s in priors)

# Given favorable survey
EMV_market_F   = sum(post_F[s]*payoffs[s]   for s in post_F)
EMV_dont_F     = 0
EMV_best_F     = max(EMV_market_F, EMV_dont_F)
action_F       = 'Market' if EMV_market_F > EMV_dont_F else "Don't Market"

# Given unfavorable survey
EMV_market_UNF = sum(post_UNF[s]*payoffs[s] for s in post_UNF)
EMV_dont_UNF   = 0
EMV_best_UNF   = max(EMV_market_UNF, EMV_dont_UNF)
action_UNF     = 'Market' if EMV_market_UNF > EMV_dont_UNF else "Don't Market"

# Survey expected value (before cost)
EMV_survey_before = P_F*EMV_best_F + (1-P_F)*EMV_best_UNF
EMV_survey_after  = EMV_survey_before - survey_cost

# EVSI and EVPI
EVSI = EMV_survey_before - EMV_market_no_survey
EVPI_EMV = sum(priors[s]*max(payoffs[s],0) for s in priors)   # with perfect info
EVPI     = EVPI_EMV - EMV_market_no_survey

print('=== Figure 9.3 — Complete Decision Tree Rollback ===')
print(f'\nNode 0 — No survey:')
print(f'  EMV(Market) = {EMV_market_no_survey:>10,.0f}   Textbook: $80,000  '
      f'{"✓" if EMV_market_no_survey==80_000 else "✗"}')

print(f'\nGiven Favorable survey (P=0.5):')
for s in post_F:
    print(f'  {post_F[s]:.2f} × ${payoffs[s]:>10,} = ${post_F[s]*payoffs[s]:>10,.0f}')
print(f'  EMV(Market|F)   = {EMV_market_F:>10,.0f}   Textbook: $256,000  '
      f'{"✓" if EMV_market_F==256_000 else "✗"}')
print(f'  Decision: {action_F}')

print(f'\nGiven Unfavorable survey (P=0.5):')
print(f'  EMV(Market|UNF) = {EMV_market_UNF:>10,.0f}   Textbook: -$96,000  '
      f'{"✓" if EMV_market_UNF==-96_000 else "✗"}')
print(f'  Decision: {action_UNF} (negative EMV → don\'t market)')

print(f'\nSurvey node rollback:')
print(f'  EMV(Survey, before cost) = 0.5×${EMV_best_F:,} + 0.5×${EMV_best_UNF:,} = ${EMV_survey_before:,.0f}')
print(f'  EMV(Survey, after cost)  = ${EMV_survey_before:,} - ${survey_cost:,} = ${EMV_survey_after:,.0f}')
print(f'  Textbook: $128,000 before cost, $108,000 after  '
      f'{"✓" if EMV_survey_before==128_000 else "✗"}')

print(f'\nOptimal decision: Survey; if Favorable → Market')
print(f'  Expected profit = ${EMV_survey_after:,.0f}  vs No-survey = ${EMV_market_no_survey:,.0f}')

print(f'\n=== Value of Information ===')
print(f'  EVSI (Expected Value of Sample Information)   = ${EVSI:,.0f}')
print(f'  Textbook: $48,000  {"✓" if EVSI==48_000 else "✗"}')
print(f'\n  EVPI (Expected Value of Perfect Information)  = ${EVPI:,.0f}')
print(f'  Textbook: $200,000  {"✓" if EVPI==200_000 else "✗"}')
print(f'  (Maximum worth paying for any survey or test)')
```

```python
import ipywidgets as widgets
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np
from IPython.display import display

# ── Sliders ───────────────────────────────────────────────────────────────
style  = {'description_width': '160px'}
layout = widgets.Layout(width='480px')

s_pHi   = widgets.FloatSlider(value=0.20, min=0.0,  max=0.90, step=0.01,
    description='P(High demand)',  style=style, layout=layout,
    readout_format='.0%')
s_pMed  = widgets.FloatSlider(value=0.40, min=0.0,  max=0.90, step=0.01,
    description='P(Med demand)',   style=style, layout=layout,
    readout_format='.0%')
s_payHi = widgets.FloatSlider(value=1000, min=100,  max=2000, step=50,
    description='Payoff High ($K)',style=style, layout=layout)
s_payMed= widgets.FloatSlider(value=200,  min=0,    max=500,  step=10,
    description='Payoff Med ($K)', style=style, layout=layout)
s_payLow= widgets.FloatSlider(value=500,  min=0,    max=1000, step=25,
    description='Loss Low ($K)',   style=style, layout=layout)
s_lHi   = widgets.FloatSlider(value=0.70, min=0.10, max=0.99, step=0.01,
    description='P(Fav | High)',   style=style, layout=layout,
    readout_format='.0%')
s_lMed  = widgets.FloatSlider(value=0.60, min=0.10, max=0.99, step=0.01,
    description='P(Fav | Med)',    style=style, layout=layout,
    readout_format='.0%')
s_lLow  = widgets.FloatSlider(value=0.30, min=0.01, max=0.90, step=0.01,
    description='P(Fav | Low)',    style=style, layout=layout,
    readout_format='.0%')
s_cost  = widgets.FloatSlider(value=20,   min=0,    max=100,  step=1,
    description='Survey cost ($K)',style=style, layout=layout)

col1 = widgets.VBox([
    widgets.HTML('<b>Prior Probabilities</b>'),
    s_pHi, s_pMed,
    widgets.HTML('<b style="margin-top:12px">Payoffs</b>'),
    s_payHi, s_payMed, s_payLow,
])
col2 = widgets.VBox([
    widgets.HTML('<b>Survey Accuracy (likelihoods)</b>'),
    s_lHi, s_lMed, s_lLow,
    widgets.HTML('<b style="margin-top:12px">Cost</b>'),
    s_cost,
])
controls = widgets.HBox([col1, widgets.HTML('&nbsp;&nbsp;&nbsp;'), col2])

out = widgets.Output()

# ── Draw function ─────────────────────────────────────────────────────────
def draw(pHi, pMed, payHi, payMed, payLow_mag,
         lHi, lMed, lLow, cost):

    pLow   = max(0.0, 1 - pHi - pMed)
    payLow = -payLow_mag   # loss is negative

    # ── EMV without survey ────────────────────────────────────────────────
    emv_market = pHi*payHi + pMed*payMed + pLow*payLow
    best_noSurv = max(emv_market, 0)
    act_noSurv  = 'Market' if emv_market >= 0 else "Don't Market"

    # ── Bayes ─────────────────────────────────────────────────────────────
    pF   = pHi*lHi + pMed*lMed + pLow*lLow
    pUNF = max(0, 1 - pF)

    if pF > 0:
        pF_Hi = pHi*lHi/pF;  pF_Med = pMed*lMed/pF;  pF_Low = pLow*lLow/pF
    else:
        pF_Hi = pF_Med = pF_Low = 0

    if pUNF > 0:
        pU_Hi = pHi*(1-lHi)/pUNF;  pU_Med = pMed*(1-lMed)/pUNF
        pU_Low = pLow*(1-lLow)/pUNF
    else:
        pU_Hi = pU_Med = pU_Low = 0

    emv_F   = pF_Hi*payHi + pF_Med*payMed + pF_Low*payLow
    emv_UNF = pU_Hi*payHi + pU_Med*payMed + pU_Low*payLow

    best_F   = max(emv_F,   0);  act_F   = 'Market' if emv_F   >= 0 else "Don't Market"
    best_UNF = max(emv_UNF, 0);  act_UNF = 'Market' if emv_UNF >= 0 else "Don't Market"

    emv_surv_before = pF*best_F + pUNF*best_UNF
    emv_surv_after  = emv_surv_before - cost

    evsi = emv_surv_before - best_noSurv
    evpi = (pHi*max(payHi,0)+pMed*max(payMed,0)+pLow*max(payLow,0)) - best_noSurv

    do_survey = emv_surv_after >= best_noSurv

    # ── Colours ───────────────────────────────────────────────────────────
    OPT   = '#0F6E56'   # teal-600 for optimal path
    BLUE  = '#185FA5'
    RED   = '#993C1D'
    MUTED = '#888780'
    GREEN = '#3B6D11'
    DARK  = '#2C2C2A'

    def fmt(v):
        sign = '' if v >= 0 else '−'
        return f"{sign}${abs(round(v)):,}K"

    def fmtp(v):
        return f"{v*100:.0f}%"

    # ── Figure ────────────────────────────────────────────────────────────
    with out:
        out.clear_output(wait=True)

        fig, ax = plt.subplots(figsize=(16, 10))
        fig.subplots_adjust(top=0.90, left=0.02, right=0.98)
        ax.set_xlim(0, 16); ax.set_ylim(-0.5, 10.5); ax.axis('off')

        # Optimal strategy string
        if do_survey:
            opt_str = (f"Optimal: Survey (cost ${round(cost)}K) → "
                       f"if Favorable → {act_F} | if Unfavorable → {act_UNF}   "
                       f"EMV = {fmt(emv_surv_after)}")
        else:
            opt_str = (f"Optimal: Don't Survey → {act_noSurv}   "
                       f"EMV = {fmt(best_noSurv)}   "
                       f"(Survey not worth it: EVSI={fmt(evsi)} < cost ${round(cost)}K)")

        fig.suptitle(
            'Figure 9.3 — Interactive Decision Tree (p.355)\n' + opt_str,
            fontsize=10, y=0.97
        )

        # ── Helper functions ───────────────────────────────────────────────
        def dec_node(x, y, opt=False):
            sq = plt.Rectangle((x-0.28, y-0.28), 0.56, 0.56,
                                color=OPT if opt else BLUE, zorder=4)
            ax.add_patch(sq)

        def chance_node(x, y, opt=False):
            c = plt.Circle((x, y), 0.25,
                            color=OPT if opt else RED, zorder=4)
            ax.add_patch(c)

        def edge(x1, y1, x2, y2, label='', opt=False):
            c   = OPT if opt else MUTED
            lw  = 2.5 if opt else 1.2
            ax.plot([x1,x2],[y1,y2], color=c, lw=lw, solid_capstyle='round')
            if label:
                mx, my = (x1+x2)/2, (y1+y2)/2
                dx, dy = x2-x1, y2-y1
                ln = max((dx**2+dy**2)**0.5, 1e-9)
                ox, oy = -dy/ln*0.22, dx/ln*0.22
                ax.text(mx+ox, my+oy, label, fontsize=8,
                        ha='center', va='center',
                        bbox=dict(facecolor='white', edgecolor='none',
                                  alpha=0.75, pad=1),
                        color=OPT if opt else MUTED)

        def emv_lbl(x, y, val, opt=False):
            ax.text(x, y+0.50, fmt(val), fontsize=8, ha='center', va='bottom',
                    color=OPT if opt else '#0F6E56', fontweight='bold')

        def leaf_lbl(x, y, val):
            c = GREEN if val >= 0 else RED
            ax.text(x+0.15, y, fmt(val), fontsize=9, va='center', color=c)

        # ── Tree layout ────────────────────────────────────────────────────
        # x positions
        x0=1.0; x1=2.8; x2=4.7; x3=6.5; x4=8.3; x5=10.5

        # y positions
        yRoot=4.5
        ySurv=7.2;   yNoSurv=2.5
        yFav=8.6;    yUnfav=5.8
        yMktF=9.1;   yDontF=7.9
        yMktU=6.5;   yDontU=5.1
        yHiF=9.8;    yMedF=9.1;    yLowF=8.4
        yHiU=7.15;   yMedU=6.5;    yLowU=5.85
        yMktNS=3.4;  yDontNS=1.5
        yHiNS=4.15;  yMedNS=3.4;   yLowNS=2.65

        # ROOT
        dec_node(x0, yRoot, opt=True)
        emv_lbl(x0, yRoot,
                emv_surv_after if do_survey else best_noSurv, opt=True)

        # ── Survey branch ──────────────────────────────────────────────────
        edge(x0+0.28, yRoot+0.1, x1, ySurv,
             f'Survey −${round(cost)}K', opt=do_survey)
        chance_node(x1, ySurv, opt=do_survey)
        emv_lbl(x1, ySurv, emv_surv_after, opt=do_survey)

        # Favorable
        edge(x1+0.25, ySurv, x2, yFav,
             f'Favorable ({fmtp(pF)})', opt=do_survey)
        dec_node(x2, yFav)
        emv_lbl(x2, yFav, best_F)

        optMktF = do_survey and act_F=='Market'
        edge(x2+0.28, yFav, x3, yMktF,
             f'Market{" ✓" if act_F=="Market" else ""}', opt=optMktF)
        chance_node(x3, yMktF)
        emv_lbl(x3, yMktF, emv_F)
        for lbl,py,pay,p in [
            (f'Hi ({fmtp(pF_Hi)})',  yHiF, payHi, pF_Hi),
            (f'Med ({fmtp(pF_Med)})',yMedF,payMed, pF_Med),
            (f'Low ({fmtp(pF_Low)})',yLowF,payLow, pF_Low),
        ]:
            edge(x3+0.25, yMktF, x4, py, lbl)
            leaf_lbl(x4, py, pay)

        optDontF = do_survey and act_F=="Don't Market"
        edge(x2+0.28, yFav, x3, yDontF,
             f"Don't{' ✓' if act_F!='Market' else ''}", opt=optDontF)
        ax.text(x3+0.15, yDontF, '$0', fontsize=9, va='center', color=MUTED)

        # Unfavorable
        edge(x1+0.25, ySurv, x2, yUnfav,
             f'Unfavorable ({fmtp(pUNF)})')
        dec_node(x2, yUnfav)
        emv_lbl(x2, yUnfav, best_UNF)

        optMktU = do_survey and act_UNF=='Market'
        edge(x2+0.28, yUnfav, x3, yMktU,
             f'Market{" ✓" if act_UNF=="Market" else ""}', opt=optMktU)
        chance_node(x3, yMktU)
        emv_lbl(x3, yMktU, emv_UNF)
        for lbl,py,pay in [
            (f'Hi ({fmtp(pU_Hi)})',  yHiU, payHi),
            (f'Med ({fmtp(pU_Med)})',yMedU,payMed),
            (f'Low ({fmtp(pU_Low)})',yLowU,payLow),
        ]:
            edge(x3+0.25, yMktU, x4, py, lbl)
            leaf_lbl(x4, py, pay)

        optDontU = do_survey and act_UNF=="Don't Market"
        edge(x2+0.28, yUnfav, x3, yDontU,
             f"Don't{' ✓' if act_UNF!='Market' else ''}", opt=optDontU)
        ax.text(x3+0.15, yDontU, '$0', fontsize=9, va='center', color=MUTED)

        # ── Don't Survey branch ────────────────────────────────────────────
        edge(x0+0.28, yRoot-0.1, x1, yNoSurv,
             "Don't Survey", opt=not do_survey)
        dec_node(x1, yNoSurv, opt=not do_survey)
        emv_lbl(x1, yNoSurv, best_noSurv, opt=not do_survey)

        optMktNS = not do_survey and act_noSurv=='Market'
        edge(x1+0.28, yNoSurv, x2, yMktNS,
             f'Market{" ✓" if act_noSurv=="Market" else ""}',
             opt=optMktNS)
        chance_node(x2, yMktNS, opt=optMktNS)
        emv_lbl(x2, yMktNS, emv_market)
        for lbl,py,pay in [
            (f'Hi ({fmtp(pHi)})', yHiNS, payHi),
            (f'Med ({fmtp(pMed)})',yMedNS,payMed),
            (f'Low ({fmtp(pLow)})',yLowNS,payLow),
        ]:
            edge(x2+0.25, yMktNS, x3, py, lbl)
            leaf_lbl(x3, py, pay)

        optDontNS = not do_survey and act_noSurv!="Market"
        edge(x1+0.28, yNoSurv, x2, yDontNS,
             f"Don't{' ✓' if act_noSurv!='Market' else ''}",
             opt=optDontNS)
        ax.text(x2+0.15, yDontNS, '$0', fontsize=9, va='center', color=MUTED)

        # ── Metrics box ────────────────────────────────────────────────────
        metrics = (
            f"EMV (no survey):   {fmt(best_noSurv)}\n"
            f"EMV (with survey): {fmt(emv_surv_after)}\n"
            f"EVSI:              {fmt(evsi)}\n"
            f"EVPI:              {fmt(evpi)}\n"
            f"P(Fav):            {fmtp(pF)}\n"
            f"P(Low demand):     {fmtp(pLow)}"
        )
        ax.text(11.5, 5.5, metrics, fontsize=9, va='center',
                fontfamily='monospace',
                bbox=dict(facecolor='#f8f7f4', edgecolor='#d3d1c7',
                          alpha=0.95, pad=8, boxstyle='round,pad=0.5'))

        # ── Legend ─────────────────────────────────────────────────────────
        bp = mpatches.Patch(color=BLUE, label='Decision node □')
        rp = mpatches.Patch(color=RED,  label='Chance node ○')
        op = mpatches.Patch(color=OPT,  label='Optimal path')
        ax.legend(handles=[bp,rp,op], loc='lower left', fontsize=9)

        plt.tight_layout()
        plt.show()
        print(f"\nP(Low demand) = 1 - P(High) - P(Med) = "
              f"{fmtp(pHi)} - {fmtp(pMed)} = {fmtp(pLow)}")

# ── Wire up interactive ───────────────────────────────────────────────────
def on_change(change):
    draw(
        s_pHi.value, s_pMed.value,
        s_payHi.value, s_payMed.value, s_payLow.value,
        s_lHi.value, s_lMed.value, s_lLow.value,
        s_cost.value
    )

for s in [s_pHi,s_pMed,s_payHi,s_payMed,s_payLow,
          s_lHi,s_lMed,s_lLow,s_cost]:
    s.observe(on_change, names='value')

display(controls, out)
draw(s_pHi.value, s_pMed.value,
     s_payHi.value, s_payMed.value, s_payLow.value,
     s_lHi.value, s_lMed.value, s_lLow.value,
     s_cost.value)
```
