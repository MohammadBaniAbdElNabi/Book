# 7.3.2 Steady State Performance Measures

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/drive/1Xp0py23RKm_OL6onmhAz_Rq5rPgUxauk?usp=sharing)

If we consider any state in the rate diagram shown in Figure 7.2, and assume the system to have reached steady state, then it must be true that the mean rate at which transitions are made into the state must equal the mean rate at which transitions are made out of the state. In fact, we can write an equation for each state that expresses this fact and that accounts for all possible ways in which transitions into and out of the state can occur. The set of equations that results from doing this are called flow balancing equations, and we can write them in the following way.
First consider state 0 , the simplest case, because there is only one path leading in and one path leading out. The mean rate out of state 0 is the probability of being in state 0 times the rate of departures from state $0, \mathrm{p}_{0} \lambda$. The mean rate in is the probability of being in state 1 times the rate of transitions from state 1 to state $0, \mathrm{p}_{1} \mu$. Therefore, the equation for state 0 is

$$
\mathrm{p}_{0} \lambda=\mathrm{p}_{1} \mu
$$

For all other states, there are two arcs leading in and two arcs leading out. Still, the rate-in = rate-out principle holds and we can write for state 1 :

$$
\begin{aligned}
\text { Rate in: } & p_{0} \lambda+p_{2} \mu \\
\text { Rate out: } & p_{1} \lambda+p_{1} \mu \\
\text { therefore } & p_{0} \lambda+p_{2} \mu=p_{1} \lambda+p_{1} \mu
\end{aligned}
$$

And similarly for state 2,

$$
\mathrm{p}_{1} \lambda+\mathrm{p}_{3} \mu=\mathrm{p}_{2} \lambda+\mathrm{p}_{2} \mu
$$

and for state n

$$
\mathrm{p}_{\mathrm{n}-1} \lambda+\mathrm{p}_{\mathrm{n}+1} \mu=\mathrm{p}_{\mathrm{n}} \lambda+\mathrm{p}_{\mathrm{n}} \mu
$$

By first solving the state 0 equation for $p_{1}$ in terms of $p_{0}$, we can proceed to the state 1 equation and solve for $\mathrm{p}_{2}$ in terms of $\mathrm{p}_{0}$, and successively solve for all $\mathrm{p}_{\mathrm{n}}$ as follows:

$$
\begin{aligned}
& \mathrm{p}_{1}=\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{0} \\
& \mathrm{p}_{2}=\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{1}=\left(\frac{\lambda}{\mu}\right)^{2} \mathrm{p}_{0} \\
& \mathrm{p}_{3}=\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{2}=\left(\frac{\lambda}{\mu}\right)^{2} \mathrm{p}_{1}=\left(\frac{\lambda}{\mu}\right)^{3} \mathrm{p}_{0} \\
& \mathrm{p}_{\mathrm{n}}=\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{\mathrm{n}-1}=\left(\frac{\lambda}{\mu}\right)^{\mathrm{n}} \mathrm{p}_{0}
\end{aligned}
$$

So, for the birth-and-death process model in which all arrivals are characterized by the parameter $\lambda$ and all departures by the parameter $\mu$, any of the $\mathrm{p}_{\mathrm{i}}$ can be computed in terms of the parameters $\lambda$ and $\mu$ and the probability $\mathrm{p}_{0}$. To obtain the value of $\mathrm{p}_{0}$, we just observe that the sum of all the $\mathrm{p}_{\mathrm{i}}$ must equal to one:

$$
\mathrm{p}_{0}+\mathrm{p}_{1}+\mathrm{p}_{2}+\ldots+\mathrm{p}_{\mathrm{n}}+\ldots=1
$$

Then,

$$
\begin{aligned}
1 & =\mathrm{p}_{0}+\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{0}+\left(\frac{\lambda}{\mu}\right)^{2} \mathrm{p}_{0}+\ldots+\left(\frac{\lambda}{\mu}\right)^{\mathrm{n}} \mathrm{p}_{0}+\ldots \\
& =\mathrm{p}_{0}\left[1+\left(\frac{\lambda}{\mu}\right)+\left(\frac{\lambda}{\mu}\right)^{2}+\ldots+\left(\frac{\lambda}{\mu}\right)^{\mathrm{n}}+\ldots\right]
\end{aligned}
$$

The series in square brackets converges, if $(\lambda / \mu)<1$, to the quantity

$$
\frac{1}{1-(\lambda / \mu)}
$$

Therefore,

$$
1=\mathrm{p}_{0} \frac{1}{1-(\lambda / \mu)}
$$

and

$$
\mathrm{p}_{0}=1-\left(\frac{\lambda}{\mu}\right)
$$

More intuitively, you may also recall that $\rho=\lambda / \mathrm{s} \mu$ is the probability of the service facility being in use at any given time. For a single service facility in which the same parameter $\mu$ characterizes service times (as in Figure 7.2), regardless of the number of customers requiring service, we let $\mathrm{s}=1$. Therefore, $\rho=\lambda / \mu$ is the probability of a busy service facility, and thus $1-(\lambda / \mu)=1-\rho$ is the probability of an idle service facility. This is exactly what is meant by the probability of there being zero customers in the system, so it is reasonable that $\mathrm{p}_{0}=1-(\lambda / \mu)$.
We can now express all of the system state probabilities in terms of $\lambda$ and $\mu$ as follows:

$$
\begin{aligned}
& \mathrm{p}_{\mathrm{n}}=\left(\frac{\lambda}{\mu}\right)^{\mathrm{n}} \mathrm{p}_{0} \\
& \mathrm{p}_{\mathrm{n}}=\left(\frac{\lambda}{\mu}\right)^{\mathrm{n}}\left[1-\left(\frac{\lambda}{\mu}\right)\right] \text { or } \\
& \mathrm{p}_{\mathrm{n}}=\rho^{\mathrm{n}}(1-\rho)
\end{aligned}
$$

Our original purpose in developing these formulae was so that we could compute system performance measures such as the expected number of customers in the system L, and the expected amount of time each customer spends in the system. By defining the expected number of customers, we know

$$
\begin{aligned}
L & =\sum_{n=0}^{\infty}\left(n \cdot p_{n}\right) \\
& =\sum_{n=0}^{\infty}\left[n \times \rho^{n}(1-\rho)\right] \\
& =(1-\rho)\left[1 \rho^{1}+2 \rho^{2}+\ldots+n \rho^{n}+\ldots\right] \\
& =(1-\rho)\left[1+\rho+\rho^{2}+\rho^{3}+\rho^{4}+\ldots\right]\left[\rho+\rho^{2}+\rho^{3}+\rho^{4}+\ldots\right] \\
& =(1-\rho)\left(\sum_{n=0}^{\infty} \rho^{n}\right)\left(\sum_{n=0}^{\infty} \rho^{n}-1\right) \\
& =(1-\rho)\left(\frac{1}{(1-\rho)}\right)\left(\frac{1}{(1-\rho)}-1\right)
\end{aligned}
$$

since

$$
\sum_{n=0}^{\infty} x^{n}=\frac{1}{(1-x)} \text { for }|x|<1
$$

Therefore,

$$
\mathrm{L}=\frac{\rho}{1-\rho}
$$

From this we can use Little's formula $L=\lambda W$ to obtain the expected time in the system W . Because $\mathrm{W}=\mathrm{W}_{\mathrm{q}}+1 / \mu$, we can then compute the expected time a customer spends in queue, $\mathrm{W}_{\mathrm{q}}$. And from this we can obtain the expected queue length $\mathrm{L}_{\mathrm{q}}$ using $\mathrm{L}_{\mathrm{q}}=\lambda \mathrm{W}_{\mathrm{q}}$.

**Example 7.3** <br>

Suppose computer programs are submitted for execution on a university's central computing facility, and that these programs arrive at a rate of 10 per minute. Assume average run-time for a program is five seconds, and that both interarrival times and run-times are exponentially distributed. During what fraction of the time is the CPU idle? What is the expected turnaround time of a job in this system? What is the average number of jobs in the job queue?

This system is assumed to be an M/M/1 queueing system with $\lambda=10$ jobs per minute and $\mu=12$ jobs per minute. We will also assume that job queues may become arbitrarily long and that there is an infinitely large user population. Since $\rho=10 / 12<1$, the system will reach steady-state and we can use the formulae developed earlier to answer these questions. Since the utilization factor $\rho=5 / 6$, the CPU will be idle $1 / 6$ of the time, or for 10 seconds out of every minute. (Since 10 jobs each take an average of five seconds, the CPU is busy for 50 seconds each minute.)

Turnaround time is defined to be waiting time plus execution time, which we call W . We know $\mathrm{L}=\rho /(1-\rho)=(5 / 6) /(1 / 6)=5$, and from this we can calculate, using Little's formula, $\mathrm{W}=\mathrm{L} / \lambda=5 / 10=1 / 2$ minute. The average queue length is $\mathrm{L}_{\mathrm{q}}=\lambda \mathrm{W}_{\mathrm{q}}$. Since we have just computed W to be $\frac{1}{2}$ minute, we can use the formula $\mathrm{W}_{\mathrm{q}}=\mathrm{W}-1 / \mu=1 / 2-1 / 12=5 / 12$ minutes (or 25 seconds) for expected waiting time spent in the queue. Then the average number of jobs in the queue is $\mathrm{L}_{\mathrm{q}}=10 \cdot 5 / 12=4 \frac{1}{6}$ jobs. (Because the job queue itself occupies some computer memory, this tells how much space is typically devoted to this system function, and it also indicates how many jobs are experiencing delay.)

Now suppose we want to know the probability that the number of jobs in the system becomes 4 or more. This can be calculated as 1 minus the probability that there are fewer than 4 (i.e., $0,1,2$, or 3 ) customers in the system:

$$
\text { Probability }[\geq 4 \text { jobs }]=1-\left[\mathrm{p}_{0}+\mathrm{p}_{1}+\mathrm{p}_{2}+\mathrm{p}_{3}\right]
$$

We know

$$
\begin{aligned}
& \mathrm{p}_{0}=1-\frac{\lambda}{\mu}=1-\frac{5}{6}=0.1667 \\
& \mathrm{p}_{1}=\rho \mathrm{p}_{0}=\left(\frac{5}{6}\right)\left(\frac{1}{6}\right)=\frac{5}{36}=0.1389 \\
& \mathrm{p}_{2}=\rho^{2} \mathrm{p}_{0}=\left(\frac{5}{6}\right)^{2}\left(\frac{1}{6}\right)=0.1158 \\
& \mathrm{p}_{3}=\rho^{3} \mathrm{p}_{0}=\left(\frac{5}{6}\right)^{3}\left(\frac{1}{6}\right)=0.0965
\end{aligned}
$$

Therefore,

$$
\text { Probability }[\geq 4 \text { jobs }]=1-[0.1667+0.1389+0.1158+0.0965]=0.48225
$$

In general, the probability that there are at least k jobs in the system is given by:

$$
\begin{aligned}
1-\sum_{n=0}^{k-1} p_{k} & =1-\sum_{n=0}^{k-1} \rho^{n}(1-\rho) \\
& =1-\frac{(1-\rho)\left(1-p^{k}\right)}{(1-\rho)} \\
& =\rho^{k}
\end{aligned}
$$

In our example, $\rho=5 / 6$, and $\mathrm{k}=4$, so the probability of at least four jobs in the system is $(5 / 6)^{4}=0.48225$.

The formulas developed earlier are valid only in queueing systems that eventually reach steady-state. Our underlying assumption that the arrival rate $\lambda$ be less than the service rate $\mu$ is sufficient to guarantee that the system will stabilize. Notice that as $\rho$ approaches 1 , both W and $\mathrm{W}_{\mathrm{q}}$ become large. Clearly, for $\rho>1$, arrivals are occurring faster than a constantly busy service facility can keep up with the demand. When $\rho=1$, the sequence is undefined. However, if we look back at the original state equations, we discover that $(\lambda / \mu)=1$ implies that $p_{0}=p_{1}=p_{2}=p_{3}=\ldots$. There are infinitely many states, all equally likely, which means that the actual probability of being in any given state is zero in the limit.

We have also assumed that the system has an infinite capacity. If this were not the case, then arriving customers would occasionally encounter a full system, and although they would arrive at the system according to the arrival parameter $\lambda$, they would not be permitted to join the system at rate $\lambda$. Thus, the effective arrival rate would not be constant and would vary in time, according to whether the system is at capacity. For this case, the $\mathrm{p}_{\mathrm{n}}$ formulae remain valid as before. However, if we let N denote the system capacity, the steady-state equation for state N is simply

$$
\mathrm{p}_{\mathrm{N}-1} \lambda=\mathrm{p}_{\mathrm{N}} \mu
$$

and

$$
\mathrm{p}_{\mathrm{N}}=\left(\frac{\lambda}{\mu}\right) \mathrm{p}_{\mathrm{N}-1}=\left(\frac{\lambda}{\mu}\right)^{\mathrm{N}} \mathrm{p}_{0}
$$

just as before. However, there is no state $\mathrm{N}+1$. We now have a finite set of states whose probabilities must sum to 1 :

$$
\mathrm{p}_{0}+\mathrm{p}_{1}+\mathrm{p}_{2}+\ldots+\mathrm{p}_{\mathrm{N}}=1
$$

Then,

$$
\mathrm{p}_{0}\left[1+\left(\frac{\lambda}{\mu}\right)+\left(\frac{\lambda}{\mu}\right)^{2}+\left(\frac{\lambda}{\mu}\right)^{3}+\ldots+\left(\frac{\lambda}{\mu}\right)^{\mathrm{N}}\right]=1
$$

The series in the brackets is a geometric progression that sums to

$$
\frac{1-(\lambda / \mu)^{N+1}}{1-(\lambda / \mu)}
$$

when $\lambda \neq \mu$; therefore,

$$
\frac{p_{0}\left(1-(\lambda / \mu)^{N+1}\right)}{(1-(\lambda / \mu))}=1
$$

or

$$
\mathrm{p}_{0}=\frac{1-\rho}{1-\rho^{\mathrm{N}+1}}
$$

When $\lambda=\mu$, we get $\lambda / \mu=1$ and $\mathrm{p}_{0}=1 /(1+\mathrm{N})$.
Other system measures can be computed as before. It can be shown that, when $\rho \neq 1$ :

$$
\mathrm{L}=\frac{\rho\left[1-(\mathrm{N}+1) \rho^{\mathrm{N}}+\mathrm{N} \rho^{\mathrm{N}+1}\right]}{(1-\rho)\left(1-\rho^{\mathrm{N}+1}\right)}
$$

When $\rho=1, \mathrm{~L}=\mathrm{N} / 2$. Moreover, if a customer arrives when the system is full (with probability $\mathrm{p}_{\mathrm{N}}$ ), the customer will not enter the system. Therefore, the effective arrival rate $\lambda_{\mathrm{e}}$ is

$$
\begin{aligned}
& \lambda_{\mathrm{e}}=\lambda\left(1-\mathrm{p}_{\mathrm{N}}\right) \\
& \mathrm{W}=\frac{\mathrm{L}}{\lambda_{\mathrm{e}}} \\
& \mathrm{~W}_{\mathrm{q}}=\frac{\mathrm{L}_{\mathrm{q}}}{\lambda_{\mathrm{e}}} \\
& \mathrm{~W}=\mathrm{W}_{\mathrm{q}}+\frac{1}{\mu}
\end{aligned}
$$

These formulas are still valid when $\lambda>\mu$. As the arrival rate increases relative to the service rate, the system just loses more customers. It is interesting to note that, even in a saturated system in which the arrival rate is greater than the service rate, there is always still some probability $\mathrm{p}_{0}$ that the system will be empty and the server will experience some idle time.

We have derived performance measures for single-server (M/M/1) systems. For multiple server systems (where $s>1$ ), the actual service rate depends on the number of customers in the system. Obviously, if there is only one customer present, then service is being rendered at rate $\mu$. But if there are two customers present, and $\mathrm{s} \geq 2$, then the system service rate is $2 \mu$. Likewise, if $\mathrm{s}=3$, then the system service rate is $3 \mu$. However, if there are s service facilities, the maximum system service rate is s $\mu$, even if there are more than s customers. This is illustrated by the rate diagram in Figure 7.3.

![](https://cdn.mathpix.com/cropped/2006c750-dc2a-4c1f-9672-416eca5f2ff4-320.jpg?height=167&width=1302&top_left_y=281&top_left_x=224)

**FIGURE 7.3** <br>
Rate diagram for multiple servers $(\mathrm{s}=\mathrm{n})$.

The results obtained from a birth-and-death process model for an M/M/s system, for $s>1$, differ from our previous results because service rates are sensitive to the current customer load. Under the assumption that $\rho<1$, that is $\lambda<\mathrm{s} \mu$, it can be shown that:

$$
p_{0}=\frac{1}{\sum_{n=0}^{s-1} \frac{(\lambda / \mu)^{n}}{n!}+\frac{(\lambda / \mu)^{s}}{s!} \frac{1}{1-(\lambda / s \mu)}}
$$

and

$$
p_{n}= \begin{cases}\frac{(\lambda / \mu)^{n} p_{0}}{n!} & \text { for } 0 \leq n \leq s \\ \frac{(\lambda / \mu)^{n} p_{0}}{s!s^{n-s}} & \text { for } n>s\end{cases}
$$

The expected number in queue can be shown to be:

$$
\mathrm{L}_{\mathrm{q}}=\frac{\mathrm{p}_{0}(\lambda / \mu)^{\mathrm{s}} \rho}{\mathrm{~s}!(1-\rho)^{2}}
$$

where $\rho=\lambda / \mathrm{s} \mu$ and, as before,

$$
\begin{aligned}
& \mathrm{L}_{\mathrm{q}}=\lambda \mathrm{W}_{\mathrm{q}} \\
& \mathrm{~W}=\mathrm{W}_{\mathrm{q}}+\frac{1}{\mu} \\
& \mathrm{~L}=\lambda \mathrm{W}=\mathrm{L}_{\mathrm{q}}+\frac{\lambda}{\mu}
\end{aligned}
$$

```python
# ── Section 7.3.2 — Interactive M/M/1 Queueing Model ────────────────
# Uses pyqueueing library + algebraic formulas
# Example 7.3 default: λ = 10 jobs/min, μ = 12 jobs/min

import sys
import subprocess
import importlib.util

# Install pyqueueing if needed
if importlib.util.find_spec("pyqueueing") is None:
    print("Installing pyqueueing...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "pyqueueing", "--quiet"])

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

try:
    from pyqueueing import MM1
    PYQUEUEING_AVAILABLE = True
except Exception:
    PYQUEUEING_AVAILABLE = False


# -------------------------------------------------------------------
# Interactive widgets
# -------------------------------------------------------------------

style = {'description_width': '160px'}
layout = widgets.Layout(width='460px')

lambda_slider = widgets.FloatSlider(
    value=10,
    min=0.1,
    max=30,
    step=0.1,
    description='Arrival rate λ:',
    continuous_update=False,
    style=style,
    layout=layout
)

mu_slider = widgets.FloatSlider(
    value=12,
    min=0.1,
    max=30,
    step=0.1,
    description='Service rate μ:',
    continuous_update=False,
    style=style,
    layout=layout
)

k_slider = widgets.IntSlider(
    value=4,
    min=1,
    max=20,
    step=1,
    description='P(N ≥ k), k:',
    continuous_update=False,
    style=style,
    layout=layout
)

nmax_slider = widgets.IntSlider(
    value=15,
    min=5,
    max=40,
    step=1,
    description='Max state n:',
    continuous_update=False,
    style=style,
    layout=layout
)

show_formula_checkbox = widgets.Checkbox(
    value=True,
    description='Show formulas and explanation'
)

out = widgets.Output()


# -------------------------------------------------------------------
# Main function
# -------------------------------------------------------------------

def run_mm1_interactive(lam, mu, k, nmax, show_formulas):
    with out:
        clear_output(wait=True)

        print("=== Section 7.3.2 — Interactive M/M/1 Queueing Model ===")
        print()
        print("M/M/1 means:")
        print("  M = Markovian/exponential interarrival times")
        print("  M = Markovian/exponential service times")
        print("  1 = one server")
        print()

        print(f"λ = {lam:.2f} arrivals per time unit")
        print(f"μ = {mu:.2f} services per time unit")
        print()

        # Stability check
        if lam >= mu:
            print("⚠️ System is unstable because λ ≥ μ.")
            print()
            print("This means arrivals are coming as fast as or faster than the server can serve them.")
            print("In steady state, L, Lq, W, and Wq do not have finite values.")
            print("Try making μ larger than λ.")
            return

        # Algebraic formulas
        rho = lam / mu
        p0 = 1 - rho
        L = rho / (1 - rho)
        Lq = rho**2 / (1 - rho)
        W = L / lam
        Wq = Lq / lam
        pk_or_more = rho**k

        # Use pyqueueing if available
        if PYQUEUEING_AVAILABLE:
            try:
                q = MM1(lam, mu)

                py_values = {
                    'ρ Utilization': q.utilization(),
                    'p0 Probability system empty': q.prob_n(0),
                    'L Average number in system': q.mean_system_size(),
                    'Lq Average number in queue': q.mean_queue_length(),
                    'W Average time in system': q.mean_system_time(),
                    'Wq Average waiting time': q.mean_wait()
                }

            except Exception:
                py_values = None
        else:
            py_values = None

        # Results table
        rows = [
            ['ρ Utilization', rho, py_values['ρ Utilization'] if py_values else np.nan],
            ['p0 Probability system empty', p0, py_values['p0 Probability system empty'] if py_values else np.nan],
            ['L Average number in system', L, py_values['L Average number in system'] if py_values else np.nan],
            ['Lq Average number in queue', Lq, py_values['Lq Average number in queue'] if py_values else np.nan],
            ['W Average time in system', W, py_values['W Average time in system'] if py_values else np.nan],
            ['Wq Average waiting time', Wq, py_values['Wq Average waiting time'] if py_values else np.nan],
            [f'P(N ≥ {k})', pk_or_more, np.nan]
        ]

        results_df = pd.DataFrame(rows, columns=['Measure', 'Algebraic Formula', 'pyqueueing'])

        if py_values:
            results_df['Match'] = results_df.apply(
                lambda r: '✓' if pd.notna(r['pyqueueing']) and abs(r['Algebraic Formula'] - r['pyqueueing']) < 1e-7 else '',
                axis=1
            )
        else:
            results_df['Match'] = ''

        print("=== Performance Measures ===")
        display(results_df)

        if show_formulas:
            print()
            print("=== Formulas Used ===")
            print(f"ρ  = λ / μ = {lam:.2f} / {mu:.2f} = {rho:.4f}")
            print(f"p0 = 1 - ρ = 1 - {rho:.4f} = {p0:.4f}")
            print(f"L  = ρ / (1 - ρ) = {L:.4f}")
            print(f"Lq = ρ² / (1 - ρ) = {Lq:.4f}")
            print(f"W  = L / λ = {L:.4f} / {lam:.2f} = {W:.4f}")
            print(f"Wq = Lq / λ = {Lq:.4f} / {lam:.2f} = {Wq:.4f}")
            print(f"P(N ≥ {k}) = ρ^{k} = {rho:.4f}^{k} = {pk_or_more:.4f}")
            print()
            print("Little's Law check:")
            print(f"L  = λW  → {L:.4f} = {lam:.2f} × {W:.4f} = {lam * W:.4f}")
            print(f"Lq = λWq → {Lq:.4f} = {lam:.2f} × {Wq:.4f} = {lam * Wq:.4f}")

        # State probabilities
        ns = np.arange(0, nmax + 1)
        pn = (1 - rho) * rho**ns

        prob_df = pd.DataFrame({
            'n customers in system': ns,
            'p_n = (1-ρ)ρ^n': pn
        })

        print()
        print("=== State Probabilities ===")
        display(prob_df)

        # ----------------------------------------------------------------
        # Plot 1: State probability distribution
        # ----------------------------------------------------------------
        plt.figure(figsize=(9, 4))
        plt.bar(ns, pn, edgecolor='black', alpha=0.75)
        plt.title(f'M/M/1 State Probabilities\nλ={lam:.2f}, μ={mu:.2f}, ρ={rho:.3f}')
        plt.xlabel('n customers in system')
        plt.ylabel('Probability p_n')
        plt.grid(True, axis='y', alpha=0.3)
        plt.show()

        # ----------------------------------------------------------------
        # Plot 2: L and Lq vs utilization
        # ----------------------------------------------------------------
        rho_values = np.linspace(0.01, 0.99, 300)
        L_values = rho_values / (1 - rho_values)
        Lq_values = rho_values**2 / (1 - rho_values)

        plt.figure(figsize=(9, 4))
        plt.plot(rho_values, L_values, linewidth=2, label='L: number in system')
        plt.plot(rho_values, Lq_values, linewidth=2, linestyle='--', label='Lq: number in queue')
        plt.axvline(rho, linestyle=':', linewidth=2, label=f'Current ρ={rho:.3f}')
        plt.scatter([rho], [L], s=80)
        plt.scatter([rho], [Lq], s=80)
        plt.title('Effect of Utilization on Queue Size')
        plt.xlabel('ρ = λ / μ')
        plt.ylabel('Expected number of customers')
        plt.ylim(0, min(30, max(10, L * 1.5)))
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()

        # ----------------------------------------------------------------
        # Plot 3: W and Wq vs utilization
        # ----------------------------------------------------------------
        W_values = 1 / (mu * (1 - rho_values))
        Wq_values = rho_values / (mu * (1 - rho_values))

        plt.figure(figsize=(9, 4))
        plt.plot(rho_values, W_values, linewidth=2, label='W: time in system')
        plt.plot(rho_values, Wq_values, linewidth=2, linestyle='--', label='Wq: waiting time')
        plt.axvline(rho, linestyle=':', linewidth=2, label=f'Current ρ={rho:.3f}')
        plt.scatter([rho], [W], s=80)
        plt.scatter([rho], [Wq], s=80)
        plt.title('Effect of Utilization on Waiting Time')
        plt.xlabel('ρ = λ / μ')
        plt.ylabel('Expected time')
        plt.ylim(0, min(5, max(1, W * 1.5)))
        plt.legend()
        plt.grid(True, alpha=0.3)
        plt.show()

        # Interpretation
        print()
        print("=== Interpretation ===")
        print(f"The server is busy {100*rho:.1f}% of the time.")

        if rho < 0.5:
            print("The system is lightly loaded. Waiting times should be small.")
        elif rho < 0.8:
            print("The system is moderately loaded. Waiting exists, but the system is still comfortable.")
        elif rho < 0.95:
            print("The system is heavily loaded. Small increases in λ can create much larger queues.")
        else:
            print("The system is very close to instability. Queues can grow very quickly.")

        print()
        print("Important idea:")
        print("As ρ approaches 1, the denominator (1 - ρ) becomes very small.")
        print("That is why L, Lq, W, and Wq grow very fast near ρ = 1.")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

def update(change=None):
    run_mm1_interactive(
        lambda_slider.value,
        mu_slider.value,
        k_slider.value,
        nmax_slider.value,
        show_formula_checkbox.value
    )


for widget in [
    lambda_slider,
    mu_slider,
    k_slider,
    nmax_slider,
    show_formula_checkbox
]:
    widget.observe(update, names='value')


display(widgets.VBox([
    widgets.HTML('<h3>Section 7.3.2 — Interactive Steady-State Performance Measures</h3>'),
    widgets.HTML('Change λ and μ to see how utilization, queue length, waiting time, and state probabilities change.'),
    lambda_slider,
    mu_slider,
    k_slider,
    nmax_slider,
    show_formula_checkbox
]))

display(out)

update()
```
