# 4.7.6 A Customer Allocation Problem

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/drive/1BCVmz0DA2VrFZeYFHKz78d6Ngc96fleJ?usp=sharing)

We will illustrate the basic method of Lagrangian relaxation by solving a distribution problem. Many companies operate multiple distribution warehouses to supply products to their customers. One of the common problems facing such companies is to determine which set of customers should be assigned to each warehouse. Because of the additional delivery costs, it usually does not make economic sense to have a customer's demand satisfied by more than one warehouse. This is referred to as a single sourcing constraint.

Consider a delivery problem in which four customers must be served from three warehouses. The cost of satisfying each customer from each warehouse is illustrated in the following table. Each customer has a demand that must be met, and each warehouse has a maximum capacity.

![alt text](image.png)

The problem can be formulated as a generalized assignment problem where $\mathrm{x}_{\mathrm{ij}}=1$ if customer i is served by warehouse j . Every customer must be served by exactly one warehouse, and every warehouse has a capacity constraint on the set of customers it can service.

$$
\begin{array}{lll}
\operatorname{minimize} & \sum_{\mathrm{i}} \sum_{\mathrm{j}} \mathrm{c}_{\mathrm{ij}} \mathrm{x}_{\mathrm{ij}} & \\
\text { subject to: } & \sum_{\mathrm{j}} \mathrm{x}_{\mathrm{ij}}=1 & \text { for all customers } \mathrm{i} \\
& \sum_{\mathrm{i}} \mathrm{~d}_{\mathrm{i}} \mathrm{x}_{\mathrm{ij}} \leq \mathrm{b}_{\mathrm{j}} & \text { for all warehouses } \mathrm{j} \\
& \mathrm{x}_{\mathrm{ij}}=0 \text { or } 1 &
\end{array}
$$

If we solve the problem as an LP with ( $0 \leq x_{i} \leq 1$ ), we get a total cost of 890 , but two of the customers are served from two warehouses. This violates the $0-1$ constraint on the variables.
We construct a Lagrangian function by penalizing the customer constraints:

$$
\begin{aligned}
& \mathrm{L}(\mathrm{x}, \mathrm{u})=\operatorname{maximum}_{\mathrm{u}} \operatorname{minimum}_{\mathrm{x}=\{0,1\}} \sum_{\mathrm{i}} \sum_{\mathrm{j}} \mathrm{c}_{\mathrm{ij}} \mathrm{x}_{\mathrm{ij}}+\sum_{\mathrm{i}} \mathrm{u}_{\mathrm{i}}\left[\sum_{\mathrm{j}} \mathrm{x}_{\mathrm{ij}}-1\right] \\
& \text { subject to: } \quad \sum_{\mathrm{i}} \mathrm{~d}_{\mathrm{i}} \mathrm{x}_{\mathrm{ij}} \leq \mathrm{b}_{\mathrm{j}} \quad \text { for all warehouses } \mathrm{j}
\end{aligned}
$$

or, equivalently:

$$
\begin{aligned}
& \mathrm{L}(\mathrm{x}, \mathrm{u})=\operatorname{maximum}_{\mathrm{u}} \operatorname{minimum}_{\mathrm{x}=\{0,1\}} \sum_{\mathrm{i}} \sum_{\mathrm{j}}\left[\mathrm{c}_{\mathrm{ij}}+\mathrm{u}_{\mathrm{i}}\right] \mathrm{x}_{\mathrm{ij}}-\sum_{\mathrm{i}} \mathrm{u}_{\mathrm{i}} \\
& \text { subject to: } \quad \sum_{\mathrm{i}} \mathrm{~d}_{\mathrm{i}} \mathrm{x}_{\mathrm{ij}} \leq \mathrm{b}_{\mathrm{j}} \quad \text { for all warehouses } \mathrm{j}
\end{aligned}
$$

Observe that because this problem is a minimization in x , we construct the Lagrangian as a maximization in u . When we substitute any fixed value of u into the Lagrangian, this problem becomes a simple multiple knapsack problem. We can treat each warehouse as an independent knapsack problem, and find the least expensive customers for that warehouse. However, because we have dropped the customer constraint, there is no reason why a customer cannot be assigned to more than one warehouse, or in fact, to no warehouse. In particular, if we set $\mathrm{u}=0$ initially, we discover that the optimal solution is $\mathrm{x}=0$. (No customers are assigned to any warehouse!) To make customers attractive to the warehouses, at least some of the costs must be negative; that is, we must choose initial values for the u vector to be negative enough to make some of the Lagrangian costs negative. We will choose $\mathrm{u}=(-475,-375,-360,-360)$, a somewhat arbitrary choice, but one in which the new Lagrangian costs have at least one negative cost for every customer. We can subtract
the second smallest cost in each row to ensure that the smallest cost will be negative. (In the first row, we can subtract 475 from each element.) Now, every customer is desired by at least one warehouse. The new Lagrangian costs are:

![alt text](image-1.png)

When we solve the knapsack problem for each warehouse we find:
Warehouse 1: Does not take any customers (all costs are zero).
Warehouse 2: Would like to take all of them, but can take only customers 2 and 3 due to capacity constraints for a cost of -405 .
Warehouse 3: Takes customer 3 for a cost of -180.

The value of the Lagrangian function is the sum of these costs minus the sum of the penalties, $\mathrm{u}_{\mathrm{i}}: 0-405-180-(-1570)=985$. This first approximation is already a better bound on the solution than the LP solution, which has a value of 890 .
When we now examine the customer constraints, we see that no warehouse took customer 1 or 4, and two warehouses took customer 3. To encourage at least one warehouse to take customers 1 and 4, we want to decrease the cost for those customers (that is, decrease $\mathrm{u}_{1}$ and $\mathrm{u}_{4}$ ).
In order to decrease the number of warehouses that want customer 3, we increase the cost slightly. There are many popular methods for doing this, but they all essentially involve trial and error. We can change all three $\mathrm{u}_{\mathrm{i}}$ values at once, or we can change them one at a time. We can take small steps, and keep increasing them until the Lagrangian stops increasing, or we can take large steps (too far) and then back up. Without elaborating on details, we will briefly illustrate the first couple of steps.
Suppose we decide to change the three $\mathrm{u}_{\mathrm{i}}$ values by 200 . (A small change by 1 or 2 does in fact increase the Lagrangian.) Then, the new $u$ values are ( $-675,-375,-160,-560$ ) and the costs will be:

![alt text](image-2.png)

The three knapsack problem solutions are:
Warehouse 1: Takes customer 4 (customer 1 will not fit) for a cost of -200 .
Warehouse 2: Takes customer 1 for a cost of -580.
Warehouse 3: Takes customer 4 for a cost of -200.
The value of the Lagrangian is: $-200-580-200-(-1770)=790$. We thought that we were moving in a direction that increased the Lagrangian; and, in fact, the Lagrangian will increase for the fixed previous value of x . Unfortunately, as we continue to increase the change in $u$, we eventually get a new minimum solution $x$, and the Lagrangian starts to decrease. Apparently, we have gone too far; so let us try again, using a smaller change for the values of $u$ by 10 . The new $u$ vector is: $(-485,-375,-350,-370)$, and the resulting cost matrix is:

![alt text](image-3.png)

The knapsack solutions are:
Warehouse 1: Takes customer 4 for a cost of -10 .
Warehouse 2: Takes customers 2 and 3 for a cost of -395.
Warehouse 3: Takes customer 3 for a cost of -170.
The Lagrangian function is: $-10-395-170-(-1580)=1005$. At this stage, customer 1 is still unserved, and customer 3 is still served by two warehouses. Decreasing $\mathrm{u}_{1}$ further and increasing $\mathrm{u}_{3}$ should lead to a further increase in the Lagrangian.
In fact, the value of the optimal solution to the Lagrangian for this problem is 1,355 , which also happens to be the optimal integer function value (with customer 4 assigned to warehouse 1; customers 2 and 3 to warehouse 2, and customer 1 to warehouse 3). Thus, for this particular example problem, the Lagrangian bound is tight.

**Code Implementation**

```python
# ── Chapter 4.7.6 — Interactive Customer Allocation Problem ─────────────
# Same notebook idea:
# 1) Direct customer allocation solution
# 2) Lagrangian relaxation by relaxing single-sourcing constraints
# 3) Modified costs cij + ui
# 4) Solve independent warehouse knapsack subproblems
# 5) Compute Lagrangian value / lower bound

import itertools
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets

from IPython.display import display, clear_output

# -------------------------------------------------------------------
# Textbook data — Section 4.7.6
# -------------------------------------------------------------------

cost = np.array([
    [475,  95, 665],
    [375, 150, 375],
    [360, 180, 180],
    [360, 180, 360]
], dtype=float)

demand = np.array([19, 15, 12, 18], dtype=float)
capacity = np.array([18, 27, 20], dtype=float)

n_cust, n_wh = cost.shape

customers = [f"Customer {i+1}" for i in range(n_cust)]
warehouses = [f"Warehouse {j+1}" for j in range(n_wh)]

# Textbook u vectors from the notebook
u_presets = {
    "Textbook Iteration 1": np.array([-475., -375., -360., -360.]),
    "Textbook Iteration 2": np.array([-675., -375., -160., -560.]),
    "Textbook Iteration 3": np.array([-485., -375., -350., -370.]),
    "Custom": np.array([-475., -375., -360., -360.])
}

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

style = {'description_width': '170px'}
layout = widgets.Layout(width='480px')

preset_dropdown = widgets.Dropdown(
    options=list(u_presets.keys()),
    value="Textbook Iteration 1",
    description="u preset:",
    style=style,
    layout=layout
)

u1_slider = widgets.FloatSlider(
    value=-475,
    min=-900,
    max=100,
    step=5,
    description='u1:',
    continuous_update=False,
    style=style,
    layout=layout
)

u2_slider = widgets.FloatSlider(
    value=-375,
    min=-900,
    max=100,
    step=5,
    description='u2:',
    continuous_update=False,
    style=style,
    layout=layout
)

u3_slider = widgets.FloatSlider(
    value=-360,
    min=-900,
    max=100,
    step=5,
    description='u3:',
    continuous_update=False,
    style=style,
    layout=layout
)

u4_slider = widgets.FloatSlider(
    value=-360,
    min=-900,
    max=100,
    step=5,
    description='u4:',
    continuous_update=False,
    style=style,
    layout=layout
)

show_modified_costs = widgets.Checkbox(
    value=True,
    description='Show modified cost matrix'
)

show_assignment_matrices = widgets.Checkbox(
    value=True,
    description='Show assignment matrices'
)

show_bound_chart = widgets.Checkbox(
    value=True,
    description='Show bound comparison chart'
)

show_preset_comparison = widgets.Checkbox(
    value=True,
    description='Compare textbook iterations'
)

out = widgets.Output()

# -------------------------------------------------------------------
# Helper functions
# -------------------------------------------------------------------

def get_current_u():
    """
    Return selected u vector.
    If a textbook preset is chosen, use that preset.
    If Custom is chosen, use slider values.
    """

    if preset_dropdown.value == "Custom":
        return np.array([
            u1_slider.value,
            u2_slider.value,
            u3_slider.value,
            u4_slider.value
        ], dtype=float)

    return u_presets[preset_dropdown.value].copy()


def direct_customer_allocation():
    """
    Solve the original customer allocation problem exactly by enumeration.

    Each customer must be assigned to exactly one warehouse.
    There are only 3^4 = 81 possible assignments, so enumeration is exact and simple.
    """

    best_cost = np.inf
    best_assignment = None
    best_capacity_used = None

    rows = []

    for assignment in itertools.product(range(n_wh), repeat=n_cust):
        capacity_used = np.zeros(n_wh)
        total_cost = 0

        for i, j in enumerate(assignment):
            capacity_used[j] += demand[i]
            total_cost += cost[i, j]

        feasible = np.all(capacity_used <= capacity + 1e-9)

        rows.append({
            "Assignment": tuple(j + 1 for j in assignment),
            "Total Cost": total_cost,
            "W1 Used": capacity_used[0],
            "W2 Used": capacity_used[1],
            "W3 Used": capacity_used[2],
            "Feasible": feasible
        })

        if feasible and total_cost < best_cost:
            best_cost = total_cost
            best_assignment = assignment
            best_capacity_used = capacity_used

    all_assignments_df = pd.DataFrame(rows)

    return best_cost, best_assignment, best_capacity_used, all_assignments_df


def solve_warehouse_knapsack(j, modified_cost):
    """
    Solve one warehouse subproblem exactly by enumeration.

    For warehouse j:
    choose any subset of customers assigned to this warehouse
    subject to capacity[j]
    minimize sum of modified costs.
    """

    best_value = np.inf
    best_subset = []
    best_weight = 0

    rows = []

    for mask in range(2 ** n_cust):
        subset = [i for i in range(n_cust) if (mask >> i) & 1]
        total_demand = sum(demand[i] for i in subset)
        total_modified_cost = sum(modified_cost[i, j] for i in subset)

        feasible = total_demand <= capacity[j] + 1e-9

        rows.append({
            "Subset": tuple(i + 1 for i in subset),
            "Modified Cost": total_modified_cost,
            "Demand Used": total_demand,
            "Feasible": feasible
        })

        if feasible and total_modified_cost < best_value:
            best_value = total_modified_cost
            best_subset = subset
            best_weight = total_demand

    return best_subset, best_value, best_weight, pd.DataFrame(rows)


def solve_lagrangian(u):
    """
    Lagrangian relaxation:

    Original assignment constraint:
        sum_j xij = 1

    Relaxed using multipliers ui:
        L(x,u) = sum_i sum_j cij xij + sum_i ui(sum_j xij - 1)

    This becomes:
        sum_i sum_j (cij + ui)xij - sum_i ui

    Each warehouse becomes an independent knapsack subproblem.
    """

    modified_cost = cost + u[:, np.newaxis]

    warehouse_results = []
    lagrangian_assignment = np.zeros((n_cust, n_wh))

    total_subproblem_value = 0

    for j in range(n_wh):
        subset, obj_value, demand_used, subset_table = solve_warehouse_knapsack(j, modified_cost)

        for i in subset:
            lagrangian_assignment[i, j] = 1

        total_subproblem_value += obj_value

        warehouse_results.append({
            "Warehouse": j + 1,
            "Selected Customers": [i + 1 for i in subset],
            "Capacity Used": demand_used,
            "Capacity": capacity[j],
            "Subproblem Value": obj_value,
            "Subset Table": subset_table
        })

    lagrangian_value = total_subproblem_value - np.sum(u)

    return modified_cost, warehouse_results, lagrangian_assignment, total_subproblem_value, lagrangian_value


def assignment_matrix_from_direct(assignment):
    matrix = np.zeros((n_cust, n_wh))

    for i, j in enumerate(assignment):
        matrix[i, j] = 1

    return matrix


def check_lagrangian_assignment(matrix):
    """
    Check whether the Lagrangian solution satisfies original assignment constraints.
    In Lagrangian relaxation, customers may be assigned to zero, one, or multiple warehouses.
    """

    row_sums = matrix.sum(axis=1)
    col_capacity_used = matrix.T @ demand

    rows = []

    for i in range(n_cust):
        if row_sums[i] == 1:
            status = "OK"
        elif row_sums[i] == 0:
            status = "Not assigned"
        else:
            status = "Assigned multiple times"

        rows.append({
            "Customer": i + 1,
            "Number of assigned warehouses": int(row_sums[i]),
            "Status": status
        })

    customer_check_df = pd.DataFrame(rows)

    capacity_check_df = pd.DataFrame({
        "Warehouse": [1, 2, 3],
        "Capacity Used": col_capacity_used,
        "Capacity": capacity,
        "Capacity OK?": col_capacity_used <= capacity + 1e-9
    })

    original_feasible = (
        np.all(row_sums == 1)
        and np.all(col_capacity_used <= capacity + 1e-9)
    )

    return customer_check_df, capacity_check_df, original_feasible


def matrix_df(matrix):
    return pd.DataFrame(
        matrix.astype(int),
        index=customers,
        columns=warehouses
    )


def draw_assignment_heatmaps(direct_matrix, lagrangian_matrix):
    """
    Show direct IP assignment and Lagrangian relaxed assignment side-by-side.
    """

    fig, axes = plt.subplots(1, 2, figsize=(10, 4.2))

    matrices = [
        (direct_matrix, "Direct 0-1 IP Assignment"),
        (lagrangian_matrix, "Lagrangian Relaxed Assignment")
    ]

    for ax, (mat, title) in zip(axes, matrices):
        ax.imshow(mat, aspect="auto")

        ax.set_title(title, fontsize=11, pad=12)
        ax.set_xticks(range(n_wh))
        ax.set_yticks(range(n_cust))
        ax.set_xticklabels([f"W{j+1}" for j in range(n_wh)])
        ax.set_yticklabels([f"C{i+1}" for i in range(n_cust)])

        for i in range(n_cust):
            for j in range(n_wh):
                ax.text(
                    j,
                    i,
                    int(mat[i, j]),
                    ha="center",
                    va="center",
                    fontsize=12,
                    fontweight="bold"
                )

    plt.tight_layout()
    plt.show()


def compute_preset_bounds():
    rows = []

    for name, u in u_presets.items():
        if name == "Custom":
            continue

        _, _, _, sub_value, lag_value = solve_lagrangian(u)

        rows.append({
            "Iteration": name,
            "u1": u[0],
            "u2": u[1],
            "u3": u[2],
            "u4": u[3],
            "Subproblem Sum": sub_value,
            "sum(u)": np.sum(u),
            "Lagrangian Bound": lag_value
        })

    return pd.DataFrame(rows)


def draw_bound_chart(selected_name, lag_value, direct_cost, preset_df):
    """
    Compare the selected Lagrangian bound with the direct optimal cost.
    For this minimization problem, the Lagrangian value is a lower bound.
    """

    labels = list(preset_df["Iteration"])
    values = list(preset_df["Lagrangian Bound"])

    if selected_name == "Custom":
        labels.append("Custom u")
        values.append(lag_value)

    labels.append("Direct IP optimum")
    values.append(direct_cost)

    plt.figure(figsize=(9, 4.8))
    plt.bar(labels, values)

    for idx, value in enumerate(values):
        plt.text(
            idx,
            value + 20,
            f"{value:.0f}",
            ha="center",
            fontsize=9
        )

    plt.title("Lagrangian Bounds Compared with Direct IP Optimum", fontsize=13, pad=16)
    plt.ylabel("Cost / bound")
    plt.xticks(rotation=20, ha="right")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()


def update_sliders_for_preset(change=None):
    """
    When choosing a preset, move sliders to that u vector.
    This makes the sliders match the displayed vector.
    """

    if preset_dropdown.value != "Custom":
        u = u_presets[preset_dropdown.value]

        u1_slider.value = u[0]
        u2_slider.value = u[1]
        u3_slider.value = u[2]
        u4_slider.value = u[3]


# -------------------------------------------------------------------
# Main update function
# -------------------------------------------------------------------

def update(change=None):
    with out:
        clear_output(wait=True)

        u = get_current_u()

        direct_cost, direct_assignment, direct_capacity_used, all_assignments_df = direct_customer_allocation()

        modified_cost, warehouse_results, lag_matrix, subproblem_sum, lag_value = solve_lagrangian(u)

        direct_matrix = assignment_matrix_from_direct(direct_assignment)

        customer_check_df, capacity_check_df, lag_original_feasible = check_lagrangian_assignment(lag_matrix)

        print("=== Chapter 4.7.6 — Interactive Customer Allocation Problem ===")
        print()
        print("Problem:")
        print("Assign each customer to exactly one warehouse while respecting warehouse capacities.")
        print("The goal is to minimize total delivery cost.")
        print()

        print("Cost matrix cij:")
        display(pd.DataFrame(cost.astype(int), index=customers, columns=warehouses))

        print("Demand and capacity:")
        demand_capacity_df = pd.DataFrame({
            "Customer": customers,
            "Demand": demand.astype(int)
        })
        display(demand_capacity_df)

        capacity_df = pd.DataFrame({
            "Warehouse": warehouses,
            "Capacity": capacity.astype(int)
        })
        display(capacity_df)

        # Direct solution
        print()
        print("=" * 60)
        print("Direct 0-1 IP Solution")
        print("=" * 60)

        print(f"Optimal cost = {int(direct_cost)}")
        print("Assignment:")

        for i, j in enumerate(direct_assignment):
            print(
                f"  Customer {i+1} → Warehouse {j+1} "
                f"(cost={int(cost[i, j])}, demand={int(demand[i])})"
            )

        print()
        print("Warehouse capacity usage:")
        display(pd.DataFrame({
            "Warehouse": warehouses,
            "Used": direct_capacity_used.astype(int),
            "Capacity": capacity.astype(int),
            "Slack": (capacity - direct_capacity_used).astype(int)
        }))

        # Lagrangian explanation
        print()
        print("=" * 60)
        print("Lagrangian Relaxation")
        print("=" * 60)

        print("Relax the single-sourcing constraints:")
        print("  sum_j xij = 1")
        print()
        print("Lagrangian objective:")
        print("  L(x,u) = sum_i sum_j cij xij + sum_i ui(sum_j xij - 1)")
        print("         = sum_i sum_j (cij + ui)xij - sum_i ui")
        print()
        print("Selected u vector:")
        display(pd.DataFrame([u], columns=["u1", "u2", "u3", "u4"], index=[preset_dropdown.value]))

        if show_modified_costs.value:
            print()
            print("Modified cost matrix cij + ui:")
            display(pd.DataFrame(
                modified_cost.astype(int),
                index=customers,
                columns=warehouses
            ))

        # Warehouse subproblems
        print()
        print("=" * 60)
        print("Independent Warehouse Subproblems")
        print("=" * 60)

        summary_rows = []

        for result in warehouse_results:
            summary_rows.append({
                "Warehouse": result["Warehouse"],
                "Selected Customers": result["Selected Customers"],
                "Capacity Used": result["Capacity Used"],
                "Capacity": result["Capacity"],
                "Subproblem Value": result["Subproblem Value"]
            })

        subproblem_df = pd.DataFrame(summary_rows)
        display(subproblem_df)

        print()
        print("Lagrangian value:")
        print(f"  Subproblem sum = {subproblem_sum:.0f}")
        print(f"  sum(u)         = {np.sum(u):.0f}")
        print(f"  L(u)           = {subproblem_sum:.0f} - ({np.sum(u):.0f}) = {lag_value:.0f}")
        print()

        print("Because this is a minimization problem, L(u) is a lower bound on the true optimal cost.")
        print(f"Direct IP optimal cost = {direct_cost:.0f}")
        print(f"Lagrangian bound       = {lag_value:.0f}")
        print(f"Gap from optimum       = {direct_cost - lag_value:.0f}")

        # Feasibility check
        print()
        print("=" * 60)
        print("Does the Lagrangian Solution Satisfy Original Assignment Rules?")
        print("=" * 60)

        print("Customer assignment check:")
        display(customer_check_df)

        print("Warehouse capacity check:")
        display(capacity_check_df)

        if lag_original_feasible:
            print("The Lagrangian solution is feasible for the original problem.")
        else:
            print("The Lagrangian solution is usually NOT feasible for the original problem because the assignment constraints were relaxed.")

        # Assignment matrices
        if show_assignment_matrices.value:
            print()
            print("=== Assignment Matrices ===")
            print("1 means the customer is assigned to that warehouse.")
            print()
            print("Direct 0-1 IP assignment matrix:")
            display(matrix_df(direct_matrix))

            print("Lagrangian relaxed assignment matrix:")
            display(matrix_df(lag_matrix))

            draw_assignment_heatmaps(direct_matrix, lag_matrix)

        # Preset comparison
        preset_df = compute_preset_bounds()

        if show_preset_comparison.value:
            print()
            print("=" * 60)
            print("Textbook Iteration Comparison")
            print("=" * 60)

            display(preset_df)

        if show_bound_chart.value:
            print()
            print("=== Bound Comparison Chart ===")
            draw_bound_chart(preset_dropdown.value, lag_value, direct_cost, preset_df)

        # Textbook verification
        print()
        print("=" * 60)
        print("Textbook Verification")
        print("=" * 60)
        print("Direct 0-1 IP optimal cost should be 1355.")
        print("Optimal assignment:")
        print("  Customer 1 → Warehouse 3")
        print("  Customer 2 → Warehouse 2")
        print("  Customer 3 → Warehouse 2")
        print("  Customer 4 → Warehouse 1")

        print()
        print("Interpretation:")
        print("The Lagrangian relaxation decomposes the allocation problem into separate warehouse knapsack problems.")
        print("This is useful because each warehouse subproblem is easier to solve independently.")
        print("However, because single-sourcing was relaxed, the Lagrangian solution may assign a customer zero times or multiple times.")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

def preset_changed(change=None):
    update_sliders_for_preset()
    update()

preset_dropdown.observe(preset_changed, names="value")

for w in [
    u1_slider,
    u2_slider,
    u3_slider,
    u4_slider,
    show_modified_costs,
    show_assignment_matrices,
    show_bound_chart,
    show_preset_comparison
]:
    w.observe(update, names="value")


# -------------------------------------------------------------------
# Display
# -------------------------------------------------------------------

display(widgets.VBox([
    widgets.HTML("<h3>Chapter 4.7.6 — Interactive Customer Allocation Problem</h3>"),
    widgets.HTML("Choose a textbook u-vector or use custom multipliers to see how the Lagrangian relaxation changes."),

    preset_dropdown,

    widgets.HTML("<h4>Custom u values</h4>"),
    u1_slider,
    u2_slider,
    u3_slider,
    u4_slider,

    widgets.HTML("<h4>Display options</h4>"),
    show_modified_costs,
    show_assignment_matrices,
    show_preset_comparison,
    show_bound_chart
]))

display(out)

update()
```
