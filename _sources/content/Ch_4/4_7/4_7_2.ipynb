Consider the following example problem, which is illustrated in Figure 4.9:

$$
\begin{array}{ll}
\operatorname{maximize} & \mathrm{z}=\mathrm{x}_{1}+2 \mathrm{x}_{2} \\
\text { subject to } & 2 \mathrm{x}_{1}+\mathrm{x}_{2} \leq 2 \\
& \mathrm{x}_{1}, \mathrm{x}_{2}=0 \text { or } 1
\end{array}
$$

![](https://cdn.mathpix.com/cropped/2006c750-dc2a-4c1f-9672-416eca5f2ff4-211.jpg?height=651&width=612&top_left_y=283&top_left_x=566)

**FIGURE 4.9** <br>
Simple Lagrangian problem.

Observe that, if a $0-1$ problem does not have any constraints, it is trivial to maximize. That is, if the objective function coefficient $\mathrm{c}_{\mathrm{j}}$ is positive, then set $\mathrm{x}_{\mathrm{j}}=1$; otherwise, set $\mathrm{x}_{\mathrm{j}}=0$. We can express the problem in Lagrangian form as:

$$
\underset{\mathrm{u} \geq 0}{\operatorname{minimum}} \underset{\mathrm{x} \in(0,1)}{\operatorname{maximum}} \quad \mathrm{x}_{1}+2 \mathrm{x}_{2}-\mathrm{u}\left(2 \mathrm{x}_{1}+\mathrm{x}_{2}-2\right)
$$

We begin with $\mathrm{u}=0$, and note the maximum of the problem is $\mathrm{L}(\mathrm{x}, 0)=3$ with $\mathrm{x}_{1}, \mathrm{x}_{2}=1$. However, this point violates the constraint, so we substitute these values of x into the Lagrangian, and consider the result as a function only of $u$.

$$
\underset{u \geq 0}{\operatorname{minimum}} 3-u
$$

This function can be minimized by choosing $u$ as large as possible. We could try $u=5$, for example; and when we substitute this value into the original Lagrangian, we get:

$$
\begin{aligned}
& \underset{\quad \underset{x=(0,1)}{\operatorname{maximum}} x_{1}+2 x_{2}-5\left(2 x_{1}+x_{2}-2\right)}{\substack{x=(0,1) \\
=\underset{x}{\operatorname{maximum}}-9 x_{1}-3 x_{2}+10}}
\end{aligned}
$$

The optimal solution to this problem is to set both decision variables to zero. The corresponding function value is $\mathrm{L}(\mathrm{x}, 5)=10$. This time, when we substitute x into the Lagrangian in terms of $u$, we find:

$$
\underset{u \geq 0}{\operatorname{minimum}} 0-u(0-2)=2 u
$$

![](https://cdn.mathpix.com/cropped/2006c750-dc2a-4c1f-9672-416eca5f2ff4-212.jpg?height=675&width=682&top_left_y=281&top_left_x=533)

**FIGURE 4.10** <br>
$\mathrm{L}\left(\mathrm{x}^{*}, \mathrm{u}\right)$ as a function of u .

This subproblem tells us to decrease $u$ as far as possible. We already know however that when $\mathrm{u}=0$, it will tell us to increase u . So, the correct value of u must lie somewhere between these two extremes.

Observe that, for any value of u , we can solve for x and find the value of $\mathrm{L}(\mathrm{x}, \mathrm{u})$. Figure 4.10 illustrates what we have learned so far about $\mathrm{L}(\mathrm{x}, \mathrm{u})$ as a function of u .

Recall that we want to minimize $L(x, u)$ as a function of $u$. When $u=0$, we found $x=(1,1)$ and the function was defined by the decreasing line as $u$ increases. This expression is valid as long as $x=(1,1)$; but then at some point, the optimal solution for $x$ changes, and we get a new linear function describing $\mathrm{L}(\mathrm{x}, \mathrm{u})$. We now know what that linear function is when $\mathrm{u}=0$ and $\mathrm{u}=5$, yet we do not know how it behaves in between these two points. The two line segments in Figure 4.10 represent our best guess at the moment. In particular, it looks as if the minimum value of the Lagrangian will be found when $\mathrm{u}=1$, so we try that next.

Substituting $\mathrm{u}=1$ into the Lagrangian gives:

$$
\begin{aligned}
& \operatorname{maximum} \mathrm{x}_{1}+2 \mathrm{x}_{2}-1\left(2 \mathrm{x}_{1}+\mathrm{x}_{2}-2\right) \\
& \quad \mathrm{x}=(0,1) \\
& =\operatorname{maximum}-\mathrm{x}_{1}+\mathrm{x}_{2}+2 \\
& \quad \mathrm{x}=(0,1)
\end{aligned}
$$

The maximum of this function is $\mathrm{L}(\mathrm{x}, 1)=3$ when $\mathrm{x}=(0,1)$. If we substitute $\mathrm{x}=(0,1)$ into the original function, we get:

$$
\mathrm{L}(\mathrm{x}, \mathrm{u})=2-\mathrm{u}(-1)=2+\mathrm{u}
$$

This new section of the Lagrangian is added to the previous approximation to get the function illustrated in Figure 4.11.

From this function, we obtain a new estimate of the minimum value of $u=0.5$. Once again, we substitute this value into the Lagrangian and solve for x .

![](https://cdn.mathpix.com/cropped/2006c750-dc2a-4c1f-9672-416eca5f2ff4-213.jpg?height=518&width=566&top_left_y=285&top_left_x=593)

**FIGURE 4.11** <br>
$\mathrm{L}\left(\mathrm{x}^{*}, \mathrm{u}\right)$ as a function of u .

$$
\begin{aligned}
& \operatorname{maximum} x_{1}+2 x_{2}-0.5\left(2 x_{1}+x_{2}-2\right) \\
& \quad x=(0,1) \\
& =\underset{x=(0,1)}{\operatorname{maximum}} 1.5 x_{2}+1
\end{aligned}
$$

The maximum of $\mathrm{L}(\mathrm{x}, 0.5)=2.5$ occurs at $\mathrm{x}=(0,1)$ or $\mathrm{x}=(1,1)$. It is easy to verify that this is the true minimum of the Lagrangian. That is, we will not find any new solutions that we have not already described in Figure 4.11.
Let us summarize a number of very useful properties of the Lagrangian, and indicate how we can make use of these properties.

- The Lagrangian method always finds an integer solution, although the solution found is not necessarily feasible.
- If the solution, $x^{\mathrm{I}}$ is feasible, and if the original function, $z^{\mathrm{I}}$ at $x^{\mathrm{I}}$ is equal to the value of the Lagrangian, then $x^{I}$ is optimal for the original integer problem.
- Most important, if $\mathrm{z}^{*}$ is the solution of the relaxed $\mathrm{LP}, \mathrm{L}(\mathrm{x}, \mathrm{u})$ is the optimal solution to the Lagrangian, and $\mathrm{z}^{\mathrm{I}}$ is the (unknown) optimal integer function value, then

$$
\mathrm{z}^{\mathrm{I}} \leq \mathrm{L}(\mathrm{x}, \mathrm{u}) \leq \mathrm{z}^{*}
$$

A proof of these relationships can be found in Fisher (1985).
In other words, the value of the Lagrangian always gives a bound on the optimal integer solution that is at least as good as the LP. Therefore, if we use the Lagrangian instead of the LP in any branch-and-bound algorithm, we may get better results. The LP bound is never better than the bound from the Lagrangian. In our simple example problem, the optimal integer function value $\mathrm{z}^{\mathrm{I}}=2$ when $\mathrm{z}^{\mathrm{I}}=(0,1)$. The LP solution occurs at $\mathrm{x}=(0.5,1)$ with $\mathrm{z}^{*}=2.5$. The LP and the Lagrangian both give the same upper bound.

**Code Implementation**

```python
# ── Chapter 4.7.2 — Interactive Lagrangian Relaxation ─────────────────
# Original notebook example:
# maximize z = x1 + 2x2
# subject to:
#   2x1 + x2 <= 2
#   x1, x2 = 0 or 1
#
# Lagrangian relaxation:
#   L(x,u) = x1 + 2x2 - u(2x1 + x2 - 2)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# -------------------------------------------------------------------
# Fixed textbook problem
# -------------------------------------------------------------------

binary_solutions = [(0, 0), (0, 1), (1, 0), (1, 1)]
textbook_u_values = [0, 0.5, 1, 5]

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

style = {'description_width': '150px'}
layout = widgets.Layout(width='440px')

u_slider = widgets.FloatSlider(
    value=0.5,
    min=0,
    max=5,
    step=0.05,
    description='Multiplier u:',
    continuous_update=False,
    style=style,
    layout=layout
)

u_max_slider = widgets.FloatSlider(
    value=3,
    min=1,
    max=10,
    step=0.5,
    description='Plot max u:',
    continuous_update=False,
    style=style,
    layout=layout
)

show_solution_lines = widgets.Checkbox(
    value=True,
    description='Show individual solution lines'
)

show_textbook_values = widgets.Checkbox(
    value=True,
    description='Show textbook u values'
)

show_all_binary_table = widgets.Checkbox(
    value=True,
    description='Show all 0-1 solutions'
)

show_annotations = widgets.Checkbox(
    value=True,
    description='Show graph annotations'
)

out = widgets.Output()

# -------------------------------------------------------------------
# Core functions
# -------------------------------------------------------------------

def original_objective(x1, x2):
    return x1 + 2*x2


def constraint_lhs(x1, x2):
    return 2*x1 + x2


def is_feasible(x1, x2):
    return constraint_lhs(x1, x2) <= 2


def lagrangian_value(x1, x2, u):
    """
    L(x,u) = x1 + 2x2 - u(2x1 + x2 - 2)
    """
    return x1 + 2*x2 - u * (2*x1 + x2 - 2)


def solve_lagrangian_for_u(u):
    """
    For a fixed u, solve:
    max L(x,u) over x1,x2 in {0,1}
    """

    rows = []

    for x1, x2 in binary_solutions:
        rows.append({
            "x1": x1,
            "x2": x2,
            "Original z": original_objective(x1, x2),
            "2x1 + x2": constraint_lhs(x1, x2),
            "Original feasible?": is_feasible(x1, x2),
            "L(x,u)": lagrangian_value(x1, x2, u)
        })

    df = pd.DataFrame(rows)

    best_index = df["L(x,u)"].idxmax()
    best_row = df.loc[best_index]

    return best_row, df


def all_possible_solutions_table():
    rows = []

    for x1, x2 in binary_solutions:
        rows.append({
            "x1": x1,
            "x2": x2,
            "Original z = x1 + 2x2": original_objective(x1, x2),
            "2x1 + x2": constraint_lhs(x1, x2),
            "Feasible?": is_feasible(x1, x2)
        })

    return pd.DataFrame(rows)


def integer_optimum():
    df = all_possible_solutions_table()
    feasible_df = df[df["Feasible?"] == True]
    best = feasible_df.loc[feasible_df["Original z = x1 + 2x2"].idxmax()]
    return best


def compute_lagrangian_bound_curve(u_max):
    u_values = np.linspace(0, u_max, 600)

    lines = {}

    for x1, x2 in binary_solutions:
        lines[(x1, x2)] = np.array([
            lagrangian_value(x1, x2, u)
            for u in u_values
        ])

    # L(u) = max over all binary solutions
    lagrangian_bound = np.maximum.reduce([
        lines[x] for x in binary_solutions
    ])

    best_index = np.argmin(lagrangian_bound)
    best_u = u_values[best_index]
    best_bound = lagrangian_bound[best_index]

    return u_values, lines, lagrangian_bound, best_u, best_bound


def draw_lagrangian_graph(u_selected, u_max):
    u_values, lines, lagrangian_bound, best_u, best_bound = compute_lagrangian_bound_curve(u_max)

    selected_best, selected_table = solve_lagrangian_for_u(u_selected)
    selected_bound = selected_best["L(x,u)"]

    plt.figure(figsize=(10.5, 6.2))

    # Individual solution lines
    if show_solution_lines.value:
        for x in binary_solutions:
            plt.plot(
                u_values,
                lines[x],
                linestyle="--",
                linewidth=1.4,
                label=f"x = {x}"
            )

    # Lagrangian bound
    plt.plot(
        u_values,
        lagrangian_bound,
        linewidth=3,
        label="Lagrangian bound L(u)"
    )

    # Best bound point
    plt.scatter(
        best_u,
        best_bound,
        s=100,
        zorder=5,
        label=f"Best bound ≈ {best_bound:.2f}"
    )

    # Selected u vertical line
    plt.axvline(
        u_selected,
        linestyle=":",
        linewidth=2,
        label=f"Selected u = {u_selected:.2f}"
    )

    # Selected u point
    plt.scatter(
        u_selected,
        selected_bound,
        s=95,
        zorder=6
    )

    # Clear separated annotations
    if show_annotations.value:

        # Minimum bound label above/right
        plt.annotate(
            f"Minimum bound\nu ≈ {best_u:.2f}\nL(u) ≈ {best_bound:.2f}",
            xy=(best_u, best_bound),
            xytext=(best_u + 0.45, best_bound + 1.05),
            arrowprops=dict(
                arrowstyle="->",
                linewidth=1.2,
                color="black"
            ),
            fontsize=9,
            ha="left",
            va="bottom",
            bbox=dict(
                boxstyle="round,pad=0.35",
                facecolor="white",
                edgecolor="black",
                alpha=0.95
            )
        )

        # Selected u label below/left
        plt.annotate(
            f"Selected u = {u_selected:.2f}\nBest x = ({int(selected_best['x1'])},{int(selected_best['x2'])})\nL = {selected_bound:.2f}",
            xy=(u_selected, selected_bound),
            xytext=(max(0.05, u_selected - 0.48), selected_bound - 1.05),
            arrowprops=dict(
                arrowstyle="->",
                linewidth=1.2,
                color="black"
            ),
            fontsize=9,
            ha="right" if u_selected > 0.6 else "left",
            va="top",
            bbox=dict(
                boxstyle="round,pad=0.35",
                facecolor="white",
                edgecolor="black",
                alpha=0.95
            )
        )

    plt.title("Lagrangian Bound as a Function of u", fontsize=14, pad=18)
    plt.xlabel("u")
    plt.ylabel("L(x*, u)")
    plt.xlim(0, u_max)

    y_max = max(6.5, np.max(lagrangian_bound) + 1.2)
    plt.ylim(0, y_max)

    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8, loc="center right")
    plt.tight_layout()
    plt.show()

    return best_u, best_bound


def textbook_u_table():
    rows = []

    for u in textbook_u_values:
        best_row, _ = solve_lagrangian_for_u(u)

        rows.append({
            "u": u,
            "Best x1": int(best_row["x1"]),
            "Best x2": int(best_row["x2"]),
            "L(x*,u)": best_row["L(x,u)"],
            "Original feasible?": best_row["Original feasible?"]
        })

    return pd.DataFrame(rows)


# -------------------------------------------------------------------
# Main update function
# -------------------------------------------------------------------

def update(change=None):
    with out:
        clear_output(wait=True)

        u = u_slider.value
        u_max = u_max_slider.value

        print("=== Chapter 4.7.2 — Interactive Lagrangian Relaxation ===")
        print()
        print("Original problem:")
        print("Maximize z = x1 + 2x2")
        print("Subject to:")
        print("  2x1 + x2 ≤ 2")
        print("  x1, x2 = 0 or 1")
        print()

        print("Lagrangian relaxation:")
        print("  L(x,u) = x1 + 2x2 - u(2x1 + x2 - 2)")
        print("  u ≥ 0")
        print()

        # All possible 0-1 solutions
        all_df = all_possible_solutions_table()

        if show_all_binary_table.value:
            print("=== All Possible 0-1 Solutions ===")
            display(all_df)

        # Integer optimum
        best_integer = integer_optimum()

        print()
        print("=== Original Integer Optimum ===")
        print(
            f"x = ({int(best_integer['x1'])}, {int(best_integer['x2'])}), "
            f"z = {int(best_integer['Original z = x1 + 2x2'])}"
        )

        # Selected u
        selected_best, selected_table = solve_lagrangian_for_u(u)

        print()
        print(f"=== Lagrangian Subproblem at u = {u:.2f} ===")
        display(selected_table.round(4))

        print()
        print("Best Lagrangian solution at selected u:")
        print(f"x1 = {int(selected_best['x1'])}")
        print(f"x2 = {int(selected_best['x2'])}")
        print(f"L(x*,u) = {selected_best['L(x,u)']:.4f}")

        if selected_best["Original feasible?"]:
            print("This selected Lagrangian solution is feasible for the original problem.")
        else:
            print("This selected Lagrangian solution is NOT feasible for the original problem.")

        # Textbook values
        if show_textbook_values.value:
            print()
            print("=== Textbook u Values ===")
            display(textbook_u_table())

        # Curve
        print()
        print("=== Lagrangian Bound Curve ===")
        best_u, best_bound = draw_lagrangian_graph(u, u_max)

        print()
        print("Best Lagrangian bound from the plotted search:")
        print(f"u ≈ {best_u:.4f}")
        print(f"L(u) ≈ {best_bound:.4f}")

        print()
        print("Textbook comparison:")
        print("Integer optimum = 2")
        print("LP bound = 2.5")
        print("Lagrangian bound = 2.5")

        print()
        print("Interpretation:")
        print("For each value of u, the Lagrangian subproblem chooses the binary solution with the largest L(x,u).")
        print("The upper envelope of these lines gives the Lagrangian bound L(u).")
        print("Because this is a maximization problem, we look for the smallest upper bound over u ≥ 0.")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

for w in [
    u_slider,
    u_max_slider,
    show_solution_lines,
    show_textbook_values,
    show_all_binary_table,
    show_annotations
]:
    w.observe(update, names="value")


# -------------------------------------------------------------------
# Display
# -------------------------------------------------------------------

display(widgets.VBox([
    widgets.HTML("<h3>Chapter 4.7.2 — Interactive Lagrangian Relaxation</h3>"),
    widgets.HTML("Change the multiplier u and see how the Lagrangian solution and bound change."),

    u_slider,
    u_max_slider,
    show_solution_lines,
    show_textbook_values,
    show_all_binary_table,
    show_annotations
]))

display(out)

update()
```
