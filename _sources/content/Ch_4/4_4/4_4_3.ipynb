The manager of an Operations Research department in a large company has a list of projects that she would like to initiate. Each project has an expected payback expressed (in thousands of dollars) as the net present value over a 10-year period. Although all of the projects would be beneficial, there are simply not enough resources (in person days) available this month to do all of them. The estimates of resources and return are:

| Project | $\mathbf{1}$ | $\mathbf{2}$ | $\mathbf{3}$ | $\mathbf{4}$ | $\mathbf{5}$ | $\mathbf{6}$ | $\mathbf{7}$ | $\mathbf{8}$ |
| :--- | :---: | :---: | :---: | :---: | :---: | :---: | :---: | ---: |
| Estimated value | 15 | 20 | 5 | 25 | 22 | 17 | 30 | 4 |
| Days | 51 | 60 | 40 | 62 | 63 | 50 | 70 | 10 |

There are 250 person-days available this month. Which projects should be selected? At the end of this month, the manager must write a report summarizing the results from
completed projects; any projects that are not completed cannot be included among the successful projects in the report.

Define $\mathrm{x}_{\mathrm{j}}=1$ if project j is selected, and 0 otherwise. The "node 0 " problem can be modeled as:

$$
\begin{array}{ll}
\text { maximize } & 15 x_{1}+20 x_{2}+5 x_{3}+25 x_{4}+22 x_{5}+17 x_{6}+30 x_{7}+4 x_{8} \\
\text { subject to } & 51 x_{1}+60 x_{2}+40 x_{3}+62 x_{4}+63 x_{5}+50 x_{6}+70 x_{7}+10 x_{8} \leq 250 \\
& x_{j}=0 \text { or } 1
\end{array}
$$

When the $0-1$ constraints are relaxed to solve the LP, we replace them with the linear constraints: $0 \leq \mathrm{x}_{\mathrm{j}} \leq 1$

**Code Implementation**

```python
# ── Chapter 4.4.3 — Interactive Knapsack Example ─────────────────────
# Same notebook idea:
# 1) Book example data
# 2) LP relaxation
# 3) scipy.milp 0-1 integer solution
# 4) Custom knapsack editing through an interactive table

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
from scipy.optimize import linprog, milp, LinearConstraint, Bounds

# -------------------------------------------------------------------
# Book example data
# -------------------------------------------------------------------

default_values  = np.array([15., 20.,  5., 25., 22., 17., 30.,  4.])
default_weights = np.array([51., 60., 40., 62., 63., 50., 70., 10.])
default_capacity = 250.

n = len(default_values)

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

style = {'description_width': '160px'}
layout = widgets.Layout(width='460px')

capacity_slider = widgets.IntSlider(
    value=250,
    min=50,
    max=400,
    step=10,
    description='Person-days:',
    continuous_update=False,
    style=style,
    layout=layout
)

show_lp_checkbox = widgets.Checkbox(
    value=True,
    description='Show LP relaxation'
)

show_custom_editor = widgets.Checkbox(
    value=False,
    description='Edit values/days manually'
)

show_chart_checkbox = widgets.Checkbox(
    value=True,
    description='Show selected-project chart'
)

# Text boxes for values and weights
value_inputs = []
weight_inputs = []

for i in range(n):
    value_inputs.append(
        widgets.IntText(
            value=int(default_values[i]),
            description=f'P{i+1} value:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='180px')
        )
    )

    weight_inputs.append(
        widgets.IntText(
            value=int(default_weights[i]),
            description=f'P{i+1} days:',
            style={'description_width': '80px'},
            layout=widgets.Layout(width='180px')
        )
    )

update_button = widgets.Button(
    description='Update solution',
    button_style='primary',
    icon='refresh'
)

reset_button = widgets.Button(
    description='Reset textbook data',
    button_style='',
    icon='undo'
)

out = widgets.Output()

# -------------------------------------------------------------------
# Helper functions
# -------------------------------------------------------------------

def get_current_data():
    values = np.array([float(w.value) for w in value_inputs])
    weights = np.array([float(w.value) for w in weight_inputs])
    capacity = float(capacity_slider.value)

    return values, weights, capacity


def solve_lp_relaxation(values, weights, capacity):
    """
    LP relaxation:
    0 <= xi <= 1
    xi can be fractional.
    """

    n = len(values)

    res = linprog(
        -values,
        A_ub=[weights],
        b_ub=[capacity],
        bounds=[(0, 1)] * n,
        method='highs'
    )

    if res.status != 0:
        return None

    return {
        'x': res.x,
        'z': -res.fun,
        'status': res.message
    }


def solve_binary_knapsack(values, weights, capacity):
    """
    0-1 integer knapsack:
    xi must be 0 or 1.
    """

    n = len(values)

    constraints = LinearConstraint([weights], -np.inf, [capacity])
    integrality = np.ones(n)
    bounds = Bounds(lb=np.zeros(n), ub=np.ones(n))

    res = milp(
        -values,
        constraints=constraints,
        integrality=integrality,
        bounds=bounds
    )

    if not res.success:
        return None

    return {
        'x': res.x,
        'z': -res.fun,
        'status': res.message
    }


def create_result_table(values, weights, lp_sol, ip_sol):
    rows = []

    for i in range(len(values)):
        lp_x = np.nan if lp_sol is None else lp_sol['x'][i]
        ip_x = 0 if ip_sol is None else int(round(ip_sol['x'][i]))

        rows.append({
            'Project': i + 1,
            'Value ($K)': int(values[i]),
            'Days': int(weights[i]),
            'LP x': round(lp_x, 3) if lp_sol is not None else '-',
            'LP fractional?': 'Yes' if lp_sol is not None and abs(lp_x - round(lp_x)) > 1e-5 else '-',
            'Selected in IP': 'YES' if ip_x == 1 else '-'
        })

    return pd.DataFrame(rows)


def draw_selected_projects_chart(result_df):
    selected_df = result_df[result_df['Selected in IP'] == 'YES']

    if selected_df.empty:
        print("No projects selected, so no chart to display.")
        return

    plt.figure(figsize=(9, 4.5))

    x_labels = [f"P{p}" for p in selected_df['Project']]

    plt.bar(x_labels, selected_df['Value ($K)'])

    for i, value in enumerate(selected_df['Value ($K)']):
        plt.text(
            i,
            value + 0.5,
            str(value),
            ha='center',
            fontsize=9
        )

    plt.title("Selected Projects in the 0-1 Knapsack Solution", fontsize=13, pad=16)
    plt.xlabel("Project")
    plt.ylabel("Value ($K)")
    plt.grid(True, axis='y', alpha=0.3)
    plt.tight_layout()
    plt.show()


def show_book_verification(values, weights, capacity, ip_sol, lp_sol):
    """
    Show verification only when the data is the original textbook data.
    """

    is_default = (
        np.allclose(values, default_values) and
        np.allclose(weights, default_weights) and
        abs(capacity - default_capacity) < 1e-9
    )

    if not is_default:
        return

    print()
    print("=== Textbook Verification ===")
    print("Book example:")
    print("  Capacity = 250 person-days")
    print("  Textbook integer answer: Projects 4, 5, 6, and 7")
    print("  Total value = 94")
    print("  Total days = 245")

    if ip_sol is not None:
        selected = [i + 1 for i, x in enumerate(ip_sol['x']) if round(x) == 1]
        print()
        print(f"Solver selected projects: {selected}")
        print(f"Solver total value: {int(round(ip_sol['z']))}")

    if lp_sol is not None:
        print()
        print(f"LP relaxation upper bound: {lp_sol['z']:.2f}")
        print("The LP relaxation can contain fractional project choices, which is why the 0-1 integer model is needed.")


# -------------------------------------------------------------------
# Main update function
# -------------------------------------------------------------------

def update(change=None):
    with out:
        clear_output(wait=True)

        values, weights, capacity = get_current_data()

        print("=== Chapter 4.4.3 — Interactive Knapsack Example ===")
        print()
        print("Problem:")
        print("Choose projects to maximize total value.")
        print("Each project can either be selected or not selected.")
        print()
        print("Mathematical model:")
        print("Maximize total value")
        print("Subject to total days used ≤ available person-days")
        print("xi ∈ {0,1}")
        print()

        data_df = pd.DataFrame({
            'Project': range(1, len(values) + 1),
            'Estimated Value ($K)': values.astype(int),
            'Days Required': weights.astype(int)
        })

        print("Project data:")
        display(data_df.set_index('Project').T)

        print(f"Available person-days: {int(capacity)}")
        print()

        # Solve LP relaxation
        lp_sol = solve_lp_relaxation(values, weights, capacity)

        # Solve IP
        ip_sol = solve_binary_knapsack(values, weights, capacity)

        # Results table
        result_df = create_result_table(values, weights, lp_sol, ip_sol)

        if show_lp_checkbox.value:
            print("=== LP Relaxation — Node 0 Upper Bound ===")

            if lp_sol is None:
                print("LP relaxation is infeasible.")
            else:
                lp_rows = []

                for i in range(len(values)):
                    lp_rows.append({
                        'Project': i + 1,
                        'x (LP)': round(lp_sol['x'][i], 3),
                        'Integer?': '✓' if abs(lp_sol['x'][i] - round(lp_sol['x'][i])) < 1e-5 else '✗ fractional'
                    })

                display(pd.DataFrame(lp_rows).set_index('Project'))

                print(f"z* = {lp_sol['z']:.2f}  (upper bound)")
                print("If any x is fractional, the LP relaxation is not a valid 0-1 project selection.")
            print()

        print("=== scipy.milp — 0-1 Knapsack Solution ===")

        if ip_sol is None:
            print("Integer problem is infeasible.")
            return

        selected_projects = []
        total_days = 0
        total_value = 0

        for i, x in enumerate(ip_sol['x']):
            if int(round(x)) == 1:
                selected_projects.append(i + 1)
                total_days += int(weights[i])
                total_value += int(values[i])

        display(result_df.set_index('Project'))

        print()
        print(f"Selected projects : {selected_projects}")
        print(f"Total days used   : {total_days} / {int(capacity)}")
        print(f"Total value       : {total_value}")

        if lp_sol is not None:
            gap = lp_sol['z'] - ip_sol['z']
            print(f"LP upper bound    : {lp_sol['z']:.2f}")
            print(f"Integrality gap   : {gap:.2f}")

        if show_chart_checkbox.value:
            print()
            print("=== Selected Projects Chart ===")
            draw_selected_projects_chart(result_df)

        show_book_verification(values, weights, capacity, ip_sol, lp_sol)

        print()
        print("Interpretation:")
        print("The LP relaxation allows fractional projects, so it gives an upper bound.")
        print("The 0-1 integer solution gives the real answer because each project must be either selected or not selected.")


def reset_data(change=None):
    for i in range(n):
        value_inputs[i].value = int(default_values[i])
        weight_inputs[i].value = int(default_weights[i])

    capacity_slider.value = int(default_capacity)
    update()


def on_button_clicked(change=None):
    update()


update_button.on_click(on_button_clicked)
reset_button.on_click(reset_data)

for w in [capacity_slider, show_lp_checkbox, show_chart_checkbox]:
    w.observe(update, names='value')


# -------------------------------------------------------------------
# Display
# -------------------------------------------------------------------

editor_rows = []

for i in range(n):
    editor_rows.append(
        widgets.HBox([
            value_inputs[i],
            weight_inputs[i]
        ])
    )

editor_box = widgets.VBox([
    widgets.HTML("<h4>Edit project values and days</h4>"),
    *editor_rows,
    widgets.HBox([update_button, reset_button])
])

def toggle_editor(change=None):
    if show_custom_editor.value:
        editor_box.layout.display = ''
    else:
        editor_box.layout.display = 'none'

show_custom_editor.observe(toggle_editor, names='value')
editor_box.layout.display = 'none'

display(widgets.VBox([
    widgets.HTML("<h3>Chapter 4.4.3 — Interactive Knapsack Example</h3>"),
    widgets.HTML("This follows the notebook: LP relaxation, 0-1 integer solution, and custom knapsack."),

    capacity_slider,
    show_lp_checkbox,
    show_chart_checkbox,
    show_custom_editor,
    editor_box
]))

display(out)

update()
```
