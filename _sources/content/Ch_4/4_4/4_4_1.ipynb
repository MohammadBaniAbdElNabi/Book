Branch-and-bound algorithms are widely considered to be the most effective methods for solving medium-sized general integer programming problems. These algorithms make no assumptions about the structure of a problem except that the objective function and the constraints must be linear. Even these restrictions can be relaxed without changing the basic framework of the technique.
In its simplest form, branch-and-bound is just an organized way of taking a hard problem and splitting it into two or more smaller (and hence easier) subproblems. If these subproblems are still too hard, we branch again and further subdivide the problems. The process is repeated until each of the subproblems can be easily solved. Branching is done in such a way that solving each of the subproblems (and selecting the best answer found) is equivalent to solving the original problem.
Consider the following simple example in two variables. A manufacturer has 300 personhours available this week and 1,800 units of raw material. These resources can be used to build two products, A and B . The requirements and the profit for each item are given as follows:

| Product | Person-Hours | Raw Material | Profit (\$) |
| :---: | ---: | :---: | :---: |
| A | 150 | 300 | 600 |
| B | 10 | 400 | 100 |

Let $\mathrm{x}_{1}$ and $\mathrm{x}_{2}$ represent the integer number of units of products A and B , respectively. We can formulate this problem as an integer linear programming problem:

$$
\begin{array}{ll}
\operatorname{maximize} & \mathrm{z}=600 \mathrm{x}_{1}+100 \mathrm{x}_{2} \\
\text { subject to } & 150 \mathrm{x}_{1}+10 \mathrm{x}_{2} \leq 300 \\
& 300 \mathrm{x}_{1}+400 \mathrm{x}_{2} \leq 1800 \\
& \mathrm{x}_{1}, \mathrm{x}_{2} \geq 0 \text { and integer }
\end{array}
$$

This problem is illustrated in Figure 4.1. The feasible region is given by the discrete set of integer points within the constraint region. The optimal LP solution occurs at $\mathrm{x}_{1}=1.789$ and $\mathrm{x}_{2}=3.158$ with a profit of $\mathrm{z}=1,389.47$. Unfortunately, we cannot sell a fractional number of items. One obvious alternative is to round down both values to $\mathrm{x}_{1}=1$ and $\mathrm{x}_{2}=3$, for a profit of $\$ 900$. We will call the feasible integer solution $x^{I}=(1,3)$ the current incumbent solution, which is the best answer found thus far. When we find a better integer solution, we will update the current incumbent. Before reading any further, try to locate the optimal integer solution to the problem in Figure 4.1, and consider how integer solutions might be found in general.

The basic branch-and-bound algorithm results from the following observations:

- The feasible integer solution $\mathrm{x}=(1,3)$ with $\mathrm{z}=900$ was fairly easy to find. The optimal integer solution cannot have a lower value of z than $\$ 900$. Thus, we write $\mathrm{z}^{\mathrm{I}}=900$ and call this a lower bound on the optimal solution. Each time we find a higher valued integer solution, we replace the lower bound $\mathrm{z}^{\mathrm{I}}$. This is the bound part of branch-and-bound methods.
- Over the whole feasible region, the largest possible value of $\mathrm{z}=1389.47$, which is the real valued solution obtained from the LP. We call this an upper bound on the optimal integer function value.
- The graphical solution shows that $\mathrm{x}_{2}=3.158$. This is infeasible because it is a fractional solution. Since $\mathrm{x}_{2}$ must be an integer, then clearly either $\mathrm{x}_{2} \leq 3$ or $\mathrm{x}_{2} \geq 4$. This is equivalent to saying that $\mathrm{x}_{2}$ cannot lie part way between 3 and 4 .

Consider the following two subproblems:

$$\begin{aligned}
\text{[A]} \quad & \text{maximize} & z = 600x_1 + 100x_2 \\
& \text{subject to} & 150x_1 + 10x_2 \le 300 \\
& & 300x_1 + 400x_2 \le 1800 \\
& & x_1 \ge 0 \text{ and integer} \\
& & x_2 \ge 4 \text{ and integer} \\
\end{aligned}$$

$$\begin{aligned}
\text{[B]} \quad & \text{maximize} & z = 600x_1 + 100x_2 \\
& \text{subject to} & 150x_1 + 10x_2 \le 300 \\
& & 300x_1 + 400x_2 \le 1800 \\
& & x_1, x_2 \ge 0 \text{ and integer} \\
& & x_2 \le 3
\end{aligned}$$

Observe that if we find the best integer solution of both of these subproblems, then one of them must be the optimal solution to the original problem. These subproblems are represented graphically in Figure 4.2, where the diagram is identical to Figure 4.1 except that the range of values for $\mathrm{x}_{2}$ between 3 and 4 is now infeasible. We say that we have separated on variable $\mathrm{x}_{2}$.

Consider problem [A] first. The LP solution occurs at $\mathrm{x}=(0.667,4)$ with an objective function value of $\mathrm{z}=\$ 800$. Notice that $\mathrm{x}_{2}$ is now integer valued. We will see that each time we separate, the chosen variable will always be integer, although it does not necessarily stay integer on subsequent iterations.

By definition, the linear programming solution is the largest value possible for the problem. Therefore, the value $\mathrm{z}=800$ is an upper bound on all possible solutions in the feasible region for problem $[\mathrm{A}]$. Any integer solution to $[\mathrm{A}]$ must be $\leq 800$. However, we already have a feasible integer solution with $\mathrm{z}^{\mathrm{I}}=900$. Therefore, problem [A] can be ignored as it cannot contain any answer better than 900. In branch-and-bound terminology, we say that problem [A] has been fathomed.

In general, a subproblem is called fathomed whenever it is no longer necessary to branch any further. A subproblem is fathomed when the LP solution is less than the current lower bound for a maximization problem, when the LP solution is infeasible, or when the LP produces an integer solution.

**FIGURE 4.2** <br>
Separate into two subproblems.

Problem [B] has its optimal LP solution at $x=(1.8,3)$ with a function value of $z=1,380$. This value gives us a new upper bound on the optimal integer solution. At each iteration of the branch-and-bound process, the upper and lower bounds can be revised until they eventually converge to the optimal solution. We now know that the optimal value lies between 900 and 1,380 . Variable $x_{2}$ is integer valued, but $x_{1}$ is still fractional. We can now further divide problem [ B ] into two subproblems based on the fact that $\mathrm{x}_{1} \leq 1$ or $\mathrm{x}_{1} \geq 2$ as follows:

$$\begin{aligned}
\text{[B1]} \quad & \text{maximize} & z = 600x_1 + 100x_2 \\
& \text{subject to} & 150x_1 + 10x_2 \le 300 \\
& & 300x_1 + 400x_2 \le 1800 \\
& & x_1, x_2 \ge 0 \text{ and integer} \\
& & x_1 \le 1 \\
& & x_2 \le 3 \\
\end{aligned}$$

$$\begin{aligned}
\text{[B2]} \quad & \text{maximize} & z = 600x_1 + 100x_2 \\
& \text{subject to} & 150x_1 + 10x_2 \le 300 \\
& & 300x_1 + 400x_2 \le 1800 \\
& & x_2 \ge 0 \text{ and integer} \\
& & x_1 \ge 2 \text{ and integer} \\
& & x_2 \le 3
\end{aligned}$$

For problem [B1], it is easy to see that the optimal LP solution occurs at point $\mathrm{x}=(1,3)$ with a function value $\mathrm{z}=900$. Since x is now integer valued, it must be optimal for this subproblem. This subproblem is considered to be fathomed because it gives us an integer solution: there is no need for further branching as the solution cannot get any better below this node. It is also considered fathomed because the solution of 900 is no better than the one we already obtained earlier. In either case, problem [B1] is finished.

Problem [B2] consists of the single point $x=(2,0)$ with a function value of $z=1,200$. This solution is both integer, and better than the previous lower bound. Since x is integer, subproblem [B2] is fathomed and no further branching is required. Our new lower bound increases to $\mathrm{z}^{\mathrm{I}}=1,200$ and $\mathrm{x}^{\mathrm{I}}=(2,0)$ becomes the new current incumbent.

At this point, we observe that all of our subproblems have been fathomed. Therefore, $x^{\mathrm{I}}=(2,0)$ is the optimal integer solution, and $\mathrm{z}^{\mathrm{I}}=\$ 1,200$ is the optimal function value.

It is often convenient to display this procedure in the form of a branch-and-bound tree. The tree corresponding to the previous example is illustrated in Figure 4.3. Each subproblem is represented by a node in the tree. Each node must either be fathomed or split into subproblems, which are shown by lower level nodes.

In Figure 4.3a, node 0 represents the original problem. We construct nodes 1 and 2 (for subproblems [A] and [B], respectively) by constraining $\mathrm{x}_{2}$ in Figure 4.3b. Node 1 is fathomed and node 2 is further subdivided into nodes 3 and 4 in Figure 4.3c, corresponding to problems [B1] and [B2].

**Code Implementation**

```python
# ── Chapter 4.4.1 — Interactive Branch-and-Bound Example ─────────────
# Same idea as your notebook:
# 1) LP relaxation
# 2) scipy.milp integer solution
# 3) Manual branch-and-bound steps
# 4) Original notebook-style branch-and-bound tree
# 5) Feasible-region graph added back

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import ipywidgets as widgets
from IPython.display import display, clear_output
from scipy.optimize import linprog, milp, LinearConstraint, Bounds

# -------------------------------------------------------------------
# Widgets
# -------------------------------------------------------------------

style = {'description_width': '190px'}
layout = widgets.Layout(width='520px')

profit_A_slider = widgets.IntSlider(
    value=600,
    min=50,
    max=1500,
    step=50,
    description='Profit A:',
    continuous_update=False,
    style=style,
    layout=layout
)

profit_B_slider = widgets.IntSlider(
    value=100,
    min=50,
    max=1500,
    step=50,
    description='Profit B:',
    continuous_update=False,
    style=style,
    layout=layout
)

hours_slider = widgets.IntSlider(
    value=300,
    min=100,
    max=1000,
    step=50,
    description='Available hours:',
    continuous_update=False,
    style=style,
    layout=layout
)

material_slider = widgets.IntSlider(
    value=1800,
    min=500,
    max=5000,
    step=100,
    description='Available material:',
    continuous_update=False,
    style=style,
    layout=layout
)

show_graph = widgets.Checkbox(
    value=True,
    description='Show feasible-region graph'
)

show_tree = widgets.Checkbox(
    value=True,
    description='Show branch-and-bound tree'
)

show_feasible_points_table = widgets.Checkbox(
    value=True,
    description='Show feasible integer points table'
)

out = widgets.Output()

# -------------------------------------------------------------------
# Solver functions
# -------------------------------------------------------------------

def solve_lp(c, A, b, extra_bounds=None):
    """
    Solve the LP relaxation with optional extra bounds.
    extra_bounds example:
        [(1, 'lb', 4)] means x2 >= 4
        [(1, 'ub', 3)] means x2 <= 3
    """

    lb = [0.0, 0.0]
    ub = [np.inf, np.inf]

    if extra_bounds:
        for idx, kind, val in extra_bounds:
            if kind == 'ub':
                ub[idx] = val
            else:
                lb[idx] = val

    res = linprog(
        -c,
        A_ub=A,
        b_ub=b,
        bounds=list(zip(lb, ub)),
        method='highs'
    )

    if res.status != 0:
        return None, None

    return -res.fun, res.x


def solve_integer_problem(c, A, b):
    """
    Solve the integer programming problem directly using scipy.milp.
    """

    constraints = LinearConstraint(A, -np.inf, b)
    integrality = np.array([1, 1])
    bounds = Bounds(lb=[0, 0], ub=[np.inf, np.inf])

    res = milp(
        -c,
        constraints=constraints,
        integrality=integrality,
        bounds=bounds
    )

    if not res.success:
        return None, None

    return -res.fun, res.x


def enumerate_integer_points(c, A, b):
    """
    Enumerate all feasible integer points for the 2-variable example.
    """

    max_x1 = int(max(b[0] / A[0, 0], b[1] / A[1, 0])) + 5
    max_x2 = int(max(b[0] / A[0, 1], b[1] / A[1, 1])) + 5

    points = []

    for x1 in range(max_x1 + 1):
        for x2 in range(max_x2 + 1):
            x = np.array([x1, x2])
            if np.all(A @ x <= b + 1e-9):
                points.append({
                    'x1 Product A': x1,
                    'x2 Product B': x2,
                    'Profit z': c @ x,
                    'Hours Used': A[0] @ x,
                    'Material Used': A[1] @ x
                })

    return pd.DataFrame(points)


def is_integer_solution(x):
    if x is None:
        return False
    return np.all(np.abs(x - np.round(x)) < 1e-6)


def fmt_x(x):
    if x is None:
        return "Infeasible"
    return f"({x[0]:.3f}, {x[1]:.3f})"


def fmt_z(z):
    if z is None:
        return "Infeasible"
    return f"{z:.2f}"


# -------------------------------------------------------------------
# Feasible-region graph
# -------------------------------------------------------------------

def draw_feasible_region_graph(c, A, b, z_lp, x_lp, z_ip, x_ip, feasible_df):
    """
    Draw graphical feasible region for the LP relaxation and integer points.
    """

    print()
    print("=== Feasible-Region Graph ===")

    # Plot limits
    x1_limit_1 = b[0] / A[0, 0]
    x1_limit_2 = b[1] / A[1, 0]
    x1_max = max(4, int(max(x1_limit_1, x1_limit_2)) + 4)

    x2_limit_1 = b[0] / A[0, 1]
    x2_limit_2 = b[1] / A[1, 1]
    x2_max = max(8, int(max(x2_limit_1, x2_limit_2)) + 4)

    x1 = np.linspace(0, x1_max, 500)

    # Constraint lines
    y_hours = (b[0] - A[0, 0] * x1) / A[0, 1]
    y_material = (b[1] - A[1, 0] * x1) / A[1, 1]

    y_hours = np.maximum(y_hours, 0)
    y_material = np.maximum(y_material, 0)

    feasible_upper = np.minimum(y_hours, y_material)

    plt.figure(figsize=(10, 7))

    # Feasible region
    plt.fill_between(
        x1,
        0,
        feasible_upper,
        where=feasible_upper >= 0,
        alpha=0.25,
        label='LP feasible region'
    )

    # Constraints
    plt.plot(x1, y_hours, linewidth=2, label='150x1 + 10x2 ≤ hours')
    plt.plot(x1, y_material, linewidth=2, label='300x1 + 400x2 ≤ material')

    # Integer points
    if not feasible_df.empty:
        plt.scatter(
            feasible_df['x1 Product A'],
            feasible_df['x2 Product B'],
            s=70,
            alpha=0.75,
            label='Feasible integer points'
        )

    # LP relaxation point
    if x_lp is not None:
        plt.scatter(
            x_lp[0],
            x_lp[1],
            s=180,
            marker='X',
            label=f'LP relaxation: z*={z_lp:.2f}'
        )

        plt.text(
            x_lp[0] + 0.08,
            x_lp[1] + 0.15,
            f"LP ({x_lp[0]:.2f}, {x_lp[1]:.2f})",
            fontsize=9
        )

    # Integer optimum point
    if x_ip is not None:
        plt.scatter(
            x_ip[0],
            x_ip[1],
            s=220,
            marker='*',
            label=f'Integer optimum: z={z_ip:.0f}'
        )

        plt.text(
            x_ip[0] + 0.08,
            x_ip[1] + 0.15,
            f"IP ({int(round(x_ip[0]))}, {int(round(x_ip[1]))})",
            fontsize=9
        )

    # Objective lines for LP and integer solution
    if x_lp is not None and c[1] != 0:
        y_obj_lp = (z_lp - c[0] * x1) / c[1]
        y_obj_lp = np.where(y_obj_lp >= 0, y_obj_lp, np.nan)

        plt.plot(
            x1,
            y_obj_lp,
            linestyle='--',
            linewidth=1.5,
            label=f'LP objective line z={z_lp:.0f}'
        )

    if x_ip is not None and c[1] != 0:
        y_obj_ip = (z_ip - c[0] * x1) / c[1]
        y_obj_ip = np.where(y_obj_ip >= 0, y_obj_ip, np.nan)

        plt.plot(
            x1,
            y_obj_ip,
            linestyle=':',
            linewidth=2,
            label=f'IP objective line z={z_ip:.0f}'
        )

    plt.title('Graphical View of LP Relaxation and Integer Solution', fontsize=13, pad=18)
    plt.xlabel('x1 = units of Product A')
    plt.ylabel('x2 = units of Product B')
    plt.xlim(0, x1_max)
    plt.ylim(0, x2_max)
    plt.grid(True, alpha=0.3)
    plt.legend(fontsize=8)
    plt.tight_layout()
    plt.show()


# -------------------------------------------------------------------
# Original notebook-style branch-and-bound tree
# -------------------------------------------------------------------

def draw_branch_and_bound_tree(c, A, b, ip_z):
    """
    Draws the same style as the original notebook tree.

    Default textbook branch:
    Root → branch on x2
    Node A: x2 >= ceil(root x2)
    Node B: x2 <= floor(root x2)
    Then Node B → branch on x1
    """

    z0, x0 = solve_lp(c, A, b)

    if x0 is None:
        print("Cannot draw tree because the root LP relaxation is infeasible.")
        return

    branch_x2_value = x0[1]
    x2_floor = int(np.floor(branch_x2_value))
    x2_ceil = int(np.ceil(branch_x2_value))

    zA, xA = solve_lp(c, A, b, [(1, 'lb', x2_ceil)])
    zB, xB = solve_lp(c, A, b, [(1, 'ub', x2_floor)])

    if xB is not None:
        branch_x1_value = xB[0]
        x1_floor = int(np.floor(branch_x1_value))
        x1_ceil = int(np.ceil(branch_x1_value))

        zB1, xB1 = solve_lp(c, A, b, [(1, 'ub', x2_floor), (0, 'ub', x1_floor)])
        zB2, xB2 = solve_lp(c, A, b, [(1, 'ub', x2_floor), (0, 'lb', x1_ceil)])
    else:
        x1_floor = None
        x1_ceil = None
        zB1, xB1 = None, None
        zB2, xB2 = None, None

    fig, ax = plt.subplots(figsize=(11, 6))
    fig.patch.set_facecolor('white')
    ax.set_facecolor('white')
    ax.axis('off')

    def draw_node(ax, x, y, lines, color='#2980b9', fathomed=False, incumbent=False):
        w, h = 2.8, 1.12

        ec = '#e74c3c' if incumbent else ('#aaaaaa' if fathomed else color)
        lw = 2.7 if incumbent else 1.5

        rect = mpatches.FancyBboxPatch(
            (x - w / 2, y - h / 2),
            w,
            h,
            boxstyle='round,pad=0.08',
            facecolor='white',
            edgecolor=ec,
            linewidth=lw,
            zorder=3
        )

        ax.add_patch(rect)

        for i, line in enumerate(lines):
            fs = 9.0 if i > 0 else 9.6
            fw = 'bold' if i == 0 else 'normal'
            col = '#2c3e50' if not fathomed else '#888888'

            ax.text(
                x,
                y + 0.34 - i * 0.27,
                line,
                ha='center',
                va='center',
                fontsize=fs,
                fontweight=fw,
                color=col,
                zorder=4
            )

    def arrow(ax, x1, y1, x2, y2, label=''):
        ax.annotate(
            '',
            xy=(x2, y2 + 0.58),
            xytext=(x1, y1 - 0.58),
            arrowprops=dict(
                arrowstyle='->',
                color='#555555',
                lw=1.2
            )
        )

        mx, my = (x1 + x2) / 2, (y1 + y2) / 2

        if label:
            ax.text(
                mx + 0.10,
                my,
                label,
                fontsize=9,
                color='#555555',
                ha='left',
                va='center',
                bbox=dict(facecolor='white', edgecolor='none', alpha=0.85, pad=1)
            )

    # Node A status
    if zA is None:
        nodeA_status = "Infeasible ← fathomed"
        nodeA_fathomed = True
    elif ip_z is not None and zA <= ip_z + 1e-9:
        nodeA_status = f"z*={zA:.0f} ← fathomed"
        nodeA_fathomed = True
    else:
        nodeA_status = f"z*={zA:.2f}"
        nodeA_fathomed = False

    # Node B status
    nodeB_status = "Infeasible"
    if zB is not None:
        nodeB_status = f"z*={zB:.2f}"

    # Node B1 status
    if zB1 is None:
        nodeB1_status = "Infeasible ← fathomed"
        nodeB1_fathomed = True
    elif is_integer_solution(xB1):
        nodeB1_status = f"z={zB1:.0f} ← fathomed"
        nodeB1_fathomed = True
    else:
        nodeB1_status = f"z*={zB1:.2f}"
        nodeB1_fathomed = False

    # Node B2 status
    if zB2 is None:
        nodeB2_status = "Infeasible"
        nodeB2_incumbent = False
        nodeB2_fathomed = True
    elif ip_z is not None and abs(zB2 - ip_z) < 1e-6 and is_integer_solution(xB2):
        nodeB2_status = f"z={zB2:.0f} ← OPTIMAL"
        nodeB2_incumbent = True
        nodeB2_fathomed = False
    elif is_integer_solution(xB2):
        nodeB2_status = f"z={zB2:.0f} ← integer"
        nodeB2_incumbent = False
        nodeB2_fathomed = True
    else:
        nodeB2_status = f"z*={zB2:.2f}"
        nodeB2_incumbent = False
        nodeB2_fathomed = False

    draw_node(
        ax,
        0,
        4,
        [
            'Node 0 (LP relaxation)',
            f'x={fmt_x(x0)}',
            f'z*={fmt_z(z0)}'
        ]
    )

    draw_node(
        ax,
        -3.5,
        2,
        [
            f'Node A  (x2 ≥ {x2_ceil})',
            f'x={fmt_x(xA)}',
            nodeA_status
        ],
        fathomed=nodeA_fathomed
    )

    draw_node(
        ax,
        3.5,
        2,
        [
            f'Node B  (x2 ≤ {x2_floor})',
            f'x={fmt_x(xB)}',
            nodeB_status
        ]
    )

    if xB is not None:
        draw_node(
            ax,
            1.5,
            0,
            [
                f'Node B1  (x1 ≤ {x1_floor})',
                f'x={fmt_x(xB1)}',
                nodeB1_status
            ],
            fathomed=nodeB1_fathomed
        )

        draw_node(
            ax,
            5.5,
            0,
            [
                f'Node B2  (x1 ≥ {x1_ceil})',
                f'x={fmt_x(xB2)}',
                nodeB2_status
            ],
            fathomed=nodeB2_fathomed,
            incumbent=nodeB2_incumbent
        )

    arrow(ax, 0, 4, -3.5, 2, f'x2 ≥ {x2_ceil}')
    arrow(ax, 0, 4, 3.5, 2, f'x2 ≤ {x2_floor}')

    if xB is not None:
        arrow(ax, 3.5, 2, 1.5, 0, f'x1 ≤ {x1_floor}')
        arrow(ax, 3.5, 2, 5.5, 0, f'x1 ≥ {x1_ceil}')

    ax.set_xlim(-6, 8)
    ax.set_ylim(-1, 5.7)

    ax.set_title(
        'Figure 4.3 — Branch-and-Bound Tree',
        fontsize=12,
        pad=16
    )

    plt.tight_layout()
    plt.show()


# -------------------------------------------------------------------
# Manual B&B text summary
# -------------------------------------------------------------------

def show_manual_branch_and_bound(c, A, b, ip_z):
    print("=== Manual Branch-and-Bound ===")
    print()

    z0, x0 = solve_lp(c, A, b)

    if x0 is None:
        print("Node 0 is infeasible.")
        return

    print("Node 0 (LP relaxation):")
    print(f"  x = ({x0[0]:.4f}, {x0[1]:.4f}),  z* = {z0:.4f}  (upper bound)")

    x2_floor = int(np.floor(x0[1]))
    x2_ceil = int(np.ceil(x0[1]))

    print(f"  x2 = {x0[1]:.3f} is fractional → branch on x2")
    print()

    zA, xA = solve_lp(c, A, b, [(1, 'lb', x2_ceil)])
    print(f"Node A  (x2 >= {x2_ceil}):")

    if zA is None:
        print("  Infeasible → fathomed")
    else:
        print(f"  x = ({xA[0]:.4f}, {xA[1]:.4f}),  z* = {zA:.4f}")

        if ip_z is not None and zA <= ip_z + 1e-9:
            print(f"  z* = {zA:.1f} ≤ current integer solution ({ip_z:.0f}) → fathomed")
        else:
            print("  Still promising")
    print()

    zB, xB = solve_lp(c, A, b, [(1, 'ub', x2_floor)])
    print(f"Node B  (x2 <= {x2_floor}):")

    if zB is None:
        print("  Infeasible → fathomed")
        return

    print(f"  x = ({xB[0]:.4f}, {xB[1]:.4f}),  z* = {zB:.4f}")

    x1_floor = int(np.floor(xB[0]))
    x1_ceil = int(np.ceil(xB[0]))

    print(f"  x1 = {xB[0]:.3f} is fractional → branch on x1")
    print()

    zB1, xB1 = solve_lp(c, A, b, [(1, 'ub', x2_floor), (0, 'ub', x1_floor)])
    print(f"Node B1 (x2 <= {x2_floor}, x1 <= {x1_floor}):")

    if zB1 is None:
        print("  Infeasible → fathomed")
    else:
        print(f"  x = ({xB1[0]:.4f}, {xB1[1]:.4f}),  z* = {zB1:.4f}")

        if is_integer_solution(xB1):
            print(f"  Integer solution → fathomed  (z={int(round(zB1))})")
        else:
            print("  Fractional solution")
    print()

    zB2, xB2 = solve_lp(c, A, b, [(1, 'ub', x2_floor), (0, 'lb', x1_ceil)])
    print(f"Node B2 (x2 <= {x2_floor}, x1 >= {x1_ceil}):")

    if zB2 is None:
        print("  Infeasible → fathomed")
    else:
        print(f"  x = ({xB2[0]:.4f}, {xB2[1]:.4f}),  z* = {zB2:.4f}")

        if is_integer_solution(xB2):
            print(f"  Integer solution → possible incumbent  z={int(round(zB2))}")
        else:
            print("  Fractional solution")
    print()


# -------------------------------------------------------------------
# Main interactive function
# -------------------------------------------------------------------

def update(change=None):
    with out:
        clear_output(wait=True)

        c = np.array([
            float(profit_A_slider.value),
            float(profit_B_slider.value)
        ])

        A = np.array([
            [150.0, 10.0],
            [300.0, 400.0]
        ])

        b = np.array([
            float(hours_slider.value),
            float(material_slider.value)
        ])

        print("=== Chapter 4.4.1 — Interactive Branch-and-Bound Simple Example ===")
        print()
        print("Problem:")
        print(f"Maximize z = {c[0]:.0f}x1 + {c[1]:.0f}x2")
        print("Subject to:")
        print(f"  150x1 + 10x2  ≤ {b[0]:.0f}     person-hours")
        print(f"  300x1 + 400x2 ≤ {b[1]:.0f}     raw material")
        print("  x1, x2 ≥ 0 and integer")
        print()

        data_df = pd.DataFrame({
            'Product': ['A', 'B'],
            'Person-Hours': [150, 10],
            'Raw Material': [300, 400],
            'Profit': c
        })

        display(data_df)

        z_lp, x_lp = solve_lp(c, A, b)
        z_ip, x_ip = solve_integer_problem(c, A, b)
        feasible_df = enumerate_integer_points(c, A, b)

        print()
        print("=== LP Relaxation ===")

        if x_lp is None:
            print("LP relaxation is infeasible.")
        else:
            print(f"  x1 = {x_lp[0]:.4f}")
            print(f"  x2 = {x_lp[1]:.4f}")
            print(f"  z* = {z_lp:.4f}")
            print("  This is an upper bound for the integer problem.")

        print()
        print("=== scipy.milp — Integer Solution ===")

        if x_ip is None:
            print("Integer problem is infeasible.")
        else:
            print(f"  x1 = {int(round(x_ip[0]))}")
            print(f"  x2 = {int(round(x_ip[1]))}")
            print(f"  z  = {int(round(z_ip))}")

        if show_graph.value:
            draw_feasible_region_graph(c, A, b, z_lp, x_lp, z_ip, x_ip, feasible_df)

        print()
        show_manual_branch_and_bound(c, A, b, z_ip)

        if show_tree.value:
            print()
            print("=== Branch-and-Bound Tree ===")
            draw_branch_and_bound_tree(c, A, b, z_ip)

        if show_feasible_points_table.value:
            print()
            print("=== Feasible Integer Points ===")

            if feasible_df.empty:
                print("No feasible integer points.")
            else:
                display(
                    feasible_df
                    .sort_values('Profit z', ascending=False)
                    .reset_index(drop=True)
                )

        print()
        print("Interpretation:")

        if x_lp is not None and x_ip is not None:
            gap = z_lp - z_ip

            print(f"The LP relaxation gives z* = {z_lp:.2f}.")
            print(f"The best integer solution gives z = {z_ip:.2f}.")
            print(f"The integrality gap is {gap:.2f}.")

            if abs(gap) < 1e-6:
                print("The LP solution is already integer, so branch-and-bound is not needed.")
            else:
                print("Because the LP solution is fractional, branch-and-bound is used to search for the best integer solution.")

        if (
            profit_A_slider.value == 600 and
            profit_B_slider.value == 100 and
            hours_slider.value == 300 and
            material_slider.value == 1800
        ):
            print()
            print("=== Textbook Default Verification ===")
            print("For the default example:")
            print("  LP relaxation: x1 = 1.789, x2 = 3.158, z* = 1389.47")
            print("  Integer optimum: x1 = 2, x2 = 0, z = 1200")


# -------------------------------------------------------------------
# Connect widgets
# -------------------------------------------------------------------

all_widgets = [
    profit_A_slider,
    profit_B_slider,
    hours_slider,
    material_slider,
    show_graph,
    show_tree,
    show_feasible_points_table
]

for w in all_widgets:
    w.observe(update, names='value')


# -------------------------------------------------------------------
# Display
# -------------------------------------------------------------------

display(widgets.VBox([
    widgets.HTML("<h3>Chapter 4.4.1 — Interactive Branch-and-Bound Example</h3>"),
    widgets.HTML("This keeps the same notebook example, includes the feasible-region graph, and keeps the branch-and-bound tree."),

    profit_A_slider,
    profit_B_slider,
    hours_slider,
    material_slider,
    show_graph,
    show_tree,
    show_feasible_points_table
]))

display(out)

update()
```
