The primary types of utility curves are illustrated in Figure 9.13. For the EMVer, each additional dollar has the same utility value. The utility curve is a straight line indicating that the marginal value of each dollar is constant. The risk averse (RA) decision-maker has a curve with a decreasing slope, indicating a decreasing value of each additional dollar. Observe that this person derives $60 \%$ of the total utility from simply breaking even. Conversely, the risk seeking (RS) gambler has a curve with an increasing rate of change. The marginal value of each additional dollar is increasing. This individual is happy only when he is very close to the top of the scale. Breaking even has a very low utility. It is important to recognize that a person's attitude toward risk is reflected by the rate of change of the slope of the curve-not by the absolute value of the slope.

A set of experiments was performed by Kahneman and Tversky (1979) in which subjects were asked to develop a utility function in the standard way. It was discovered that the typical curve of the majority of cases resembled the form shown in Figure 9.15. Most respondents exhibited RA behavior above zero, but they became gamblers when the values were negative. Apparently, people have a financial target of breaking even and are willing to take considerable risks to avoid a loss.

In decision analysis, this behavior is called The Zero Illusion. The problem is that the zero point is relative to the scale that has been defined in the choice of units. For example, if we use net profit as the outcome measure, zero represents the point at which the company makes or loses money on the product. If we use net assets to measure the effect of a marketing decision, we have not changed the problem, but zero now represents the point at which the company is in the black or in the red. Profit could be described in terms of this year's profit-and-loss statement. In each case, the method used to calculate outcome values has no effect on the real decision problem, but the scale has simply been shifted.

When a decision-maker produces a curve with this structure, he can usually be convinced to abandon the zero target by reformulating the questions in terms of a few different scales. He will soon adjust his answers to a more consistent behavior. Zero is indeed an imaginary target. We will come back to this notion when we discuss the framing effect in Section 9.5.5.

A classic example of apparently irrational behavior with respect to utility theory is known as the Allais Paradox. Consider the two decision problems shown in Figure 9.16. For the problem in Figure 9.16a, most people prefer to choose the certain outcome of $\$ 500,000$ because the lottery branch looks too risky. Although there is a $10 \%$ chance of getting $\$ 1,000,000$, there is also a $1 \%$ chance of getting nothing, so why take the risk?

In Figure 9.16b, the two lotteries look very similar, except that the second one has a payoff of $\$ 1,000,000$ while the first gives only $\$ 500,000$. In this case, the majority of people will

Consider an individual who plays the aforementioned two lotteries as described for the majority. Not everyone will agree with this person, but the selections certainly appear to be logical. The paradox arises when we attempt to develop a utility function for this person: it cannot be done.
Without loss of generality, we can assume a utility value of $\mathrm{u}(\$ 0)=0, \mathrm{u}(\$ 1,000,000)=1$, and $\mathrm{u}(\$ 500,000)=\mathrm{p}$ for some value of p . We wish to find a value of p such that the utilities of the two decision problems correspond with the choices of our decision-maker.
In the first problem, the utility of $\$ 500,000$ certain is simply p . The utility of the lottery is

$$
0.89 \times p+0.10
$$

Because our subject chose the certain branch, we conclude that

$$
p>0.89 \times p+0.10
$$

or

$$
0.11 \times \mathrm{p}>0.10
$$

Similarly, for the second problem, the utility of the first branch is given by $0.11 \times \mathrm{p}$, while the utility of the second branch is 0.10 . Because the second branch is preferred, we conclude that

$$
0.11 \times \mathrm{p}<0.10
$$

In other words, no matter how the utility function is defined, the decision-maker must either select the first branch in both problems, or the second branch in both problems.
There is no rational way to explain this dilemma. In the first problem, the decision-maker is frightened away from the lottery because there is a 0.01 chance of obtaining a zero. However, in the second problem, there is an extra 0.01 chance of getting zero in the second lottery. This additional risk apparently is not recognized. The decision-maker is effectively saying that a probability of 0.01 is significant enough to avoid the risk, but, at the same time, probabilities of 0.89 or 0.90 are essentially equivalent. A difference in probability of 0.01 should either deter a person in both cases or not at all. Even when we point this out, many people will stick with their original, irrational selections. We will discuss this and other aspects of irrational behavior in the following section.

Many practitioners believe that utility theory is the only solution to decision-making problems. Others argue that, although it is reasonable to assume that people have an implicit personal utility function, it is not a practical tool for decision analysis. The main objection, in addition to the problems already mentioned, is that the assessment procedures for determining a person's utility function are basically artificial. When people are asked to play lottery games, the prizes are not representative of real decision problems. Therefore, it is difficult for people to treat the answers seriously, especially as the number of questions increases. Despite these criticisms, utility theory has been used in a wide variety of practical decision situations, some of which are described and discussed in Raiffa (1968) and Keeney and Raiffa (1976).

**Code Implementation**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

style = {'description_width': '180px'}
layout = widgets.Layout(width='500px')

# ─────────────────────────────────────────────────────────────
# Helper formatting
# ─────────────────────────────────────────────────────────────

def money_k(x):
    sign = "-" if x < 0 else ""
    return f"{sign}${abs(x):,.0f}K"


# ─────────────────────────────────────────────────────────────
# PART 1 — Basic Utility Curves
# ─────────────────────────────────────────────────────────────

risk_type = widgets.Dropdown(
    options=[
        ('Risk Averse', 'averse'),
        ('EMVer / Risk Neutral', 'neutral'),
        ('Risk Seeking', 'seeking'),
        ('Show All', 'all')
    ],
    value='all',
    description='Curve Type:',
    style=style,
    layout=layout
)

curvature = widgets.FloatSlider(
    value=2.0,
    min=0.3,
    max=5.0,
    step=0.1,
    description='Curvature:',
    style=style,
    layout=layout
)

out_basic = widgets.Output()


def draw_basic_curves(curve_type, c):
    with out_basic:
        out_basic.clear_output(wait=True)

        x = np.linspace(-20, 40, 400)
        x_norm = (x - x.min()) / (x.max() - x.min())

        # Risk averse: increasing but decreasing slope
        y_averse = x_norm ** (1 / c)

        # Risk neutral / EMVer: straight line
        y_neutral = x_norm

        # Risk seeking: increasing slope
        y_seeking = x_norm ** c

        fig, ax = plt.subplots(figsize=(8, 6))

        if curve_type in ['averse', 'all']:
            ax.plot(x, y_averse, label='Risk Averse')

        if curve_type in ['neutral', 'all']:
            ax.plot(x, y_neutral, label='EMVer / Risk Neutral')

        if curve_type in ['seeking', 'all']:
            ax.plot(x, y_seeking, label='Risk Seeking')

        ax.set_title('Basic Utility Curves')
        ax.set_xlabel('Profit in thousand dollars')
        ax.set_ylabel('Utility')
        ax.set_ylim(0, 1.05)
        ax.grid(True, alpha=0.3)
        ax.legend()

        plt.show()

        if curve_type == 'averse':
            display(Markdown("""
**Risk averse:** The curve rises quickly at first, then flattens.
This means the person values safety and dislikes large risk.
"""))
        elif curve_type == 'neutral':
            display(Markdown("""
**EMVer / risk neutral:** The curve is straight.
Each additional dollar has the same value.
"""))
        elif curve_type == 'seeking':
            display(Markdown("""
**Risk seeking:** The curve rises slowly at first, then faster.
This person is willing to take risk for a large possible gain.
"""))
        else:
            display(Markdown("""
This compares the three common utility curve shapes:

- **Risk averse:** decreasing slope
- **EMVer / risk neutral:** constant slope
- **Risk seeking:** increasing slope
"""))


def update_basic(change):
    draw_basic_curves(risk_type.value, curvature.value)


risk_type.observe(update_basic, names='value')
curvature.observe(update_basic, names='value')

display(Markdown("## Part 1 — Basic Utility Curves"))
display(widgets.VBox([risk_type, curvature]), out_basic)
draw_basic_curves(risk_type.value, curvature.value)
```

```python
# ─────────────────────────────────────────────────────────────
# PART 2 — Target-Level Utility Curve
# ─────────────────────────────────────────────────────────────

target = widgets.FloatSlider(
    value=25,
    min=-10,
    max=40,
    step=1,
    description='Target Level ($K):',
    style=style,
    layout=layout
)

steepness = widgets.FloatSlider(
    value=0.13,
    min=0.03,
    max=0.40,
    step=0.01,
    description='Goal Sensitivity:',
    style=style,
    layout=layout
)

out_target = widgets.Output()


def draw_target_curve(target_value, k):
    with out_target:
        out_target.clear_output(wait=True)

        x = np.linspace(-20, 40, 500)

        # S-shaped logistic curve centered around target
        y = 1 / (1 + np.exp(-k * (x - target_value)))
        y = (y - y.min()) / (y.max() - y.min())

        fig, ax = plt.subplots(figsize=(8, 6))
        ax.plot(x, y, linewidth=2)

        ax.axvline(target_value, linestyle='--', alpha=0.7)
        ax.scatter([target_value], [np.interp(target_value, x, y)], s=80)

        ax.text(
            target_value + 1,
            np.interp(target_value, x, y),
            f"Target = {money_k(target_value)}",
            fontsize=10
        )

        ax.set_title('Utility Curve with a Target Level')
        ax.set_xlabel('Profit in thousand dollars')
        ax.set_ylabel('Utility')
        ax.set_ylim(0, 1.05)
        ax.grid(True, alpha=0.3)

        plt.show()

        display(Markdown(f"""
This curve shows a person who has a **target level** of **{money_k(target_value)}**.

Below the target, the person may become more willing to take risk to reach the goal.
Once the target is reached, the person becomes more conservative again.

This explains why someone can be:

- **Risk averse** far below the target
- **Risk seeking** near the target
- **Risk averse** again after reaching the target
"""))


def update_target(change):
    draw_target_curve(target.value, steepness.value)


target.observe(update_target, names='value')
steepness.observe(update_target, names='value')

display(Markdown("## Part 2 — Utility Curve with a Target Level"))
display(widgets.VBox([target, steepness]), out_target)
draw_target_curve(target.value, steepness.value)
```

```python
# ─────────────────────────────────────────────────────────────
# PART 3 — Zero-Illusion Utility Curve
# ─────────────────────────────────────────────────────────────

zero_point = widgets.FloatSlider(
    value=0,
    min=-100,
    max=100,
    step=5,
    description='Reference Zero ($K):',
    style=style,
    layout=layout
)

loss_aversion = widgets.FloatSlider(
    value=2.0,
    min=0.5,
    max=5.0,
    step=0.1,
    description='Loss Aversion:',
    style=style,
    layout=layout
)

out_zero = widgets.Output()


def draw_zero_illusion(zero, loss_factor):
    with out_zero:
        out_zero.clear_output(wait=True)

        x = np.linspace(-100, 400, 600)
        relative = x - zero

        # Prospect-theory-like shape
        y = np.where(
            relative >= 0,
            np.sqrt(relative + 1),
            -loss_factor * np.sqrt(np.abs(relative) + 1)
        )

        # Normalize to 0–1 for display
        y = (y - y.min()) / (y.max() - y.min())

        fig, ax = plt.subplots(figsize=(8, 6))
        ax.plot(x, y, linewidth=2)

        ax.axvline(zero, linestyle='--', alpha=0.7)
        ax.text(zero + 5, 0.08, f"Reference zero = {money_k(zero)}", fontsize=10)

        ax.set_title('Zero-Illusion Utility Curve')
        ax.set_xlabel('Profit in thousand dollars')
        ax.set_ylabel('Utility')
        ax.set_ylim(0, 1.05)
        ax.grid(True, alpha=0.3)

        plt.show()

        display(Markdown(f"""
This curve shows the **zero illusion** idea.

The important point is that people react differently to gains and losses depending on what they treat as **zero**.

Current reference zero: **{money_k(zero)}**

A person may act:

- **Risk seeking below zero**, because they want to escape a loss
- **Risk averse above zero**, because they want to protect gains

Changing the reference zero changes how the same outcome feels.
"""))


def update_zero(change):
    draw_zero_illusion(zero_point.value, loss_aversion.value)


zero_point.observe(update_zero, names='value')
loss_aversion.observe(update_zero, names='value')

display(Markdown("## Part 3 — Zero-Illusion Utility Curve"))
display(widgets.VBox([zero_point, loss_aversion]), out_zero)
draw_zero_illusion(zero_point.value, loss_aversion.value)
```
